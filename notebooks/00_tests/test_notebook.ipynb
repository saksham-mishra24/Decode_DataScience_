### To test hypotheses, statisticians created statistical tests. The first and most fundamental one is the Z-Test.

#### A Z-test helps us check whether a sample mean is significantly different from a population mean.

Suppose a company produces light bulbs.

The company claims:

Average lifespan = 1000 hours

This is the population mean.

Now a quality inspector tests 50 bulbs and finds:

Sample average = 960 hours

Riya asks:

Riya: Sir, is the company lying?

Professor Arjun laughs.

Professor Arjun: Not necessarily. The difference might just be due to random variation.

So we perform a Z-test.

1. Sample size is large (n ≥ 30)
2. Population standard deviation is known
3. Data is approximately normally distributed

```python
from statsmodels.stats.weightstats import ztest

# statsmodels is a module that provides classes and functions for the estimation of many different statistical models, as well as for conducting statistical tests and statistical data exploration. 
# The ztest function is used to perform a z-test for the mean of a 
# single sample or for the difference between two samples.

# weightstats is a submodule of statsmodels that provides classes and 
# functions for performing weighted statistical analyses, 
# including weighted means, variances, and tests.
```

```python
# Example sample data (lifespan of bulbs)
data = [960, 950, 970, 940, 965, 955, 945, 960, 950, 955]
```

```python
z_stats, p_value = ztest(data, value = 1000)
```

```python
print(f"Z-statistic: {z_stats}, P-value: {p_value}")
```
