# 🧠 Simple Meaning of Gini Impurity

Gini Impurity also measures:

```
How mixed the classes are
```

It answers the question:

> **If we randomly pick a data point, how likely is it to be wrongly classified?**

---

# 🎯 Example Dataset

| Student | Result |
| ------- | ------ |

| B       | Pass   |
| C       | Fail   |

This dataset is **mixed**, so impurity is **high**.

# 🧠 Gini Formula (Simple Version)

```
Gini = 1 − (p₁² + p₂²)
```

Where:

| Symbol | Meaning             |
| ------ | ------------------- |
| p₁     | Probability of Pass |
| p₂     | Probability of Fail |

---


So impurity is **0.5 (high)**.

# Now we calculate **Gini Impurity using Python**.


````python
# Import numpy for numerical calculations
import numpy as np

# 1 means Pass
# 0 means Fail

results = np.array([1,1,0,0])
````

# Step 3 — Calculate Probabilities


````python

# Count number of Pass results
num_pass = np.sum(results == 1)

# Count number of Fail results
num_fail = np.sum(results == 0)

# Total number of samples
total = len(results)

# Calculate probabilities
p_pass = num_pass / total
p_fail = num_fail / total
````


```
p_pass → probability of pass
p_fail → probability of fail
```

# Step 4 — Calculate Gini


````python
# Apply the Gini formula

gini = 1 - (p_pass**2 + p_fail**2)

# Print the result
print("Gini Impurity:", gini)
````

Output:

```
Gini Impurity: 0.5
```

Meaning dataset is **mixed**.


# 🧠 Creating a Reusable Function


````python

# We now create a **function** so Milo can calculate Gini anytime.

# Function to calculate Gini impurity

def gini_impurity(labels):
    
    # Count total samples
    total = len(labels)
    
    # Count pass and fail
    num_pass = np.sum(labels == 1)
    num_fail = np.sum(labels == 0)
    
    # Convert counts to probabilities
    p_pass = num_pass / total
    p_fail = num_fail / total
    
    # Apply Gini formula
    gini = 1 - (p_pass**2 + p_fail**2)
    
    return gini
````

````python
# Example dataset
results = np.array([1,1,0,0])

# Calculate Gini impurity
gini_value = gini_impurity(results)

# Print result
print("Gini Impurity:", gini_value)
````
