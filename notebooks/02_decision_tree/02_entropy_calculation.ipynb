# 🧠 Simple Meaning of Entropy

Entropy simply means:

```text
How mixed or uncertain the data is
```

Think of it like **confusion level**.

| Dataset              | Entropy          |
| -------------------- | ---------------- |
| All Pass             | 0 (no confusion) |
| All Fail             | 0 (no confusion) |
| Half Pass, Half Fail | High entropy     |

So:

```text
More mixing → higher entropy
Less mixing → lower entropy
```

---

# 🧠 Goal of Decision Trees

Decision Trees always try to:

```text
Reduce entropy
```

Meaning they try to create **groups where results are mostly the same**.

Example:

Before splitting:

```text
Pass
Fail
Pass
Fail
```

Mixed → High entropy.

After splitting:

Group 1

```text
Pass
Pass
```

Group 2

```text
Fail
Fail
```

Now both groups are **pure**.

Entropy becomes **low**.

---

# 🧮 Entropy Formula (Explain Slowly)

The mathematical formula for entropy is:

```text
Entropy = − (p log₂ p + q log₂ q)
```

Where:

| Symbol | Meaning             |
| ------ | ------------------- |
| p      | Probability of Pass |
| q      | Probability of Fail |

---

### Example Calculation

Dataset:

| Result |
| ------ |
| Pass   |
| Pass   |
| Fail   |
| Fail   |

Counts:

```text
Pass = 2

Total = 4
```

Probabilities:

```text
p = 2/4 = 0.5
q = 2/4 = 0.5
```

Entropy:

```text
Entropy = -(0.5 log₂ 0.5 + 0.5 log₂ 0.5)
```

Result:

```text
Entropy = 1
```

Meaning **high uncertainty**.

# 🧪 Step 1 — Import Libraries
Now we calculate entropy using Python.


````python

# Import numpy
# numpy helps us work with numerical values
import numpy as np
````

# 🧪 Step 2 — Create Dataset


````python

# 1 represents Pass
# 0 represents Fail

results = np.array([1,1,0,0])
````

```
This dataset means:

```text
Pass
Pass
Fail
Fail
```

---

# 🧪 Step 3 — Calculate Probabilities


````python
# Count number of Pass results
num_pass = np.sum(results == 1)


# Count number of Fail results
num_fail = np.sum(results == 0)


# Total number of samples
total = len(results)


# Calculate probabilities
p_pass = num_pass / total
p_fail = num_fail / total
````

Explanation:

```text
p_pass → probability of passing
p_fail → probability of failing
```

# 🧪 Step 4 — Calculate Entropy


````python
# Calculate entropy using the entropy formula

entropy = -(
    p_pass * np.log2(p_pass) +
    p_fail * np.log2(p_fail)
)

print("Entropy:", entropy)
````



```

Output:

```text
Entropy: 1.0
```

Meaning:

```text
High uncertainty
```

---

# 📊 Visualizing Entropy


````python
# Create probabilities from 0 to 1
p = np.linspace(0.01,0.99,100)


# Calculate entropy for each probability
entropy_values = -(p*np.log2(p) + (1-p)*np.log2(1-p))


# Plot the entropy curve
import matplotlib.pyplot as plt

plt.plot(p, entropy_values)

plt.xlabel("Probability of Pass")

plt.ylabel("Entropy")

plt.title("Entropy vs Probability")

plt.show()
````

Now we see how entropy changes when probability changes.

# 🤖 How Milo Uses Entropy

When building a decision tree, Milo:

```text
1. Calculates entropy of the dataset
2. Tries different questions
3. Splits the data
4. Calculates entropy after splitting
5. Chooses the split that reduces entropy the most
```
