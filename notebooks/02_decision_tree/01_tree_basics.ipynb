# Decision Tree


> **What if the relationship between input and output is not a straight line?**

Until now, with **Logistic Regression**, we were learning patterns like:

```text
More study hours → higher chance of passing
```

That worked well when the data followed a **smooth curve**.

But many real-world problems work differently. They follow **rules and decisions**, not smooth curves.

Examples:

| Problem           | Decision                           |
| ----------------- | ---------------------------------- |
| Loan approval     | Income > 50k → Approve             |
| Email filtering   | Contains "free money" → Spam       |
| Medical diagnosis | Fever + cough → Possible infection |
| Student exam      | Hours < 3 → Fail                   |

This means we need an algorithm that can **learn rules**.

That algorithm is called a **Decision Tree**.

# 🌳 What is a Decision Tree?

A **Decision Tree** is a model that learns a **series of questions** to make a prediction.

It works like a **flowchart**.

Example tree:

```text
           Hours < 3 ?
            /      \
         Yes        No
        Fail    Attendance High?
                    /    \
                  Yes     No
                 Pass    Fail
```

Each question splits the data into smaller groups.

---

# 🌳 Parts of a Decision Tree

| Part          | Meaning                |
| ------------- | ---------------------- |
| Root Node     | First question         |
| Decision Node | Intermediate questions |
| Leaf Node     | Final prediction       |

Example:

```text
           Hours < 3 ?  ← Root Node
            /      \
         Yes        No
        Fail    Attendance High?  ← Decision Node
                    /    \
                  Pass   Fail  ← Leaf Nodes
```

---

# 🧠 Why Decision Trees Are Powerful

Decision trees can handle problems that **linear models struggle with**.

| Feature                  | Decision Trees |
| ------------------------ | -------------- |
| Non-linear relationships | ✅              |
| Categorical data         | ✅              |
| Easy to explain          | ✅              |
| Rule-based predictions   | ✅              |

That is why they are used in:

* Medical diagnosis
* Fraud detection
* Credit scoring
* Customer behavior prediction

# 🧪 Step 1 — Import Libraries


Now we will build a **Decision Tree in Python**.


````python
# Import numpy
# numpy helps us work with numerical arrays
import numpy as np

# Import matplotlib
# matplotlib is used to create graphs and visualizations
import matplotlib.pyplot as plt


# Import the Decision Tree classifier from sklearn
# This model learns decision rules from the data
from sklearn.tree import DecisionTreeClassifier


# Import function to visualize the tree
from sklearn.tree import plot_tree
````

# 🧪 Step 2 — Create Dataset


````python

# Hours studied by students
hours = np.array([1,2,3,4,5,6,7,8])


# Exam results
# 0 = Fail
# 1 = Pass
results = np.array([0,0,0,1,1,1,1,1])


# Convert input data into column format
# reshape(-1,1) turns the array into a column vector
X = hours.reshape(-1,1)


# Target variable
y = results
````

# 📊 Visualize the Dataset


````python
# Before training the model, we visualize the data.

# Create a scatter plot

plt.scatter(X, y)


# Label for x-axis
plt.xlabel("Hours Studied")


# Label for y-axis
plt.ylabel("Result (0 = Fail, 1 = Pass)")


# Title of the graph
plt.title("Student Exam Dataset")


# Display the graph
plt.show()
````


```text
1 |        ● ● ● ● ●
  |
0 |  ● ● ●
  +-------------------
      Hours studied
```

Meaning:

```text
Low study hours → Fail
High study hours → Pass
```

# 🧪 Step 3 — Train the Decision Tree


````python

# Now we train the decision tree model.


# Create a decision tree model

model = DecisionTreeClassifier()


# Train the model using the dataset
# fit() means the algorithm learns patterns from the data
model.fit(X, y)
````

```text
model.fit(X, y)
```

This tells the algorithm:

```text
Look at study hours and exam results
Find rules that separate Fail and Pass
```

# 🧪 Step 4 — Make Predictions


````python
# Now we test the model.

# Predict result for a student studying 5 hours

prediction = model.predict([[5]])


# Print the prediction
print("Prediction:", prediction)
````

```

Example output:

```text
Prediction: [1]
```

Meaning:

```text
1 = Pass
```


# 🧪 Step 5 — Visualize the Decision Tree


````python
# One of the **best things about Decision Trees** is that we can actually **see the model**.

# Set figure size for the tree visualization
plt.figure(figsize=(6,4))


# Draw the decision tree
plot_tree(
    model,
    feature_names=["Hours Studied"],
    class_names=["Fail","Pass"],
    filled=True
)


# Show the tree
plt.show()
````


The model learned this rule automatically.

```text
Hours ≤ 3.5 → Fail
Hours > 3.5 → Pass
```
