# 🧠 What Is a Cost Function?

A **Cost Function** tells the model:

```text
How wrong your predictions are
```

Think of it like **marks in an exam**.

| Prediction | Actual | Correct? |
| ---------- | ------ | -------- |
| 0.30       | 0      | Good     |
| 0.80       | 1      | Good     |
| 0.20       | 1      | Very Bad |

The cost function gives a **penalty score**.

```text
More mistakes → Higher cost
Better predictions → Lower cost
```

The model's goal:

```text
MINIMIZE THE COST
```

---

# 🎬 Chapter 5 — The Punishment Rule (Log Loss)

Professor Arjun explains:

Log Loss punishes the model **more if it is confidently wrong**.

Example:

| Prediction | Actual | Error         |
| ---------- | ------ | ------------- |
| 0.9        | 1      | small error   |
| 0.6        | 1      | small error   |
| 0.1        | 1      | huge error 🚨 |

Why?

Because predicting **0.1 when the answer is 1** is a terrible mistake.

Log Loss punishes this **very strongly**.



---

# 📊 Log Loss Graph (Visual)



````python

# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Probability values
p = np.linspace(0.01, 0.99, 100)

# Log loss when actual = 1
loss = -np.log(p)

# Plot
plt.plot(p, loss)

plt.xlabel("Predicted Probability")
plt.ylabel("Log Loss")

plt.title("Log Loss when Actual = 1")

plt.show()
````



## 🧠 Code Explanation

### Generate probabilities

```python
p = np.linspace(0.01, 0.99, 100)
```

We generate values from **0.01 to 0.99**.

Why not 0?

Because:

```text
log(0) = infinity
```

---

### Calculate Log Loss

```python
loss = -np.log(p)
```

This calculates the **penalty**.

---

# 📈 What you Will See

Graph behavior:

```text
Prediction near 1 → Loss very small
Prediction near 0 → Loss extremely large
```

Meaning:

```text
Confidently wrong predictions get punished heavily.
```

---

# 🎯 Final Log Loss Formula

```text
Loss = -[y log(p) + (1-y) log(1-p)]
```

```text
Correct prediction → small loss
Wrong prediction → big loss
```

# 🎬 Chapter 6 — How the Model Improves (Gradient Descent)





Now Riya asks:

> "Sir… Milo knows the error now…
> But how does he fix it?"

Professor Arjun smiles.

> "He searches for better parameters."

Remember this equation:

```text
z = mx + b
```

The model must find the **best m and b**.

---

# 🧠 Idea of Gradient Descent

Imagine Milo standing on a **mountain**.

Top of mountain:

```text
High cost
```

Bottom of valley:

```text
Lowest cost
```

Goal:

```text
Find the lowest point
```

Gradient Descent simply means:

```text
Take small steps downhill
```

# 📊 Visualizing Gradient Descent


````python
import numpy as np
import matplotlib.pyplot as plt

# Cost values
x = np.linspace(-5,5,100)
y = x**2

plt.plot(x,y)

plt.title("Cost Function Curve")

plt.xlabel("Parameter Value")
plt.ylabel("Cost")

plt.show()
````

--- 

## 🧠 Code

### Create range

```python
x = np.linspace(-5,5,100)
```

Different possible **parameter values**.

---

### Cost function

```python
y = x**2
```

This simulates **cost**.

The lowest point is:

```text
x = 0
```

---

# 📈  Explanation

You will see a **U-shaped curve**.

```text
      Cost
       ▲
       |        ●
       |     ●
       |   ●
       | ●
       |●
       +--------------►
             Parameter
```

Gradient Descent moves like this:

```text
Step → Step → Step → Lowest cost
```



---

# 🤖 Milo's Learning Process

Each training step:

```text
1. Make prediction
2. Calculate log loss
3. Adjust m and b slightly
4. Repeat
```

After many iterations:

```text
Model finds best parameters
```
