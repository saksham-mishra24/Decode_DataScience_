| ------------- | ---------- |
| 1             | 40         |

````python
import matplotlib.pyplot as plt
import numpy as np

hours = np.array([1,2,3,4,5])
scores = np.array([40,50,60,70,80])

plt.scatter(hours, scores)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.title("Linear Regression Example")

plt.show()
````

# 📈 Add the Linear Regression Line


````python
m = 10
b = 30

predicted = m*hours + b

plt.scatter(hours, scores)
plt.plot(hours, predicted)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.title("Best Fit Line")

plt.show()
````

# 🧠 The Core Problem

Linear regression predicts **any number**.

```
-10
2.5
100
```

But classification needs only:

```
0 or 1
```

Or in probability form:

```
0 to 1
```

So we need a model that **never goes below 0 or above 1**.

---


Linear regression line:

```
-∞  ----------------  +∞
```

But probabilities must stay between:

```
0 ----------- 1
```

So we need a **curve that stays between 0 and 1**.

---



# 🤖 The Solution Appears

Professor draws this curve.

```
      ________
     /
    /
___/
```

This is called the **Sigmoid Curve**.

It looks like a **lazy S**.

And the magic rule:

```
Output always between 0 and 1
```

---



# 🎯 This Is Logistic Regression

Logistic Regression =

```
Linear Regression
+
Sigmoid Function
```

First we calculate:

```
z = mx + b
```

Then we convert it using sigmoid:

```
Probability = sigmoid(z)
```

Now predictions become:

```
0.02
0.31
0.76
0.91
```

All valid probabilities.

---

# 📊 Let's Draw the Sigmoid Curve 


````python

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)

y = 1/(1+np.exp(-x))

plt.plot(x,y)

plt.title("Sigmoid Function")
plt.xlabel("Input")
plt.ylabel("Probability")

plt.show()
````

```

When x is very small:

```
Probability ≈ 0
```

When x is very big:

```
Probability ≈ 1
```

---



# 🎉 The Big Idea

Logistic Regression does:

```
Step 1 → Draw a line
Step 2 → Pass it through sigmoid
Step 3 → Convert to probability
```
