# 🧠 Step 1 — Create Our Dataset

First we create a simple dataset.

This dataset represents:

```text
Hours studied → Pass or Fail
```

````python

# Import libraries
# numpy is used for numerical calculations
# matplotlib is used for visualization

import numpy as np
import matplotlib.pyplot as plt

# Create dataset
hours = np.array([1,2,3,4,5,6,7,8])
# Actual results
# 0 = Fail
# 1 = Pass

result = np.array([0,0,0,0,1,1,1,1])
````



---

```python
hours = np.array([1,2,3,4,5,6,7,8])
```

```python
result = np.array([0,0,0,0,1,1,1,1])
```

# 🧠 Step 2 — Define Sigmoid Function


Now Milo needs the **probability function**.


````python
# Define sigmoid function

def sigmoid(z):
    
    # sigmoid formula
    return 1 / (1 + np.exp(-z))
````

---

## Code Explanation

This function converts **any number** into probability.

Example:

| Input | Output |
| ----- | ------ |
| -5    | 0.006  |
| 0     | 0.5    |
| 5     | 0.99   |

So sigmoid keeps outputs **between 0 and 1**.

---

# 🧠 Step 3 — Start With Random Parameters


````python
# Milo begins with **random guesses**.

# Random starting parameters

m = -1
b = 2
````


This creates the equation:

```
z = -1 × hours + 2
```

But this line is **probably wrong**.

---

# 🧠 Step 4 — Calculate Predictions


````python
# Calculate z values

z = m * hours + b


# Convert z into probabilities using sigmoid

predictions = sigmoid(z)


print(predictions)
````

````python
plt.scatter(hours, result, label="Actual Data")

plt.scatter(hours, predictions, label="Predicted Probabilities")

plt.xlabel("Hours Studied")
plt.ylabel("Probability")

plt.legend()

plt.title("Initial Predictions")

plt.show()
````

## What we Notice

The predictions will **not match perfectly**.

Example:

| Hours | Actual | Predicted |
| ----- | ------ | --------- |
| 2     | 0      | 0.62      |
| 3     | 0      | 0.50      |
| 4     | 0      | 0.38      |
| 5     | 1      | 0.27      |

The model is **very wrong**.

---

# 🎯 Now Milo Must Improve

Professor Arjun explains:

```
If predictions are wrong
→ adjust m and b
```

This is **Gradient Descent**.

---

# 🧠 Step 5 — Adjust Parameters

Instead of jumping to calculus, we simulate learning.

````python
learning_rate = 0.1

for step in range(10):

    z = m * hours + b
    predictions = sigmoid(z)

    # Calculate error
    error = result - predictions

    # Update parameters
    m = m + learning_rate * np.sum(error * hours)
    b = b + learning_rate * np.sum(error)

    print("Step:", step)
    print("m:", m)
    print("b:", b)
````



# Simulate learning


---

# 🧠 Code Explanation

### Learning Rate

```python
learning_rate = 0.1
```

This controls **how big each learning step is**.

Think of it like **step size while walking downhill**.

Too big → overshoot
Too small → slow learning

---

### Loop

```python
for step in range(10):
```

We repeat learning **10 times**.

Each loop = **one training step**.

---

### Calculate Error

```python
error = result - predictions
```

This measures:

```
Actual value − predicted probability
```

---

### Update Parameters

```python
m = m + learning_rate * np.sum(error * hours)
```

Adjust the **slope**.

---

```python
b = b + learning_rate * np.sum(error)
```

Adjust the **intercept**.

---


# 📈 Visualizing Learning Progress


````python
plt.scatter(hours, result, label="Actual")

plt.scatter(hours, predictions, label="Final Predictions")

plt.xlabel("Hours")
plt.ylabel("Probability")

plt.legend()

plt.title("Model After Learning")

plt.show()
````

You will see predictions **move closer to correct values**.
