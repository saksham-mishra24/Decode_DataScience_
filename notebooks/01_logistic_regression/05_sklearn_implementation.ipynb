# 🧪 Step 1 — Import Libraries


````python
# Import NumPy for numerical calculations
import numpy as np

# Import Matplotlib for graphs
import matplotlib.pyplot as plt

# Import Logistic Regression model from scikit-learn 
# linear_model is a module in scikit-learn that contains various linear models, including Logistic Regression.
# logistic_regression is a class within the linear_model module that implements the Logistic Regression algorithm, which is used for classification tasks.
from sklearn.linear_model import LogisticRegression 
````

# 🧪 Step 2 — Create Dataset


````python
# Hours students studied
hours = np.array([1,2,3,4,5,6,7,8])

# Exam results
# 0 = Fail
# 1 = Pass

result = np.array([0,0,0,0,1,1,1,1])
````

---

## Important Note

Scikit-Learn expects **input features in 2D format**.

So we must reshape the data.


````python
# Reshape hours into column format
# -1 means automatically calculate rows
# 1 means one column

X = hours.reshape(-1,1)

# Target values

y = result
````

---

## 🧠 Why reshape?

Right now hours looks like:

```
[1 2 3 4 5 6 7 8]
```

But the model expects:

```
[[1]
 [2]
 [3]
 [4]
 [5]
 [6]
 [7]
 [8]]
```

Each row = **one student**.

````python
# Plot the dataset

plt.scatter(X, y)

plt.xlabel("Hours Studied")
plt.ylabel("Pass (1) / Fail (0)")

plt.title("Student Exam Dataset")

plt.show()
````

### What we See

```
1 |        ● ● ● ●
  |
0 |  ● ● ● ●
  |
  +----------------
     Hours studied
```

- Left → Fail
- Right → Pass

---


# 🧠 Step 3 — Create the Model


````python
# Create logistic regression model

model = LogisticRegression()
# this is creating an object because without creating an object, we cannot use the methods of the LogisticRegression class.
````


## Explanation

This line creates the **machine learning model object**.

Think of it like creating an **empty brain**.

Right now the model **has not learned anything**.

# 🧠 Step 4 — Train the Model


````python
# Train the model using the dataset

model.fit(X, y)
````


## Explanation

This is the **most important line**.

```
model.fit()
```

means:

```
Learn patterns from the data
```

During this step the algorithm:

- 1️⃣ Applies sigmoid
- 2️⃣ Calculates log loss
- 3️⃣ Uses gradient descent
- 4️⃣ Finds best parameters **m and b**

All automatically.

---

# 🤖 Milo Has Now Learned

The model now knows:

```
More study hours → higher chance of passing

# 🧪 Step 5 — Make Predictions


````python
# Predict class for new students

prediction = model.predict([[4]])

print(prediction)
````

## Code Explanation

We ask the model:

```
If a student studies 4 hours,
will they pass?
```

Input must be **2D**.

So we write:

```
[[4]]
```

Example output:

```
[0]
```

Meaning:

```
Fail
```

---

# 🧪 Step 6 — Predict Probability


Now we ask Milo:

> "How confident are you?"


````python
# Predict probability

prob = model.predict_proba([[4]])

print(prob)
````



---

## Output 

```
[[0.63 0.37]]
```

Meaning:

| Class | Probability |
| ----- | ----------- |
| Fail  | 63%         |
| Pass  | 37%         |

So Milo predicts:

```
Fail
```

Because:

```
0.37 < 0.5
```


# 📊 Visualizing the Decision Boundary
Now we draw the **sigmoid curve learned by the model**.


````python
# Create smooth range of hours

x_test = np.linspace(0,10,100).reshape(-1,1)

# Get probabilities

probabilities = model.predict_proba(x_test)[:,1]

# Plot sigmoid curve

plt.plot(x_test, probabilities)

# Plot actual data

plt.scatter(X, y)

# Decision boundary

plt.axhline(0.5)

plt.xlabel("Hours Studied")
plt.ylabel("Probability of Passing")

plt.title("Logistic Regression Decision Curve")

plt.show()
````


---

# 🧠 Code Explanation

### Generate smooth inputs

```
np.linspace(0,10,100)
```

Creates **100 values between 0 and 10**.

This helps draw a **smooth curve**.

---

### Get probability of class 1

```
model.predict_proba(x_test)[:,1]
```

This extracts the probability of **passing**.

---

### Plot sigmoid curve

```
plt.plot()
```

Draws the **logistic curve**.
