# 🧠 The Solution — Train Test Split

Instead of using **all data for training**, we divide it.

```text
Dataset
   |
   |---- Training Data (used for learning)
   |
   |---- Testing Data (used for evaluation)
```

Example:

| Dataset Size | Split              |
| ------------ | ------------------ |
| 100 rows     | 80 train + 20 test |

The model learns **only from training data**.

Then we test it on **unseen data**.

---

# 🧪 Step 1 — Import Train Test Split


````python
# Import required libraries

import numpy as np
import matplotlib.pyplot as plt

# Import logistic regression model
from sklearn.linear_model import LogisticRegression

# Import train_test_split function
from sklearn.model_selection import train_test_split
# model_selection module is used to split the data into training and testing sets
````

# 🧪 Step 2 — Create Dataset


````python
# Input feature (hours studied)
hours = np.array([1,2,3,4,5,6,7,8])

# Output labels (pass/fail)
result = np.array([0,0,0,0,1,1,1,1])

# Reshape input data into column format
X = hours.reshape(-1,1)

# Target variable
y = result
````

# 🧪 Step 3 — Split Dataset


````python

# Split dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,   # 25% data used for testing
    random_state=42   # ensures same split every time
)
````

---

## 🧠 Code Explanation

### test_size

```python
test_size = 0.25
```

Means:

```text
25% of data → Testing
75% of data → Training
```

---

### random_state

```python
random_state = 42
```

This ensures the **same random split every time**.

Without it, results might change.

# 🧠 Step 4 — Train Model


````python
# Create logistic regression model
model = LogisticRegression()

# Train model using only training data
model.fit(X_train, y_train)
````


Notice something important:

```text
The model NEVER sees the test data.
```

So testing is **fair**.


# 🧪 Step 5 — Make Predictions on Test Data


````python
# Predict results for test dataset

predictions = model.predict(X_test)

print("Actual values:", y_test)
print("Predicted values:", predictions)
````


Example output:

```text
Actual values: [1 0]
Predicted values: [1 0]
```

# 🎬 Chapter 10 — Understanding the Confusion Matrix

Professor Arjun draws a table.

|             | Predicted Fail | Predicted Pass |
| ----------- | -------------- | -------------- |
| Actual Fail | TN             | FP             |
| Actual Pass | FN             | TP             |


---

# 🧠 Confusion Matrix Meaning

| Term | Meaning                          |
| ---- | -------------------------------- |
| TP   | Correctly predicted Pass         |
| TN   | Correctly predicted Fail         |
| FP   | Predicted Pass but actually Fail |
| FN   | Predicted Fail but actually Pass |

---

# 🧪 Generate Confusion Matrix


````python

# Import confusion matrix function

from sklearn.metrics import confusion_matrix

# Calculate confusion matrix

cm = confusion_matrix(y_test, predictions)

print(cm)
````

# 📊 Visualizing Confusion Matrix


````python
import seaborn as sns

# Plot confusion matrix heatmap

sns.heatmap(cm, annot=True, cmap="Blues")
# annot=True adds numerical values to each cell in the heatmap
# cm is the confusion matrix data
# why heatmap because it provides a visual representation of the confusion matrix, making it easier to interpret the results 

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.title("Confusion Matrix")

plt.show()
````


---

# 🧠 Evaluation Metrics

Now we convert this matrix into **scores**.

---

# 🎯 Accuracy

Accuracy measures:

```text
How many predictions were correct
```

Formula:

```text
Accuracy = (TP + TN) / Total
```


# 🧪 Accuracy


````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, predictions)

print("Accuracy:", accuracy)
````


Example output:

```text
Accuracy: 0.9
```

Meaning:

```text
Model is 90% correct


# 🎯 Precision

Precision answers:

```text
When the model predicts PASS,
how often is it correct?
```

Formula:

```text
Precision = TP / (TP + FP)
```

---

# 🎯 Recall

Recall answers:

```text
Out of all actual PASS cases,
how many did we detect?
```

Formula:

```text
Recall = TP / (TP + FN)
```

---

# 🧪 Precision and Recall


````python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test, predictions)

recall = recall_score(y_test, predictions)

print("Precision:", precision)
print("Recall:", recall)
````
