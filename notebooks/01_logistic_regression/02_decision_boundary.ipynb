# 🧠 The Secret Rule

Professor Arjun explains:

Machines use a **cutoff rule**.

```
If Probability ≥ 0.5 → Class = 1 (Pass)

If Probability < 0.5 → Class = 0 (Fail)
```

This rule is called:

# 🎯 Decision Boundary

Think of it like a **line that separates two worlds**.

```
Fail Zone | Pass Zone
```

---

# 📊 Example Probabilities

| Hours Studied | Probability | Prediction |
| ------------- | ----------- | ---------- |
| 2             | 0.10        | Fail       |
| 3             | 0.22        | Fail       |
| 4             | 0.40        | Fail       |
| 5             | 0.55        | Pass       |
| 6             | 0.80        | Pass       |

Notice something interesting.

Between **4 and 5 hours** the prediction flips.

That point is the **decision boundary**.

---

# 📈 Let's Visualize This With Code


````python
# Import libraries
# numpy helps with mathematical operations
# matplotlib helps us draw graphs

import numpy as np
import matplotlib.pyplot as plt


# Create input values from -10 to 10
# This simulates different possible inputs to the sigmoid function

x = np.linspace(-10, 10, 100)


# Define the sigmoid function
# sigmoid(x) = 1 / (1 + e^(-x))
# np.exp() calculates the exponential value

y = 1 / (1 + np.exp(-x))


# Plot the sigmoid curve

plt.plot(x, y)


# Draw a horizontal line at probability = 0.5
# This shows the decision boundary

plt.axhline(0.5)


# Label the axes so students know what they mean

plt.xlabel("Input Value (z)")
plt.ylabel("Probability")


# Add a title to explain what the graph shows

plt.title("Sigmoid Function and Decision Boundary")


# Display the graph

plt.show()
````


These numbers represent **inputs to the sigmoid function**.

---

## 2️⃣ Sigmoid Formula

```
y = 1 / (1 + np.exp(-x))
```

This is the **sigmoid formula**.

It converts **any number** into a value between:

```
0 and 1
```

Example conversions:

| Input | Output |
| ----- | ------ |
| -5    | 0.006  |
| 0     | 0.5    |
| 5     | 0.99   |

---

## 3️⃣ Plotting the Curve

```
plt.plot(x,y)
```

This draws the **sigmoid curve**.

---

## 4️⃣ Drawing the Decision Line

```
plt.axhline(0.5)
```

This draws a **horizontal line at 0.5**.

Why?

Because **0.5 is the decision boundary**.

# 🤖 How Milo Actually Decides

Step by step:

```

Step 3: Compare with 0.5
Step 4: Decide class
```

```
z = 1.2
```


Decision rule:

```
0.77 > 0.5
```

Prediction:

```

```

# 🎭 Analogy

Professor Arjun tells Milo:

Imagine a **door with a guard**.

```
Probability < 0.5 → Guard says NO
Probability ≥ 0.5 → Guard says YES
```

So the guard decides:

```
Fail or Pass
```

That guard is the **decision boundary**.
