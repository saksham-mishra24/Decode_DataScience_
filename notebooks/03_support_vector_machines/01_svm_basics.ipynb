# 📊 Our Dataset

We will use two features:

| Feature     | Meaning                      |
| ----------- | ---------------------------- |
| Study Hours | How long the student studied |
| Sleep Hours | How long the student slept   |

Dataset:

| Study | Sleep | Result |
| ----- | ----- | ------ |
| 2     | 6     | Fail   |
| 3     | 5     | Fail   |
| 4     | 6     | Fail   |
| 6     | 7     | Pass   |
| 7     | 6     | Pass   |
| 8     | 7     | Pass   |

---

````python
# Import numpy
# numpy is used to create and handle numerical datasets
import numpy as np

# Import the Support Vector Machine classifier
# SVC stands for Support Vector Classifier
from sklearn.svm import SVC

# Import matplotlib for plotting graphs
import matplotlib.pyplot as plt
````



# 🧪 Step 2 — Create the Dataset



````python
# X represents the input features
# Each row represents one student

# [Study Hours, Sleep Hours]

X = np.array([
    [2,6],
    [3,5],
    [4,6],
    [6,7],
    [7,6],
    [8,7]
])


# y represents the output labels
# 0 means Fail
# 1 means Pass

y = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])
````



Each row is a **vector describing a student**.

Example:

```text
[2,6]
```

means

```text
Study Hours = 2
Sleep Hours = 6
```

---

Now we create the **labels (answers)**.



Meaning:

| Vector | Result |
| ------ | ------ |
| [2,6]  | Fail   |
| [3,5]  | Fail   |
| [4,6]  | Fail   |
| [6,7]  | Pass   |
| [7,6]  | Pass   |
| [8,7]  | Pass   |

---


# 🧪 Step 3 — Create the SVM Model


````python
# Create the SVM model

# kernel="linear" means the decision boundary will be a straight line

model = SVC(kernel="linear")
````

### Explanation

Important parameter:

```text
kernel
```

Kernel tells the model **what type of boundary to create**.

| Kernel | Meaning          |
| ------ | ---------------- |
| linear | straight line    |
| rbf    | curved boundary  |
| poly   | polynomial curve |

Right now we use:

```text
linear
```

---

# 🧪 Step 4 — Train the Model


````python
# Train the model using the dataset

# X contains the input features
# y contains the correct answers

model.fit(X, y)
````



What happens here?

```text
1 Milo sees all vectors
2 Milo finds the best hyperplane
3 Milo identifies support vectors
4 Milo maximizes the margin
```

# 🧪 Step 5 — Check the Support Vectors


````python

# Print the support vectors

print("Support Vectors:")

print(model.support_vectors_)
````


Example output might look like:

```
Support Vectors:
[[4 6]
 [6 7]]
```

Meaning:

```text
These points are closest to the boundary
```

They control the **position of the hyperplane**.

---


# 🧪 Step 6 — Make Predictions


````python


# Now we test Milo.

# Suppose a student studies **5 hours** and sleeps **6 hours**.

# Predict result for a new student

prediction = model.predict([[5,6]])

print("Prediction:", prediction)
````

# 🧪 Step 7 — Visualizing the Dataset


````python
# Plot the data points

# Loop through each point in the dataset
for i in range(len(X)):
    
    # If the label is 0 (Fail)
    if y[i] == 0:
        
        # Plot a red circle
        plt.scatter(X[i][0], X[i][1], color='red',marker='*')
    
    # If label is 1 (Pass)
    else:
        
        # Plot a blue triangle
        plt.scatter(X[i][0], X[i][1], color='blue')

# Label the axes
plt.xlabel("Study Hours")
plt.ylabel("Sleep Hours")

# Title of the graph
plt.title("Student Dataset")

# Show the graph
plt.show()
````
