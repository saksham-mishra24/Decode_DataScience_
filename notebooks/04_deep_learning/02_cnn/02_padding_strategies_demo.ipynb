## Cell 4 — First CNN model: `padding='valid'`

Ab hum ek CNN bana rahe hain.

### Model ka idea
Ye model image ke andar edges, curves, strokes jaise patterns dhoondta hai, taaki digit identify kar sake.

### Line by line breakdown

- `model = Sequential()`  
  Iska matlab ek aisa model banana jo layers ko sequence me add karega.

- `model.add(Conv2D(32, kernel_size=(3,3), padding='valid', activation='relu', input_shape=(28,28,1)))`  
  Ye pehli convolution layer hai:
  - `32` = 32 filters, yani 32 different pattern detectors
  - `kernel_size=(3,3)` = 3x3 ka chhota window
  - `padding='valid'` = border ke bahar extra pixels add nahi honge
  - `activation='relu'` = negative values ko zero bana deta hai, model ko non-linear banata hai
  - `input_shape=(28,28,1)` = input image 28x28 pixels ki hai, aur `1` channel grayscale ka hai

- Doosri aur teesri `Conv2D` layers same kaam karti hain. Har layer deeper features seekhti hai.

- `model.add(Flatten())`  
  Convolution ka output 3D hota hai. `Flatten()` usko ek lambi 1D line me badal deta hai, taaki Dense layer use samajh sake.

- `model.add(Dense(128,activation='relu'))`  
  Ye hidden layer hai. Isme 128 neurons hain.

- `model.add(Dense(10,activation='softmax'))`  
  Final output layer. 10 classes hain, kyunki digits 0 se 9 tak hain.  
  `softmax` probabilities deta hai — kis digit ke hone ki chance kitni hai.

### `valid` padding ka matlab
Input ke border par extra padding nahi lagti, isliye image size har convolution ke baad chhota hota jata hai.

````python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
````

## Cell 5 — First model ka summary output samajhna

`model.summary()` model ki poori structure table me dikhata hai.

Is table me 4 cheezein important hain:

- **Layer (type)** → kaunsi layer hai
- **Output Shape** → layer ke baad data ka shape kya ban gaya
- **Param #** → layer me kitne trainable numbers hain
- **Total params** → total trainable parameters

### Output ko line by line samjho

- `conv2d_4 (Conv2D) -> (None, 26, 26, 32)`  
  Input 28x28 tha. 3x3 filter aur `valid` padding ki wajah se size 26x26 ho gaya.  
  `32` means 32 feature maps.

- `conv2d_5 -> (None, 24, 24, 32)`  
  Dobara 3x3 convolution laga, aur size aur chhota ho gaya.

- `conv2d_6 -> (None, 22, 22, 32)`  
  Teesri convolution ke baad size aur reduce hua.

- `flatten_1 -> (None, 15488)`  
  22 × 22 × 32 = 15488  
  Ab 3D data ek long vector ban gaya.

- `dense_2 -> (None, 128)`  
  Fully connected layer output 128 numbers me badal deti hai.

- `dense_3 -> (None, 10)`  
  Final 10 probabilities.

### Parameters ka simple meaning
Parameters model ke andar woh numbers hote hain jo training ke dauran learn hote hain.  
Jitne zyada parameters, utna model zyada flexible ho sakta hai — lekin training bhi heavier ho sakti hai.

### Total params
- Total parameters = **2,002,698**

Matlab ye model kaafi bada hai.

````python
model.summary()
````

## Cell 6 — Second CNN model: `padding='same'` aur `strides=(2,2)`

Ab hum second model bana rahe hain, jisme do major changes hain:

- `padding='same'`
- `strides=(2,2)`

### Inka matlab

- `padding='same'` → output size ko roughly same rakhne ki koshish hoti hai
- `strides=(2,2)` → filter image par 2 steps karke move karta hai, isse size faster reduce hota hai

### Line by line breakdown
Baaki structure pehle model jaisa hi hai:
- 3 convolution layers
- flatten
- dense 128
- dense 10

Difference sirf ye hai ki convolution ka movement aur padding alag hai, isliye output shape alag aati hai.

````python
model = Sequential()

model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
````

## Cell 7 — Second model ka summary output samajhna

Ab summary table compare karo pehle model se.

### Output shapes

- `conv2d -> (None, 14, 14, 32)`  
  28x28 image par stride 2 lagaya, to size half ho gaya.

- `conv2d_1 -> (None, 7, 7, 32)`  
  Dobara half hua.

- `conv2d_2 -> (None, 4, 4, 32)`  
  Teesri convolution ke baad aur small ho gaya.

- `flatten -> (None, 512)`  
  4 × 4 × 32 = 512  
  Isliye flatten output sirf 512 numbers ka hai.

- `dense -> (None, 128)`  
  Same hidden layer.

- `dense_1 -> (None, 10)`  
  Final 10 classes.

### Sabse important difference
Pehle model me flatten output **15488** tha, ab sirf **512** hai.

Iska matlab:
- second model much smaller representation banata hai
- parameters bhi kam ho jaate hain
- total params = **85,770**

### Compare karne layak baat
- `padding='valid'` model = much larger
- `padding='same'` + stride 2 model = much smaller and faster

Yehi notebook ka main lesson hai:  
**padding aur stride output shape ko kaise change karte hain.**
