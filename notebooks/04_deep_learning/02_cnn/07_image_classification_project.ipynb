````python

# 🧪 Step 4 — Import Libraries

# Deep learning library
import tensorflow as tf

# Image processing
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Model building
from tensorflow.keras import layers, models

# Visualization
import matplotlib.pyplot as plt
````

# 🧠 Explanation

* **ImageDataGenerator** → loads and prepares images
* **layers, models** → build CNN
* **matplotlib** → show images

````python

# 🧪 Step 5 — Data Preprocessing

# Create image generator for training
train_datagen = ImageDataGenerator(
    rescale=1./255,        # normalize pixel values
    shear_range=0.2,       # slight distortion
    zoom_range=0.2,        # zoom in/out
    horizontal_flip=True   # flip images
)

# Create generator for testing (no augmentation)
test_datagen = ImageDataGenerator(rescale=1./255)
````




# 🧠 Simple explanation

* **rescale** → convert 0–255 → 0–1
* **flip/zoom** → create more variety → better learning

---

````python


# 🧪 Step 6 — Load Images from Folder


# Load training images
train_data = train_datagen.flow_from_directory(
    "dataset/train",       # folder path
    target_size=(64, 64),  # resize images
    batch_size=32,         # images per batch
    class_mode="binary"    # 2 classes (dog/cat)
)

# Load test images
test_data = test_datagen.flow_from_directory(
    "dataset/test",
    target_size=(64, 64),
    batch_size=32,
    class_mode="binary"
)
````


```

---

# 🧠 Explanation

* **target_size** → all images resized to same size
* **batch_size** → process 32 images at a time
* **class_mode="binary"** → dog = 0, cat = 1

---

# 🧪 Step 7 — Build CNN Model

```python
# Create model
model = models.Sequential()

# Convolution + Pooling
model.add(layers.Conv2D(32, (3,3), activation="relu", input_shape=(64,64,3)))
model.add(layers.MaxPooling2D((2,2)))

# Another Conv layer
model.add(layers.Conv2D(64, (3,3), activation="relu"))
model.add(layers.MaxPooling2D((2,2)))

# Flatten
model.add(layers.Flatten())

# Dense layers
model.add(layers.Dense(128, activation="relu"))

# Output layer
model.add(layers.Dense(1, activation="sigmoid"))
```

---

# 🧠 Code Explanation (VERY IMPORTANT)

### Conv Layer

```python
Conv2D(32, (3,3))
```

→ 32 filters → detect patterns

---

### Pooling

```python
MaxPooling2D
```

→ reduce size

---

### Flatten

```python
Flatten()
```

→ convert image → list

---

### Dense

```python
Dense(128)
```

→ learn complex patterns

---

### Output

```python
Dense(1, activation="sigmoid")
```

→ binary output (dog or cat)

---

# 🧪 Step 8 — Compile Model

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

---

# 🧠 Explanation

* binary problem → use **binary_crossentropy**
* sigmoid → gives probability

---

# 🧪 Step 9 — Train Model

```python
model.fit(
    train_data,
    epochs=5,
    validation_data=test_data
)
```

# 🧪 Step 10 — Test Prediction

```python
# Load a single image
img = tf.keras.preprocessing.image.load_img(
    "dataset/test/cats/cat1.jpg",
    target_size=(64,64)
)

# Convert to array
img_array = tf.keras.preprocessing.image.img_to_array(img)

# Normalize
img_array = img_array / 255.0

# Add batch dimension
img_array = img_array.reshape(1, 64, 64, 3)

# Predict
prediction = model.predict(img_array)

print(prediction)
```
