# --- RNN parameters ---


```python
input_dim = 1
hidden_dim = 2
output_dim = 1
```

```python
np.random.seed(42)

# random seed used when generating random numbers, so that the same random numbers are generated every time the code is run
```

```python
w_x = np.random.rand(input_dim, hidden_dim)
# weights for the input to hidden layer connections
```

```python
w_h = np.random.rand(hidden_dim, hidden_dim)
# hiden to hidden layer connections beccause we will be using the hidden state from the previous time step as input to the current time step
```

```python
w_y = np.random.rand( output_dim, hidden_dim)
```

```python
b_h = np.random.rand(hidden_dim)
# bias for the hidden layer 
b_y = np.random.rand(output_dim)
# bias for the output layer
```

```python
h_prev = np.zeros(hidden_dim)
# initial hidden state, initialized to zeros
```

```python
x.shape[0]
# we used shape[0] because we want the number of rows in the input, which corresponds to the number of time steps in the sequence.
```

```python
for t in range(x.shape[0]):
    x_t = x[t]
# in each time step, we take the corresponding input from the sequence x, which is x[t], and store it in x_t. 
#  allows us to process each time step of the input sequence sequentially in the loop.
    x_t = x[t].reshape(1, -1)
# reshape x_t to have a shape of (1, input_dim) so that it can be multiplied with the weight matrices w_x and w_h.
    h_t = np.tanh(np.dot(x_t, w_x) + np.dot(h_prev, w_h) + b_h)
    # compute the hidden state at time step t using the input x_t, the previous hidden state h_prev, and the weights and biases.
    y_t = np.dot(h_t, w_y.T) + b_y
    # compute the output at time step t using the hidden state h_t and the weights and biases for the output layer. 
    # T is used to transpose w_y so that the dimensions match for the matrix multiplication.
    h_prev = h_t
    print(f"Time step {t}:")
    print(f"Input x_t: {x_t}")
    print(f"Hidden state h_t: {h_t}")
    print(f"Output y_t: {y_t}")
```
