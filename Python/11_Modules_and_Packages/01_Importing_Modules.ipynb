## 🤔 What is a Module?

> A **module** is a Python file (.py) containing functions, classes, and variables that you can **reuse** in other programs.

### Analogy: Toolbox 🧰
Instead of making your own hammer every time, you pick one from the toolbox. Modules are Python's toolboxes!

---

```python
# Different ways to import

# Way 1: Import entire module
import math
print("Square root of 144:", math.sqrt(144))
print("Value of π:", math.pi)
print("Factorial of 5:", math.factorial(5))

# Way 2: Import specific functions
from math import sqrt, pi
print("\nsqrt(225):", sqrt(225))   # No need for math.
print("pi:", pi)

# Way 3: Import with alias (nickname)
import math as m
print("\nm.ceil(4.3):", m.ceil(4.3))   # 5
print("m.floor(4.7):", m.floor(4.7))   # 4
```

```python
# random module — Generate random numbers
import random

print("Random number (1-10):", random.randint(1, 10))
print("Random float (0-1):", round(random.random(), 4))

# Random choice from list
fruits = ["Apple", "Banana", "Mango", "Grapes"]
print("Random fruit:", random.choice(fruits))

# Shuffle a list
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("Shuffled:", numbers)
```

```python
# datetime module — Work with dates and times
from datetime import datetime, date, timedelta

now = datetime.now()
print("Current date/time:", now)
print("Date:", now.strftime("%d-%m-%Y"))
print("Time:", now.strftime("%I:%M %p"))
print("Day:", now.strftime("%A"))

# Date arithmetic
today = date.today()
future = today + timedelta(days=30)
print(f"\nToday: {today}")
print(f"30 days later: {future}")
```

```python
# os module — Interact with operating system
import os

print("Current directory:", os.getcwd())
print("Files here:", os.listdir(".")[:5])   # First 5 files
print("OS name:", os.name)
```
