## 📌 Make your own reusable Python files

---

````python
# Step 1: Create a module file
# Let's create a helper module called 'my_math.py'

module_code = '''
"""My custom math module"""

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero!"
    return a / b

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

PI = 3.14159
'''

# Save it as a .py file
with open("my_math.py", "w") as f:
    f.write(module_code)

print("✅ Module 'my_math.py' created!")
````

````python
# Step 2: Import and use your module
import my_math

print("5 + 3 =", my_math.add(5, 3))
print("10 - 4 =", my_math.subtract(10, 4))
print("6 × 7 =", my_math.multiply(6, 7))
print("15 / 4 =", my_math.divide(15, 4))
print("5! =", my_math.factorial(5))
print("PI =", my_math.PI)
````

````python
# Import specific functions
from my_math import add, multiply, PI

print("Quick add:", add(100, 200))
print("Quick multiply:", multiply(5, 5))
print("PI value:", PI)
````

## 📦 What is a Package?

A **package** is a folder containing multiple modules with a special `__init__.py` file.

```
my_package/
├── __init__.py     ← Makes it a package
├── math_tools.py   ← Module 1
├── string_tools.py ← Module 2
└── file_tools.py   ← Module 3
```

Usage:
```python
from my_package import math_tools
from my_package.string_tools import clean_text
```
