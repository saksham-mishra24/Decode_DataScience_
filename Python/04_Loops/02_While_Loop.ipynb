## 🤔 What is a While Loop?

A **while loop** keeps running **as long as a condition is True**.

### Real-Life Analogy: Waiting for the Bus 🚌

**WHILE** the bus hasn't arrived:
- Keep waiting
- Check if the bus came

When the bus arrives → Stop waiting!

### Syntax:
```python
while condition:
    do_something
    update_condition  # Important! Or it loops forever!
```

### For vs While:

| For Loop | While Loop |
|---|---|
| Know how many times to repeat | Don't know how many times |
| Goes through a sequence | Runs until condition becomes False |
| "Do this 10 times" | "Do this until I say stop" |

---

````python
# Counting 1 to 5 with while
count = 1

while count <= 5:
    print("Count:", count)
    count += 1       # Very important! Without this → infinite loop!

print("Loop finished!")
````

````python
# Countdown timer
seconds = 5

print("Countdown starting!")
while seconds > 0:
    print(seconds, "...")
    seconds -= 1

print("🚀 BLAST OFF!")
````

````python
# Keep asking until correct password
correct_password = "python123"

password = ""
attempts = 0

while password != correct_password:
    password = input("Enter password: ")
    attempts += 1
    
    if password != correct_password:
        print("❌ Wrong! Try again.")

print(f"✅ Login successful after {attempts} attempt(s)!")
````

````python
# Menu-driven program
print("=== Simple Menu ===")

choice = ""
while choice != "3":
    print("\n1. Say Hello")
    print("2. Show Date")
    print("3. Exit")
    choice = input("Enter choice: ")
    
    if choice == "1":
        print("Hello! 👋")
    elif choice == "2":
        import datetime
        print("Today:", datetime.date.today())
    elif choice == "3":
        print("Goodbye! 👋")
    else:
        print("Invalid choice!")
````

## ⚠️ Avoiding Infinite Loops

An **infinite loop** runs forever and never stops! Always make sure the condition eventually becomes False.

```python
# ❌ INFINITE LOOP (DON'T RUN THIS!)
# while True:
#     print("This never stops!")

# ✅ CORRECT - condition changes
count = 0
while count < 5:
    print(count)
    count += 1    # This makes count eventually reach 5
```

**If you accidentally create an infinite loop:** Press `Ctrl + C` to stop it, or click the Stop button (⬛) in Jupyter.

## 🌾 Real-Life Examples

````python
# Sum numbers until user enters 0
print("Enter numbers to add. Type 0 to stop.")

total = 0
count = 0
number = -1   # Start with non-zero value

while number != 0:
    number = float(input("Enter a number (0 to stop): "))
    if number != 0:
        total += number
        count += 1

print(f"\nYou entered {count} numbers.")
print(f"Total sum: {total}")
if count > 0:
    print(f"Average: {total / count}")
````

## ✅ Answer

````python
secret = 7
guess = 0
attempts = 0

print("Guess the number (1-10)!")
while guess != secret:
    guess = int(input("Your guess: "))
    attempts += 1
    if guess < secret:
        print("Too low! ⬆️")
    elif guess > secret:
        print("Too high! ⬇️")
    else:
        print(f"🎉 Correct! You got it in {attempts} tries!")
````
