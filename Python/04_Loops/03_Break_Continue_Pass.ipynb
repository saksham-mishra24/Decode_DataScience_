# 🛑 Break, Continue, and Pass

---

## 📌 What You Will Learn

- `break` — Exit the loop immediately
- `continue` — Skip current iteration, go to next
- `pass` — Do nothing (placeholder)

---

## 1️⃣ Break — Stop the Loop! 🛑

`break` **immediately exits** the loop, even if it hasn't finished.

### Analogy: Finding Your Lost Phone 📱
You search each room one by one. When you FIND it, you STOP searching (break!).

```python
# Break example: Search for a number
numbers = [4, 7, 2, 9, 5, 1, 8, 3]
target = 9

print(f"Searching for {target}...")
for num in numbers:
    print(f"  Checking {num}...")
    if num == target:
        print(f"  ✅ Found {target}! Stopping search.")
        break  # Exit the loop immediately
```

```python
# Break with while: Exit on command
print("Type 'quit' to exit")

while True:  # This would be infinite without break
    text = input("Enter something: ")
    if text == "quit":
        print("Goodbye! 👋")
        break
    print("You said:", text)
```

## 2️⃣ Continue — Skip This One! ⏭️

`continue` **skips** the current iteration and goes to the next one.

### Analogy: Skip Rotten Mangoes 🥭
You're sorting mangoes. If one is rotten → skip it, move to next.

```python
# Skip even numbers, print only odd
print("Odd numbers from 1 to 10:")

for i in range(1, 11):
    if i % 2 == 0:  # If even
        continue     # Skip to next iteration
    print(i, end=" ")
```

```python
# Skip students who are absent
students = ["Rahul", "Priya", "ABSENT", "Amit", "ABSENT", "Sunita"]

print("=== Present Students ===")
for student in students:
    if student == "ABSENT":
        continue   # Skip absent entries
    print("✅", student)
```

## 3️⃣ Pass — Do Nothing (Placeholder) 📝

`pass` does **absolutely nothing**. It's used as a placeholder when you need code syntactically but haven't written it yet.

```python
# Pass as placeholder
for i in range(5):
    if i == 3:
        pass   # TODO: Handle special case later
    print(i)

# Without pass, an empty if block would give an error!
# if True:
#     (nothing here → Error!)

if True:
    pass   # This is OK — Python accepts pass as a placeholder
```

## 📊 Comparison

| Keyword | What It Does | Analogy |
|---|---|---|
| `break` | Exits the entire loop | Finding your phone — stop searching |
| `continue` | Skips ONE iteration | Skipping rotten mango — check next |
| `pass` | Does nothing | Empty placeholder — fill in later |

## ✏️ Practice Exercise

Print numbers 1 to 20, but skip multiples of 3, and stop completely if you reach 15.

```python
# Write your code here
```

---

## ✅ Answer

```python
for i in range(1, 21):
    if i == 15:
        print("\nReached 15! Stopping.")
        break
    if i % 3 == 0:
        continue   # Skip multiples of 3
    print(i, end=" ")
```
