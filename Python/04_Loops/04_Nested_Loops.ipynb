## 🤔 What is a Nested Loop?

A **nested loop** is a loop inside another loop.

### Real-Life Analogy: Classroom Rows and Columns 🏫

A classroom has 3 rows and 4 columns of desks:

```
Row 1:  [Desk] [Desk] [Desk] [Desk]
Row 2:  [Desk] [Desk] [Desk] [Desk]
Row 3:  [Desk] [Desk] [Desk] [Desk]
```

To check every desk: **For each row → check each column**. That's a nested loop!

---

````python
# Simple nested loop
for row in range(1, 4):      # Outer loop: 3 rows
    for col in range(1, 5):  # Inner loop: 4 columns
        print(f"R{row}C{col}", end="  ")
    print()   # New line after each row
````

````python
# Multiplication tables 1-5
for table in range(1, 6):
    print(f"\n=== Table of {table} ===")
    for i in range(1, 11):
        print(f"{table} × {i} = {table * i}")
````

````python
# Star rectangle
rows = 4
cols = 8

for i in range(rows):
    for j in range(cols):
        print("* ", end="")
    print()
````

## ✏️ Practice: Star Triangle

````python
# Right-angled triangle
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print("* ", end="")
    print()
````
