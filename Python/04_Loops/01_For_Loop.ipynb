## 🤔 Why Do We Need Loops?

Imagine you want to print "Hello" 5 times. Without loops:

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

Now imagine printing it 1000 times! 😱 That's where **loops** save us.

### Real-Life Analogy: Roll Call in Class 📋

A teacher calls out each student's name one by one from the register:
- "Rahul" → Present!
- "Priya" → Present!
- "Amit" → Present!
- ... (continues for ALL students)

A **for loop** does this — it goes through each item **one by one** and does something with it.

## 💻 Basic For Loop

### Syntax:
```python
for variable in sequence:
    do_something
```

````python
# Looping through a list
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]

for fruit in fruits:
    print("I like", fruit, "🍎")
````

````python
# Roll call example
students = ["Rahul", "Priya", "Amit", "Sunita", "Ravi"]

print("=== Roll Call ===")
for student in students:
    print(student, "→ Present! ✅")
````

## 🔢 Using range() Function

`range()` generates a sequence of numbers. Very useful with for loops!

| Syntax | Generates | Example |
|---|---|---|
| `range(n)` | 0 to n-1 | `range(5)` → 0,1,2,3,4 |
| `range(start, end)` | start to end-1 | `range(1,6)` → 1,2,3,4,5 |
| `range(start, end, step)` | start to end-1, skip by step | `range(0,10,2)` → 0,2,4,6,8 |

````python
# Counting from 1 to 10
print("Counting 1 to 10:")
for i in range(1, 11):
    print(i, end=" ")
print()  # New line

# Even numbers from 2 to 20
print("\nEven numbers 2-20:")
for i in range(2, 21, 2):
    print(i, end=" ")
print()

# Countdown!
print("\nCountdown:")
for i in range(10, 0, -1):
    print(i, end=" ")
print("🚀 Launch!")
````

````python
# Print "Hello" 5 times
for i in range(5):
    print("Hello!", "(time", i+1, ")")
````

````python
# Multiplication table
num = int(input("Enter a number for multiplication table: "))

print(f"\n=== Table of {num} ===")
for i in range(1, 11):
    print(f"{num} × {i} = {num * i}")
````

## 📝 Looping Through Strings

````python
# Going through each character
name = "PYTHON"

for letter in name:
    print(letter, "🔤")
````

## 🧮 For Loop with Calculations

````python
# Sum of numbers 1 to 100
total = 0
for i in range(1, 101):
    total += i
print("Sum of 1 to 100 =", total)

# Factorial of a number (5! = 5×4×3×2×1)
n = 5
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"\n{n}! = {factorial}")
````

````python
# Average marks calculator
print("=== Average Marks Calculator ===")
num_subjects = int(input("How many subjects? "))

total_marks = 0
for i in range(1, num_subjects + 1):
    marks = int(input(f"Enter marks for subject {i}: "))
    total_marks += marks

average = total_marks / num_subjects
print(f"\nTotal: {total_marks}")
print(f"Average: {round(average, 2)}")
````

## 📊 Summary

| Syntax | What It Does |
|---|---|
| `for x in list:` | Goes through each item in a list |
| `for x in range(n):` | Counts from 0 to n-1 |
| `for x in range(a,b):` | Counts from a to b-1 |
| `for x in string:` | Goes through each character |
