## 🤔 Why Do We Need Elif?

`if-else` gives us only 2 choices. But sometimes we need MORE than 2!

### Real-Life Analogy: Grade System 📝

- Marks >= 90 → Grade A
- Marks >= 80 → Grade B
- Marks >= 70 → Grade C
- Marks >= 60 → Grade D
- Marks < 60 → Fail

That's 5 different conditions! We need `elif` (short for "else if").

### Syntax:
```python
if condition1:
    action1
elif condition2:
    action2
elif condition3:
    action3
else:
    default_action
```

---

````python
# Grade System
marks = int(input("Enter your marks (out of 100): "))

if marks >= 90:
    print("Grade: A+ 🌟 Excellent!")
elif marks >= 80:
    print("Grade: A 😊 Very Good!")
elif marks >= 70:
    print("Grade: B 👍 Good!")
elif marks >= 60:
    print("Grade: C 📚 Average")
elif marks >= 33:
    print("Grade: D ✅ Passed")
else:
    print("Grade: F ❌ Failed")
````

````python
# Day of the week
day = int(input("Enter day number (1-7): "))

if day == 1:
    print("Monday")
elif day == 2:
    print("Tuesday")
elif day == 3:
    print("Wednesday")
elif day == 4:
    print("Thursday")
elif day == 5:
    print("Friday")
elif day == 6:
    print("Saturday 🎉")
elif day == 7:
    print("Sunday 🎉")
else:
    print("Invalid day! Enter 1-7.")
````

````python
# Simple Menu-Based Calculator
print("=== 🧮 Calculator ===")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("\nEnter your choice (1-4): "))
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == 1:
    print("Result:", num1, "+", num2, "=", num1 + num2)
elif choice == 2:
    print("Result:", num1, "-", num2, "=", num1 - num2)
elif choice == 3:
    print("Result:", num1, "×", num2, "=", num1 * num2)
elif choice == 4:
    if num2 != 0:
        print("Result:", num1, "÷", num2, "=", round(num1 / num2, 2))
    else:
        print("Error: Cannot divide by zero!")
else:
    print("Invalid choice! Please enter 1-4.")
````

---

## 📊 Summary

| Structure | When to Use |
|---|---|
| `if` only | When you have 1 condition |
| `if-else` | When you have 2 choices |
| `if-elif-else` | When you have 3 or more choices |

**Key Points:**
- You can have as many `elif` blocks as needed
- `else` is optional but recommended as a "catch-all"
- Only ONE block runs — Python stops checking after finding a True condition

Calculate electricity bill based on units consumed:
- First 100 units: ₹5 per unit
- Next 100 units (101-200): ₹7 per unit
- Above 200 units: ₹10 per unit

## ✅ Answer

````python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Units:", units)
print("Bill: ₹", bill)
````
