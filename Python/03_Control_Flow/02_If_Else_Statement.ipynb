## 🤔 Why Do We Need Else?

With just `if`, we can do something when a condition is True. But what about when it's False?

### Real-Life Analogy:

- **IF** it rains → Take an umbrella ☔
- **ELSE** → Wear sunglasses 🕶️

### Syntax:
```python
if condition:
    do_this          # Runs if condition is True
else:
    do_that_instead  # Runs if condition is False
```

---

````python
# Pass or Fail
marks = int(input("Enter your marks: "))

if marks >= 33:
    print("🎉 You PASSED!")
else:
    print("😔 You FAILED. Try harder next time!")
````

````python
# Even or Odd
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is EVEN")
else:
    print(number, "is ODD")
````

````python
# Positive or Negative
num = float(input("Enter a number: "))

if num >= 0:
    print(num, "is positive (or zero) ✅")
else:
    print(num, "is negative ➖")
````

````python
# Age-based ticket pricing
print("=== 🎬 Movie Ticket System ===")
age = int(input("Enter your age: "))

if age <= 12:
    price = 100
    print("Child ticket: ₹", price)
else:
    price = 250
    print("Adult ticket: ₹", price)
````

````python
# Login system
correct_password = "python123"
entered_password = input("Enter password: ")

if entered_password == correct_password:
    print("✅ Login successful! Welcome!")
else:
    print("❌ Wrong password! Access denied.")
````

````python
# Leap year checker
year = int(input("Enter a year: "))

if year % 4 == 0:
    print(year, "is a Leap Year! 📅")
else:
    print(year, "is NOT a Leap Year")
````

---

## 📊 Summary

```
if condition:          ← Check the condition
    action_if_true     ← Runs when condition is True
else:                  ← Otherwise
    action_if_false    ← Runs when condition is False
```

**Key Point:** `if` and `else` are **always partners** — only ONE of them runs, never both!
