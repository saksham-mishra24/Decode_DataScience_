## 🤔 What is the Ternary Operator?

The ternary operator is a **shortcut** way to write a simple if-else in just ONE line.

### Syntax:
```python
result = value_if_true if condition else value_if_false
```

### Comparison:

**Normal if-else (3+ lines):**
```python
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
```

**Ternary (1 line):**
```python
status = "Adult" if age >= 18 else "Minor"
```

Both do the SAME thing! The ternary is just shorter.

---

````python
# Example 1: Adult or Minor
age = 20
status = "Adult" if age >= 18 else "Minor"
print("Age:", age, "→ Status:", status)

# Example 2: Even or Odd
num = 15
result = "Even" if num % 2 == 0 else "Odd"
print(num, "is", result)

# Example 3: Pass or Fail
marks = 45
verdict = "PASS ✅" if marks >= 33 else "FAIL ❌"
print("Marks:", marks, "→", verdict)
````

````python
# Example 4: Max of two numbers
a = 10
b = 25
maximum = a if a > b else b
print("Max of", a, "and", b, "is", maximum)

# Example 5: Absolute value
num = -7
absolute = num if num >= 0 else -num
print("Absolute value of", num, "is", absolute)

# Example 6: Directly in print
temperature = 42
print("Weather is", "Hot 🔥" if temperature > 35 else "Normal 😊")
````

---

## ⚠️ When to Use vs When NOT to Use

| ✅ Use Ternary When | ❌ Don't Use When |
|---|---|
| Simple True/False choice | Complex logic with multiple elif |
| Assigning a value based on condition | Multiple actions needed |
| Short and readable | Would make code confusing |
