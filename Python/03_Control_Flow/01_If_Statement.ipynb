## 🤔 What is an If Statement?

An **if statement** lets Python **make decisions** — do something ONLY IF a condition is true.

### Real-Life Analogy: Traffic Light 🚦

- **IF** the light is GREEN → Go
- **IF** the light is RED → Stop

You make decisions based on conditions every day! Python can do the same.

### Syntax:
```python
if condition:
    do_something    # This runs ONLY if condition is True
```

**⚠️ Important:** Notice the **colon (:)** after the condition and the **4 spaces** (indentation) before the action!

---

````python
# Simple if statement
age = 20

if age >= 18:
    print("You are an adult! 🎉")
    print("You can vote!")

print("This line ALWAYS runs (it's outside the if)")
````

````python
# When condition is False — the indented code is SKIPPED
age = 15

if age >= 18:
    print("You are an adult!")    # This is SKIPPED
    print("You can vote!")         # This is also SKIPPED

print("Program continues here")   # This ALWAYS runs
````

---

## 📏 Understanding Indentation

In Python, **indentation** (spaces at the beginning of a line) is very important!

- Lines with the **same indentation** belong to the same block
- Standard indentation is **4 spaces** (press Tab key)

```python
if condition:
    line 1    ← inside if (4 spaces)
    line 2    ← inside if (4 spaces)
line 3        ← OUTSIDE if (no spaces) — always runs
```

````python
# Indentation matters!
temperature = 45

if temperature > 40:
    print("It's very hot! 🌡️")       # Inside if
    print("Stay indoors!")              # Inside if
    print("Drink plenty of water! 💧")  # Inside if

print("Have a nice day!")  # OUTSIDE if — always runs
````

## 🌾 More Real-Life Examples

````python
# Example 1: Check if student passed
marks = int(input("Enter your marks (out of 100): "))

if marks >= 33:
    print("🎉 Congratulations! You PASSED!")

if marks >= 90:
    print("🌟 Outstanding! You got a distinction!")

if marks == 100:
    print("🏆 PERFECT SCORE! Amazing!")
````

````python
# Example 2: Check if number is positive
number = int(input("Enter a number: "))

if number > 0:
    print(number, "is a POSITIVE number ✅")

if number < 0:
    print(number, "is a NEGATIVE number ➖")

if number == 0:
    print("The number is ZERO")
````

````python
# Example 3: Even or Odd
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is EVEN")

if num % 2 != 0:
    print(num, "is ODD")
````

## ⚠️ Common Mistakes

| Mistake | What Happens | Fix |
|---|---|---|
| Forgetting the colon `:` | SyntaxError | `if age >= 18:` (add colon) |
| Wrong indentation | IndentationError | Use exactly 4 spaces |
| Using `=` instead of `==` | Assigns instead of comparing | Use `==` for comparison |
