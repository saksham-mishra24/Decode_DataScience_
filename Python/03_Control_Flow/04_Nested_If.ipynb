## 🤔 What is Nested If?

A **nested if** is an if statement **inside another** if statement.

### Real-Life Analogy: Going to the Movie 🎬

```
IF you have money:
    IF the movie is good:
        Go watch the movie! 🎬
    ELSE:
        Stay home and watch TV 📺
ELSE:
    Stay home (no money!) 💰
```

The second decision depends on the first one being true.

---

````python
# Example 1: ATM Withdrawal
print("=== 🏧 ATM Machine ===")

pin = input("Enter your PIN: ")
correct_pin = "1234"

if pin == correct_pin:
    print("✅ PIN correct!")
    
    balance = 5000
    amount = int(input("Enter amount to withdraw: ₹"))
    
    if amount <= balance:
        if amount > 0:
            balance -= amount
            print("💵 Please collect ₹", amount)
            print("Remaining balance: ₹", balance)
        else:
            print("❌ Invalid amount!")
    else:
        print("❌ Insufficient balance! You have only ₹", balance)
else:
    print("❌ Wrong PIN! Access denied.")
````

````python
# Example 2: Exam Result with Division
marks = int(input("Enter total marks (out of 500): "))
percentage = (marks / 500) * 100

if percentage >= 33:  # Passed
    print("✅ PASSED!")
    
    if percentage >= 60:
        print("🥇 First Division!")
    elif percentage >= 45:
        print("🥈 Second Division")
    else:
        print("🥉 Third Division")
    
    print("Percentage:", round(percentage, 2), "%")
else:
    print("❌ FAILED")
    print("Percentage:", round(percentage, 2), "%")
    print("You needed", 33 - percentage, "% more to pass.")
````

````python
# Example 3: Theme Park Entry
print("=== 🎢 Theme Park Entry ===")

age = int(input("Enter your age: "))
height = float(input("Enter your height in cm: "))

if age >= 5:
    print("✅ Age requirement met!")
    
    if height >= 120:
        print("✅ Height requirement met!")
        print("🎢 You can enjoy ALL rides!")
    else:
        print("⚠️ You can enjoy only kids' rides.")
        print("You need", 120 - height, "cm more for big rides.")
else:
    print("❌ Sorry! Minimum age is 5 years.")
````

---

## ⚠️ Tips for Nested If

1. **Don't nest too deep** — 3 levels is usually the maximum
2. **Consider using elif** instead of nesting when possible
3. **Keep indentation correct** — each level adds 4 more spaces
