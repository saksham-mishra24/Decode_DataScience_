## 📌 Writing and appending text files

---

```python
# Write mode ('w') — Creates new file or OVERWRITES existing
with open("my_file.txt", "w") as f:
    f.write("This is my first line.\n")
    f.write("This is my second line.\n")
    f.write("Python makes file handling easy!\n")

print("✅ File written!")

# Read to verify
with open("my_file.txt", "r") as f:
    print(f.read())
```

```python
# Append mode ('a') — Add to END of file (doesn't overwrite!)
with open("my_file.txt", "a") as f:
    f.write("This line was APPENDED.\n")
    f.write("And this one too!\n")

print("✅ Lines appended!")
with open("my_file.txt", "r") as f:
    print(f.read())
```

```python
# Write a list of lines using writelines()
students = ["Rahul - 85\n", "Priya - 92\n", "Amit - 67\n", "Sunita - 78\n"]

with open("students.txt", "w") as f:
    f.write("=== Student Marks ===\n")
    f.writelines(students)

print("✅ Student file created!")
with open("students.txt", "r") as f:
    print(f.read())
```

```python
# Real-life: Save user input to file
import os

print("=== 📝 Diary Entry ===")
date = input("Enter today's date: ")
entry = input("Write your diary entry: ")

with open("diary.txt", "a") as f:
    f.write(f"\n📅 {date}\n")
    f.write(f"{entry}\n")
    f.write("-" * 30 + "\n")

print("✅ Diary entry saved!")

# Read all entries
if os.path.exists("diary.txt"):
    with open("diary.txt", "r") as f:
        print("\n=== All Diary Entries ===")
        print(f.read())
```

---

## ⚠️ Important Notes
- `'w'` mode **deletes** existing content! Be careful!
- `'a'` mode **preserves** existing content and adds to end
- Always use `with` — it handles closing automatically
