# 📊 Working with CSV Files

---

## 📌 CSV = Comma Separated Values (like Excel!)

CSV files store data in rows and columns, separated by commas.

---

```python
import csv

# Writing a CSV file
students = [
    ["Name", "Age", "Marks", "City"],
    ["Rahul", 20, 85, "Pune"],
    ["Priya", 19, 92, "Mumbai"],
    ["Amit", 21, 67, "Delhi"],
    ["Sunita", 20, 78, "Jaipur"]
]

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(students)

print("✅ CSV file created!")
```

```python
# Reading a CSV file
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
# Reading as dictionary (with headers)
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    print("=== Student Records ===")
    for row in reader:
        print(f"Name: {row['Name']}, Marks: {row['Marks']}, City: {row['City']}")
```

```python
# Writing with DictWriter
students = [
    {"name": "Ravi", "marks": 88, "grade": "A"},
    {"name": "Meena", "marks": 95, "grade": "A+"},
    {"name": "Suresh", "marks": 72, "grade": "B"}
]

with open("grades.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "marks", "grade"])
    writer.writeheader()     # Write header row
    writer.writerows(students)

print("✅ Grades CSV created!")

# Read and display
with open("grades.csv", "r") as f:
    print(f.read())
```
