# 📋 Working with JSON Files

---

## 📌 JSON = JavaScript Object Notation (structured data)

JSON looks like Python dictionaries! Used for APIs, configs, data exchange.

---

```python
import json

# Writing JSON
student = {
    "name": "Rahul Kumar",
    "age": 20,
    "city": "Pune",
    "marks": {"Hindi": 78, "English": 85, "Math": 92},
    "hobbies": ["Reading", "Cricket", "Coding"]
}

# Save to file
with open("student.json", "w") as f:
    json.dump(student, f, indent=4)  # indent for readability

print("✅ JSON file saved!")
```

```python
# Reading JSON
with open("student.json", "r") as f:
    data = json.load(f)

print("Name:", data["name"])
print("Age:", data["age"])
print("Math marks:", data["marks"]["Math"])
print("Hobbies:", data["hobbies"])
```

```python
# Convert Python → JSON string
python_dict = {"name": "Priya", "age": 19}
json_string = json.dumps(python_dict, indent=2)
print("JSON string:")
print(json_string)

# Convert JSON string → Python
json_text = '{"name": "Amit", "age": 21}'
python_obj = json.loads(json_text)
print("\nPython dict:", python_obj)
print("Name:", python_obj["name"])
```

```python
# Save multiple records
classroom = [
    {"name": "Rahul", "marks": 85},
    {"name": "Priya", "marks": 92},
    {"name": "Amit", "marks": 67}
]

with open("classroom.json", "w") as f:
    json.dump(classroom, f, indent=2)

# Read and process
with open("classroom.json", "r") as f:
    students = json.load(f)

print("=== Class Toppers ===")
for s in sorted(students, key=lambda x: x['marks'], reverse=True):
    print(f"{s['name']}: {s['marks']} marks")
```

---

## 📊 JSON Functions Summary

| Function | Purpose |
|---|---|
| `json.dump(data, file)` | Write Python → JSON file |
| `json.load(file)` | Read JSON file → Python |
| `json.dumps(data)` | Python → JSON string |
| `json.loads(string)` | JSON string → Python |
