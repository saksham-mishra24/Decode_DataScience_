## 🤔 Why File Handling?

So far, all data disappears when the program ends. Files let us **save data permanently** — like writing in a notebook 📓 that you can read later.

---

```python
# First, let's create a sample file to read
with open("sample.txt", "w") as f:
    f.write("Hello, this is line 1.\n")
    f.write("Python is amazing!\n")
    f.write("File handling is easy.\n")
    f.write("This is the last line.\n")
print("✅ Sample file created!")
```

```python
# Method 1: read() — Read ENTIRE file as one string
with open("sample.txt", "r") as f:
    content = f.read()
print("=== Full Content ===")
print(content)
```

```python
# Method 2: readline() — Read ONE line at a time
with open("sample.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()
print("Line 1:", line1.strip())   # strip() removes \n
print("Line 2:", line2.strip())
```

```python
# Method 3: readlines() — Read ALL lines as a list
with open("sample.txt", "r") as f:
    lines = f.readlines()
print("Lines as list:", lines)

# Loop through lines
print("\n=== Line by Line ===")
for i, line in enumerate(lines, 1):
    print(f"Line {i}: {line.strip()}")
```

```python
# Method 4: Loop directly (most Pythonic!)
print("=== Direct Loop ===")
with open("sample.txt", "r") as f:
    for line in f:
        print("📄", line.strip())
```

---

## 📊 File Opening Modes

| Mode | Meaning |
|---|---|
| `'r'` | Read (default) |
| `'w'` | Write (overwrites!) |
| `'a'` | Append (add to end) |
| `'x'` | Create new file |

**Always use `with` statement** — it automatically closes the file!
