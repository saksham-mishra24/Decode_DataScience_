# 🎀 Decorators

---

## 📌 Functions that modify other functions

### Analogy: Gift Wrapping 🎁
A decorator **wraps** a function with extra functionality, like wrapping a gift with colorful paper.

---

```python
# A simple decorator
def greeting_decorator(func):
    def wrapper():
        print("🎀 Before the function")
        func()    # Call the original function
        print("🎀 After the function")
    return wrapper

@greeting_decorator     # Apply the decorator
def say_hello():
    print("Hello World!")

say_hello()
```

```python
# Practical: Timer decorator
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"⏱️ {func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    total = 0
    for i in range(1000000):
        total += i
    return total

result = slow_function()
print(f"Result: {result}")
```

```python
# Practical: Logger decorator
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"📝 Calling {func.__name__} with args={args}")
        result = func(*args, **kwargs)
        print(f"📝 {func.__name__} returned {result}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

@logger
def multiply(a, b):
    return a * b

add(5, 3)
print()
multiply(4, 7)
```
