## 🔄 Iterators

An **iterator** is an object that produces one value at a time. Every time you use a `for` loop, you're using an iterator!

---

```python
# Manual iterator
numbers = [10, 20, 30]
my_iter = iter(numbers)    # Create iterator

print(next(my_iter))   # 10
print(next(my_iter))   # 20
print(next(my_iter))   # 30
# print(next(my_iter))   # StopIteration error!
```

## ⚡ Generators

A **generator** is a special function that uses `yield` instead of `return`. It produces values one at a time — very memory efficient!

```python
# Generator function
def count_up_to(n):
    """Generate numbers from 1 to n"""
    i = 1
    while i <= n:
        yield i     # Produces value and PAUSES
        i += 1

# Use the generator
for num in count_up_to(5):
    print(num, end=" ")
print()
```

```python
# Generator for even numbers
def even_numbers(limit):
    n = 0
    while n <= limit:
        yield n
        n += 2

print("Even numbers up to 20:")
for num in even_numbers(20):
    print(num, end=" ")
print()
```

```python
# Generator expression (like list comprehension but with ())
squares_list = [x**2 for x in range(10)]   # List — stores all in memory
squares_gen = (x**2 for x in range(10))    # Generator — creates one at a time

print("List:", squares_list)
print("Generator:", squares_gen)   # Just shows the generator object
print("Generator values:", list(squares_gen))  # Convert to see values
```

---

## 📊 List vs Generator

| Feature | List | Generator |
|---|---|---|
| Memory | Stores all at once | One at a time |
| Speed | Fast access | Lazy (computes when needed) |
| Syntax | `[x for x in ...]` | `(x for x in ...)` |
| Reusable | ✅ Yes | ❌ One-time only |
| Best for | Small data | Large/infinite data |
