## 1️⃣ Variables and Data Types

```python
# Variables
name = "Rahul"         # str (string)
age = 20               # int (integer)
height = 5.7           # float (decimal)
is_student = True      # bool (boolean)

# Check type
type(name)  # <class 'str'>

# Type conversion
int("25")    # String → Integer
float(10)    # Integer → Float
str(100)     # Integer → String
```

## 2️⃣ Operators

```python
# Arithmetic
+ - * / // % **

# Comparison
== != > < >= <=

# Logical
and  or  not

# Assignment
= += -= *= /=
```

## 3️⃣ Strings

```python
s = "Hello World"

# Methods
s.upper()         # HELLO WORLD
s.lower()         # hello world
s.strip()         # Remove whitespace
s.split(" ")      # ['Hello', 'World']
s.replace("o","0") # Hell0 W0rld
s.find("World")   # 6
len(s)            # 11

# f-strings
f"My name is {name} and I am {age}"

# Slicing
s[0:5]   # Hello
s[-5:]   # World
s[::-1]  # dlroW olleH
```

## 4️⃣ Control Flow

```python
# If-elif-else
if condition:
    action1
elif condition2:
    action2
else:
    default

# Ternary
result = "Yes" if condition else "No"
```

## 5️⃣ Loops

```python
# For loop
for i in range(10):       # 0-9
for item in list:         # Each item
for i, item in enumerate(list):  # Index + item

# While loop
while condition:
    action

# Control
break       # Exit loop
continue    # Skip iteration
pass        # Do nothing
```

## 6️⃣ Data Structures

```python
# List (ordered, changeable)
my_list = [1, 2, 3]
my_list.append(4)      # Add
my_list.remove(2)      # Remove
my_list.sort()         # Sort
my_list[0]             # Access

# Tuple (ordered, unchangeable)
my_tuple = (1, 2, 3)
a, b, c = my_tuple     # Unpacking

# Set (unique, unordered)
my_set = {1, 2, 3}
set_a | set_b          # Union
set_a & set_b          # Intersection

# Dictionary (key-value)
my_dict = {"key": "value"}
my_dict["key"]         # Access
my_dict.get("key")     # Safe access
my_dict.items()        # All pairs

# List Comprehension
[x**2 for x in range(10)]
[x for x in list if x > 5]
```

## 7️⃣ Functions

```python
# Define
def my_func(param1, param2="default"):
    """Docstring"""
    return result

# Lambda
square = lambda x: x ** 2

# *args, **kwargs
def func(*args, **kwargs):
    pass
```

## 8️⃣ File Handling

```python
# Read
with open("file.txt", "r") as f:
    content = f.read()

# Write
with open("file.txt", "w") as f:
    f.write("Hello")

# Append
with open("file.txt", "a") as f:
    f.write("More text")

# CSV
import csv
# JSON
import json
```

## 🔟 OOP (Object-Oriented Programming)

```python
class MyClass:
    def __init__(self, name):
        self.name = name     # Attribute
    
    def greet(self):         # Method
        return f"Hello {self.name}"

# Inheritance
class Child(Parent):
    def __init__(self):
        super().__init__()

# Access Levels
self.name      # Public
self._name     # Protected
self.__name    # Private
```
