# 📒 Mini Project 4: Contact Book

---

## Uses: Dictionaries, File Handling, Functions, Error Handling

---

```python
import json
import os

FILENAME = "contacts.json"

def load_contacts():
    if os.path.exists(FILENAME):
        with open(FILENAME, 'r') as f:
            return json.load(f)
    return {}

def save_contacts(contacts):
    with open(FILENAME, 'w') as f:
        json.dump(contacts, f, indent=2)

def add_contact(contacts):
    name = input("Enter name: ").strip()
    phone = input("Enter phone: ").strip()
    email = input("Enter email (or press Enter to skip): ").strip()
    contacts[name] = {"phone": phone, "email": email or "N/A"}
    save_contacts(contacts)
    print(f"✅ {name} added successfully!")

def view_contacts(contacts):
    if not contacts:
        print("📒 Contact book is empty!")
        return
    print(f"\n{'='*50}")
    print(f"{'📒 CONTACT BOOK':^50}")
    print(f"{'='*50}")
    for i, (name, info) in enumerate(contacts.items(), 1):
        print(f"{i}. {name}")
        print(f"   📞 {info['phone']}")
        print(f"   📧 {info['email']}")
    print(f"{'='*50}")
    print(f"Total contacts: {len(contacts)}")

def search_contact(contacts):
    query = input("Search name: ").strip().lower()
    found = {k: v for k, v in contacts.items() if query in k.lower()}
    if found:
        for name, info in found.items():
            print(f"\n✅ {name}: 📞 {info['phone']} | 📧 {info['email']}")
    else:
        print("❌ No contacts found!")

def delete_contact(contacts):
    name = input("Enter name to delete: ").strip()
    if name in contacts:
        del contacts[name]
        save_contacts(contacts)
        print(f"✅ {name} deleted!")
    else:
        print(f"❌ {name} not found!")

def contact_book():
    contacts = load_contacts()
    
    while True:
        print("\n📒 CONTACT BOOK MENU")
        print("1. Add Contact")
        print("2. View All Contacts")
        print("3. Search Contact")
        print("4. Delete Contact")
        print("5. Exit")
        
        choice = input("\nChoice: ")
        
        if choice == '1': add_contact(contacts)
        elif choice == '2': view_contacts(contacts)
        elif choice == '3': search_contact(contacts)
        elif choice == '4': delete_contact(contacts)
        elif choice == '5':
            print("👋 Goodbye!")
            break
        else:
            print("❌ Invalid choice!")

contact_book()
```
