# 🎯 Mini Project 3: Number Guessing Game

## Uses: Loops, Conditionals, Random, Functions

---

```python
import random

def number_guessing_game():
    """A fun number guessing game!"""
    print("="*40)
    print("   🎯 NUMBER GUESSING GAME")
    print("="*40)
    
    # Choose difficulty
    print("\nChoose difficulty:")
    print("1. Easy   (1-50,  10 chances)")
    print("2. Medium (1-100, 7 chances)")
    print("3. Hard   (1-200, 5 chances)")
    
    diff = input("\nYour choice (1/2/3): ")
    
    if diff == '1':
        max_num, max_tries = 50, 10
    elif diff == '3':
        max_num, max_tries = 200, 5
    else:
        max_num, max_tries = 100, 7
    
    secret = random.randint(1, max_num)
    tries = 0
    
    print(f"\nI'm thinking of a number between 1 and {max_num}.")
    print(f"You have {max_tries} chances. Good luck! 🍀")
    
    while tries < max_tries:
        try:
            guess = int(input(f"\nChance {tries+1}/{max_tries} — Your guess: "))
            tries += 1
            
            if guess == secret:
                print(f"\n🎉🎉🎉 CORRECT! The number was {secret}!")
                print(f"You got it in {tries} tries!")
                
                if tries == 1:
                    print("🏆 UNBELIEVABLE! First try!")
                elif tries <= 3:
                    print("⭐ Excellent!")
                elif tries <= 5:
                    print("👍 Good job!")
                else:
                    print("😅 Close call!")
                return
            elif guess < secret:
                remaining = max_tries - tries
                print(f"⬆️ Too LOW! ({remaining} chances left)")
            else:
                remaining = max_tries - tries
                print(f"⬇️ Too HIGH! ({remaining} chances left)")
        except ValueError:
            print("❌ Please enter a valid number!")
    
    print(f"\n😔 Game Over! The number was {secret}. Better luck next time!")

# Play the game!
number_guessing_game()
```
