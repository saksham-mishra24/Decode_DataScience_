# 🧠 Mini Project 5: Python Quiz Game

## Uses: Lists, Dictionaries, Loops, Functions, Random

---

```python
import random

def python_quiz():
    questions = [
        {
            "question": "What is the output of print(2 ** 3)?",
            "options": ["6", "8", "9", "5"],
            "answer": "8"
        },
        {
            "question": "Which data type is immutable?",
            "options": ["List", "Dictionary", "Set", "Tuple"],
            "answer": "Tuple"
        },
        {
            "question": "What does len('Hello') return?",
            "options": ["4", "5", "6", "Error"],
            "answer": "5"
        },
        {
            "question": "Which keyword is used to define a function?",
            "options": ["function", "func", "def", "define"],
            "answer": "def"
        },
        {
            "question": "What is 10 // 3 in Python?",
            "options": ["3.33", "3", "4", "3.0"],
            "answer": "3"
        },
        {
            "question": "Which symbol is used for comments?",
            "options": ["//", "/*", "#", "--"],
            "answer": "#"
        },
        {
            "question": "What does 'append()' do to a list?",
            "options": ["Removes last item", "Adds item to end", "Sorts the list", "Reverses the list"],
            "answer": "Adds item to end"
        },
        {
            "question": "What is the correct file extension for Python?",
            "options": [".python", ".pt", ".py", ".pyt"],
            "answer": ".py"
        },
        {
            "question": "Which loop is best when you know the number of iterations?",
            "options": ["while", "do-while", "for", "repeat"],
            "answer": "for"
        },
        {
            "question": "What does 'break' do in a loop?",
            "options": ["Skips one iteration", "Exits the loop", "Pauses the loop", "Restarts the loop"],
            "answer": "Exits the loop"
        }
    ]
    
    random.shuffle(questions)
    score = 0
    total = len(questions)
    
    print("="*50)
    print(f"{'🧠 PYTHON QUIZ GAME':^50}")
    print("="*50)
    print(f"Total Questions: {total}\n")
    
    for i, q in enumerate(questions, 1):
        print(f"\nQ{i}. {q['question']}")
        shuffled_options = q['options'][:]
        random.shuffle(shuffled_options)
        for j, opt in enumerate(shuffled_options, 1):
            print(f"   {j}. {opt}")
        
        try:
            ans = int(input("Your answer (1-4): "))
            if 1 <= ans <= 4 and shuffled_options[ans-1] == q['answer']:
                print("✅ Correct!")
                score += 1
            else:
                print(f"❌ Wrong! Answer: {q['answer']}")
        except (ValueError, IndexError):
            print(f"❌ Invalid input! Answer: {q['answer']}")
    
    # Results
    percentage = (score / total) * 100
    print(f"\n{'='*50}")
    print(f"{'RESULTS':^50}")
    print(f"{'='*50}")
    print(f"Score: {score}/{total} ({percentage:.0f}%)")
    
    if percentage == 100: print("🏆 PERFECT SCORE! Python Master!")
    elif percentage >= 80: print("⭐ Excellent! Great knowledge!")
    elif percentage >= 60: print("👍 Good! Keep learning!")
    elif percentage >= 40: print("📚 Need more practice.")
    else: print("📖 Keep studying! You'll get better!")

python_quiz()
```
