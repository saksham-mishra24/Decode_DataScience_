# 📝 Mini Project 2: Student Report Card System

---

## Uses: Dictionaries, Loops, Functions, File Handling, OOP

---

```python
import json

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
        self.marks = {}
    
    def add_marks(self, subject, marks):
        self.marks[subject] = marks
    
    def get_total(self):
        return sum(self.marks.values())
    
    def get_percentage(self):
        if not self.marks:
            return 0
        return round(self.get_total() / len(self.marks), 2)
    
    def get_grade(self):
        p = self.get_percentage()
        if p >= 90: return 'A+'
        elif p >= 80: return 'A'
        elif p >= 70: return 'B'
        elif p >= 60: return 'C'
        elif p >= 33: return 'D'
        else: return 'F'
    
    def display_report(self):
        print(f"\n{'='*45}")
        print(f"{'STUDENT REPORT CARD':^45}")
        print(f"{'='*45}")
        print(f"Name:    {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"{'-'*45}")
        print(f"{'Subject':<20} {'Marks':>10} {'Status':>10}")
        print(f"{'-'*45}")
        
        for subject, marks in self.marks.items():
            status = '✅ Pass' if marks >= 33 else '❌ Fail'
            print(f"{subject:<20} {marks:>10} {status:>10}")
        
        print(f"{'-'*45}")
        print(f"{'Total':<20} {self.get_total():>10}")
        print(f"{'Percentage':<20} {self.get_percentage():>10}%")
        print(f"{'Grade':<20} {self.get_grade():>10}")
        print(f"{'Result':<20} {'PASS ✅' if self.get_grade() != 'F' else 'FAIL ❌':>10}")
        print(f"{'='*45}")
```

```python
# Create a student and add marks
student = Student("Rahul Kumar", "2024001")
student.add_marks("Hindi", 78)
student.add_marks("English", 85)
student.add_marks("Mathematics", 92)
student.add_marks("Science", 67)
student.add_marks("Social Studies", 75)

student.display_report()
```

```python
# Interactive version
def create_report_card():
    print("\n📝 CREATE STUDENT REPORT CARD")
    print("-" * 30)
    
    name = input("Enter student name: ")
    roll = input("Enter roll number: ")
    
    s = Student(name, roll)
    
    subjects = ["Hindi", "English", "Mathematics", "Science", "Social Studies"]
    
    for subject in subjects:
        while True:
            try:
                marks = int(input(f"Enter {subject} marks (0-100): "))
                if 0 <= marks <= 100:
                    s.add_marks(subject, marks)
                    break
                else:
                    print("Marks should be between 0 and 100!")
            except ValueError:
                print("Please enter a valid number!")
    
    s.display_report()
    
    # Save to file
    data = {
        "name": s.name, "roll_no": s.roll_no,
        "marks": s.marks, "total": s.get_total(),
        "percentage": s.get_percentage(), "grade": s.get_grade()
    }
    with open(f"report_{roll}.json", "w") as f:
        json.dump(data, f, indent=2)
    print(f"\n✅ Report saved to report_{roll}.json")

create_report_card()
```
