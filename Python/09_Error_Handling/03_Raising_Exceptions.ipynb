## 📌 Creating and raising your own errors

---

```python
# raise — Create your own error
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    if age > 150:
        raise ValueError("Age cannot be more than 150!")
    return f"Age {age} is valid ✅"

try:
    print(check_age(25))
    print(check_age(-5))   # This will raise an error
except ValueError as e:
    print(f"❌ Error: {e}")
```

```python
# Custom Exception class
class InsufficientBalanceError(Exception):
    """Raised when withdrawal amount exceeds balance"""
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(
            f"Cannot withdraw ₹{amount}. Balance is only ₹{balance}."
        )
    return balance - amount

# Test it
balance = 5000
try:
    balance = withdraw(balance, 3000)
    print(f"✅ Withdrawal successful. Balance: ₹{balance}")
    balance = withdraw(balance, 5000)  # This will fail
except InsufficientBalanceError as e:
    print(f"❌ {e}")
```
