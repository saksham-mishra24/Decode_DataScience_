## 📌 Types of Errors in Python

### 1️⃣ Syntax Errors — Mistakes in writing code

```python
# Syntax errors — Python can't even run the code
# These are like spelling/grammar mistakes

# ❌ Missing colon
# if True
#     print("Hello")

# ✅ Correct
if True:
    print("Hello")
```

### 2️⃣ Runtime Errors — Errors that happen while running

```python
# Common runtime errors:

# ZeroDivisionError
# print(10 / 0)   # Can't divide by zero!

# NameError
# print(xyz)      # Variable 'xyz' doesn't exist!

# TypeError
# print("age: " + 25)   # Can't add string + number!

# IndexError
# my_list = [1, 2, 3]
# print(my_list[10])     # Index 10 doesn't exist!

# ValueError
# int("hello")    # Can't convert 'hello' to integer!

# FileNotFoundError
# open("nonexistent_file.txt")  # File doesn't exist!

print("Common errors listed above (commented out to prevent crashes)")
```

### 3️⃣ Logical Errors — Code runs but gives wrong result

```python
# Logical error example
# Intent: Calculate average of 3 numbers

a, b, c = 10, 20, 30

# ❌ Wrong (missing parentheses — divides only c)
wrong_avg = a + b + c / 3
print("Wrong average:", wrong_avg)   # 40.0 (wrong!)

# ✅ Correct
right_avg = (a + b + c) / 3
print("Right average:", right_avg)   # 20.0 (correct!)
```

## 📊 Common Error Types

| Error | When It Happens | Example |
|---|---|---|
| `SyntaxError` | Code written wrong | Missing colon, quotes |
| `NameError` | Variable not defined | Using `x` before creating it |
| `TypeError` | Wrong type operation | `"hello" + 5` |
| `ValueError` | Wrong value | `int("abc")` |
| `ZeroDivisionError` | Divide by zero | `10 / 0` |
| `IndexError` | Index out of range | `list[100]` |
| `KeyError` | Key not in dict | `dict["missing"]` |
| `FileNotFoundError` | File doesn't exist | `open("missing.txt")` |
