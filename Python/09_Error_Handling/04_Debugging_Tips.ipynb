## 🐛 What is Debugging?

A **bug** is an error in your code. **Debugging** is finding and fixing it.

### Tip 1: Read the Error Message Carefully!
Error messages tell you exactly WHERE and WHAT went wrong.

```
Traceback (most recent call last):
  File "example.py", line 5    ← WHERE the error is
    print(10 / 0)
ZeroDivisionError: division by zero  ← WHAT the error is
```

---

````python
# Tip 2: Use print() to inspect variables
def calculate_average(marks):
    print(f"DEBUG: marks = {marks}")          # Debug line
    print(f"DEBUG: length = {len(marks)}")     # Debug line
    
    total = sum(marks)
    print(f"DEBUG: total = {total}")           # Debug line
    
    average = total / len(marks)
    return average

result = calculate_average([78, 85, 92, 67])
print(f"\nFinal Average: {result}")
````

````python
# Tip 3: Use type() to check data types
x = "25"
print(f"x = {x}, type = {type(x)}")  # It's a string!
# This helps catch type-related bugs early
````

## 🔧 Common Debugging Checklist

| Problem | Check For |
|---|---|
| `SyntaxError` | Missing colons, brackets, quotes |
| `NameError` | Typo in variable name, or not defined yet |
| `TypeError` | Mixing types (string + int) |
| `IndexError` | Index too large for list |
| Wrong output | Check BODMAS, wrong variable used |
| Infinite loop | Check if loop condition ever becomes False |

### Golden Rules of Debugging:
1. 📖 **Read the error message** — it tells you the answer!
2. 🖨️ **Print variables** — see what values they hold
3. 🔍 **Check types** — is it really a number or a string?
4. 📝 **Comment out code** — isolate the problem
5. 🔄 **Test small pieces** — run parts separately
6. 🌐 **Google the error** — someone else had the same problem!
