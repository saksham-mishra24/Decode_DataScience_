## 📌 How to handle errors gracefully instead of crashing

### Analogy: Safety Net 🥅
A trapeze artist uses a safety net. If they fall (error), the net catches them (except) instead of hitting the ground (crash).

---

```python
# Without error handling — CRASHES!
# num = int(input("Enter number: "))  # If user types 'abc' → CRASH!

# With error handling — SAFE!
try:
    num = int(input("Enter a number: "))
    print(f"You entered: {num}")
except ValueError:
    print("❌ That's not a valid number! Please enter digits only.")
```

```python
# Handling specific errors
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b
    print(f"{a} / {b} = {result}")
except ValueError:
    print("❌ Please enter valid numbers!")
except ZeroDivisionError:
    print("❌ Cannot divide by zero!")
except Exception as e:
    print(f"❌ Something went wrong: {e}")
```

```python
# try-except-else-finally
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("❌ Invalid number!")
else:
    print(f"✅ Great! {num} × 2 = {num * 2}")  # Only runs if NO error
finally:
    print("🔚 This ALWAYS runs, error or not.")  # Always runs
```

```python
# Real-life: Safe file reading
filename = input("Enter filename to read: ")

try:
    with open(filename, "r") as f:
        content = f.read()
    print("File content:")
    print(content)
except FileNotFoundError:
    print(f"❌ File '{filename}' not found!")
except PermissionError:
    print(f"❌ No permission to read '{filename}'!")
finally:
    print("File operation completed.")
```

```python
# Retry pattern — keep asking until valid input
while True:
    try:
        age = int(input("Enter your age: "))
        if age < 0 or age > 150:
            print("⚠️ Please enter a realistic age!")
            continue
        break  # Exit loop if valid
    except ValueError:
        print("❌ Please enter a number!")

print(f"✅ Your age: {age}")
```

---

## 📊 Summary

| Block | Purpose | Runs When |
|---|---|---|
| `try` | Code that might cause error | Always attempted |
| `except` | Handle the error | Only if error occurs |
| `else` | Code if NO error | Only if NO error |
| `finally` | Cleanup code | ALWAYS (error or not) |
