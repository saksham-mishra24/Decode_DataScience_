## 📖 Understanding print()

The `print()` function is like a **loudspeaker** 📢

- You give it a message
- It shows (prints) that message on the screen

### Rules for using print():

1. Always write `print` in **small letters** (lowercase)
2. Put **round brackets** `()` after print
3. Put your message inside the brackets
4. If your message is text, put it inside **quotes** (`""` or `''`)

---

## 💻 More Examples of print()

### Example 1: Printing Text Messages

````python
# Using double quotes
print("I am learning Python!")

# Using single quotes (both work the same!)
print('Python is fun!')

# Printing a longer message
print("My name is Student and I am from India.")
````

### Example 2: Printing Numbers

````python
# Numbers DON'T need quotes
print(100)
print(3.14)
print(-25)
````

### Example 3: Printing Math Calculations

````python
# Python can do math and print the answer!
print(5 + 3)       # Addition
print(10 - 4)      # Subtraction
print(6 * 7)       # Multiplication
print(20 / 5)      # Division
````

### Example 4: Printing Multiple Things Together

````python
# Use comma (,) to print multiple things in one line
print("My age is", 20)
print("5 + 3 =", 5 + 3)
print("Name:", "Rahul", "| Age:", 21, "| Village:", "Sundargram")
````

### Example 5: Printing on Multiple Lines

````python
# Each print() automatically goes to a new line
print("Line 1: Hello")
print("Line 2: How are you?")
print("Line 3: I am fine!")
````

````python
# You can also use \n to create a new line WITHIN one print
print("Hello\nHow are you?\nI am fine!")
````

### Example 6: Printing an Empty Line

````python
print("First line")
print()              # This prints an empty line (for spacing)
print("Third line (with a gap above)")
````

### Example 7: Printing Special Characters

````python
# Printing with emojis and special characters
print("🌟 Stars are shining!")
print("I ❤️ Python!")
print("☀️ Good Morning!")
print("🇮🇳 Jai Hind!")
````

## ⚠️ Common Mistakes to Avoid

### Mistake 1: Writing Print with capital P

````python
# ❌ WRONG - Capital P
# Print("Hello")   # This will give an error!

# ✅ CORRECT - Small p
print("Hello")
````

### Mistake 2: Forgetting the Quotes

````python
# ❌ WRONG - No quotes around text
# print(Hello)   # This will give an error!

# ✅ CORRECT - Text in quotes
print("Hello")
````

### Mistake 3: Forgetting the Brackets

````python
# ❌ WRONG - No brackets
# print "Hello"   # This will give an error!

# ✅ CORRECT - With brackets
print("Hello")
````

### Mistake 4: Mixing Quote Types

````python
# ❌ WRONG - Starting with " but ending with '
# print("Hello')   # This will give an error!

# ✅ CORRECT - Same type of quotes
print("Hello")  # Both double quotes
print('Hello')  # Both single quotes
````
