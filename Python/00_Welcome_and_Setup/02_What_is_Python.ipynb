## 🤔 What is Python?

> **Python** is a programming language — a way to give instructions to a computer.

It was created by **Guido van Rossum** in **1991** (more than 30 years ago!).

### Fun Fact 🎉
The name "Python" does NOT come from the snake! 🐍  
Guido named it after a British comedy show called **"Monty Python's Flying Circus"** which he liked to watch.

---

## 🌟 Why Python is Special?

Python is the **most popular** programming language in the world today. Here's why:

### 1. ✅ Very Easy to Read and Write
Python code looks almost like English sentences!

**Compare these examples:**

**In C language (difficult to read):**
```c
#include <stdio.h>
int main() {
    printf("Hello World");
    return 0;
}
```

**In Java (medium difficulty):**
```java
public class Hello {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
```

**In Python (very easy!):**
```python
print("Hello World")
```

See the difference? Python does the same thing in just **1 line**! 🎯

- Millions of people around the world use Python

### 5. ✅ Very Powerful
- Python can do almost anything: websites, apps, AI, data analysis, automation

## 💻 Let's See Python in Action!

Run each code cell below by clicking on it and pressing `Shift + Enter`:

````python
# Python can print messages
print("Python is amazing!")
print("I am going to learn Python!")
````

````python
# Python can do mathematics
print("Addition:", 10 + 5)
print("Subtraction:", 20 - 8)
print("Multiplication:", 7 * 6)
print("Division:", 100 / 4)
````

````python
# Python can store information
my_name = "Student"
my_age = 20
my_village = "My Village"

print("My name is", my_name)
print("My age is", my_age)
print("I live in", my_village)
````

````python
# Python can make decisions
marks = 75

if marks >= 40:
    print("Congratulations! You PASSED! 🎉")
else:
    print("Sorry, you need to study more.")
````

````python
# Python can repeat tasks
print("Let me count from 1 to 5:")
for number in range(1, 6):
    print(number)
````
