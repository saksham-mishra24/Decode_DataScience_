---

## 📥 Step 1: Download Python

### What You Need:
- A computer (Windows, Mac, or Linux)
- Internet connection

### Steps:

1. **Open your web browser** (Chrome, Firefox, Edge — any browser)

2. **Go to the Python website:**
   - Type this in the address bar: **https://www.python.org**
   - Press Enter

3. **Click on "Downloads"** tab at the top

4. **Click the big yellow button** that says "Download Python 3.x.x"
   - The 'x.x' will be the latest version number
   - Just click the biggest download button you see!

5. **Wait** for the file to download (it's about 25-30 MB)

---

## 🔧 Step 2: Install Python on Windows

### Steps:

1. **Find the downloaded file**
   - Go to your Downloads folder
   - You will see a file like `python-3.x.x-amd64.exe`

2. **Double-click** on the file to start installation

3. **⚠️ VERY IMPORTANT: Check the box that says:**
   ```
   ☑ Add Python to PATH
   ```
   This is at the BOTTOM of the installer window. **YOU MUST CHECK THIS BOX!**

4. **Click "Install Now"**

5. **Wait** for the installation to complete (takes 2-5 minutes)

6. **Click "Close"** when it says "Setup was successful"

### ✅ How to Check if Python is Installed:

1. Press `Windows Key + R` on your keyboard
2. Type `cmd` and press Enter
3. In the black window that opens, type:
   ```
   python --version
   ```
4. Press Enter
5. You should see something like: `Python 3.12.4`

**If you see the version number, Python is installed! 🎉**

---

## 📓 Step 3: Install Jupyter Notebook

Jupyter Notebook is a tool where you can write and run Python code. The files you are reading right now are Jupyter Notebook files!

### Steps:

1. **Open Command Prompt (cmd)**
   - Press `Windows Key + R`
   - Type `cmd` and press Enter

2. **Type this command and press Enter:**
   ```
   pip install notebook
   ```

3. **Wait** for it to download and install (may take 5-10 minutes depending on internet speed)

4. **When it finishes**, you will see "Successfully installed" message

### ✅ How to Check if Jupyter is Installed:

In the same command prompt, type:
```
jupyter --version
```
If you see version numbers, Jupyter is installed! 🎉

---

## 🚀 Step 4: How to Open Jupyter Notebook

### Steps:

1. **Open Command Prompt**

2. **Navigate to your Python folder:**
   ```
   cd Desktop\Python
   ```

3. **Type this command:**
   ```
   jupyter notebook
   ```

4. **Your web browser will open** with Jupyter Notebook running!

5. **You will see your files and folders** listed in the browser

6. **Click on any `.ipynb` file** to open it

---

## 📖 Step 5: Using Jupyter Notebook

### What is a Cell?
Jupyter Notebook is made up of **cells**. Each cell is a box where you can write:
- **Code** (Python code that runs)
- **Text** (Explanations, like what you're reading now)

### How to Run a Code Cell:
1. **Click** on the code cell
2. Press **`Shift + Enter`** on your keyboard
3. The code will run and show the output below!

### Important Keyboard Shortcuts:

| Shortcut | What It Does |
|---|---|
| `Shift + Enter` | Run current cell and move to next |
| `Ctrl + Enter` | Run current cell and stay |
| `Ctrl + S` | Save the notebook |
| `Esc` then `A` | Add a new cell above |
| `Esc` then `B` | Add a new cell below |
| `Esc` then `DD` | Delete current cell |

## ⚠️ Common Problems and Solutions

| Problem | Solution |
|---|---|
| "python is not recognized" | You forgot to check "Add to PATH". Reinstall Python with PATH checked. |
| "pip is not recognized" | Same as above — reinstall Python with PATH checked. |
| Installation is very slow | Check your internet connection. Be patient. |
| Can't open Jupyter | Make sure you typed `jupyter notebook` correctly (with a space). |
