## 🤔 Why Do We Need Input?

So far, we have been putting values directly in our code:
```python
name = "Rahul"    # We decided the name
age = 20          # We decided the age
```

But what if we want the **user** (the person running the program) to type their own name and age?

That's where `input()` comes in!

### Real-Life Analogy: The Form 📋

When you go to a bank to open an account, they give you a **form** to fill:
- Name: ____________
- Age: _____________
- Address: __________

The `input()` function works the same way — it asks a question and waits for the user to type an answer!

---

## 💻 Using input()

```python
variable = input("Your question here: ")
```

- Python shows the question
- Waits for the user to type something
- Stores what they typed in the variable

### Example 1: Getting a Name

````python
# Getting input from user
name = input("What is your name? ")
print("Hello,", name, "! Nice to meet you! 🙏")
````

````python
# Getting multiple inputs
name = input("Enter your name: ")
village = input("Enter your village: ")
favorite_food = input("What is your favorite food? ")

print("\n--- Your Details ---")
print("Name:", name)
print("Village:", village)
print("Favorite Food:", favorite_food)
````

---

## ⚠️ IMPORTANT: input() Always Returns a String!

No matter what the user types, `input()` always gives you a **string** (text).

Even if the user types a number like `25`, Python treats it as the text `"25"`, not the number `25`!

````python
# See the problem:
age = input("Enter your age: ")   # User types: 20
print("Type of age:", type(age))   # <class 'str'> ← It's a STRING!

# This won't work for math:
# next_year_age = age + 1   # ERROR! Can't add string + number
````

### Solution: Convert to Number

Use `int()` or `float()` to convert the input to a number:

````python
# Getting a whole number (integer)
age = int(input("Enter your age: "))     # Converts to integer
next_year = age + 1
print("Next year, you will be", next_year, "years old!")
````

````python
# Getting a decimal number (float)
height = float(input("Enter your height in feet: "))  # Converts to float
print("Your height is", height, "feet")
print("In centimeters:", round(height * 30.48, 2), "cm")
````

## 🌾 Real-Life Programs Using Input

````python
# Program 1: Simple Calculator
print("=== Simple Calculator ===")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\n--- Results ---")
print(num1, "+", num2, "=", num1 + num2)
print(num1, "-", num2, "=", num1 - num2)
print(num1, "×", num2, "=", num1 * num2)
print(num1, "÷", num2, "=", round(num1 / num2, 2))
````

````python
# Program 2: Student Report Card
print("=== Student Report Card ===")

name = input("Enter student name: ")
hindi = int(input("Enter Hindi marks: "))
english = int(input("Enter English marks: "))
math_marks = int(input("Enter Math marks: "))
science = int(input("Enter Science marks: "))
social = int(input("Enter Social Science marks: "))

total = hindi + english + math_marks + science + social
percentage = (total / 500) * 100

print("\n" + "=" * 35)
print("    REPORT CARD")
print("=" * 35)
print("Student Name:", name)
print("-" * 35)
print("Hindi:         ", hindi)
print("English:       ", english)
print("Mathematics:   ", math_marks)
print("Science:       ", science)
print("Social Science:", social)
print("-" * 35)
print("Total:         ", total, "/ 500")
print("Percentage:    ", percentage, "%")
print("=" * 35)
````

````python
# Program 3: Farmer's Crop Calculator
print("=== 🌾 Crop Income Calculator ===")

crop_name = input("Enter crop name: ")
quantity_kg = float(input("How many kg did you harvest? "))
price_per_kg = float(input("Selling price per kg (₹): "))
expense = float(input("Total expense (₹): "))

income = quantity_kg * price_per_kg
profit = income - expense

print("\n--- Results ---")
print("Crop:", crop_name)
print("Total Income: ₹", income)
print("Total Expense: ₹", expense)
print("Profit: ₹", profit)

if profit > 0:
    print("🎉 You made a profit!")
else:
    print("😔 You had a loss.")
````

## 📊 Summary

| What You Want | How to Do It | Example |
|---|---|---|
| Get text input | `input()` | `name = input("Name: ")` |
| Get integer input | `int(input())` | `age = int(input("Age: "))` |
| Get decimal input | `float(input())` | `price = float(input("Price: "))` |

### Key Points:
- `input()` always returns a **string**
- Use `int()` to convert to whole number
- Use `float()` to convert to decimal number
- Always include a clear message inside `input()`
