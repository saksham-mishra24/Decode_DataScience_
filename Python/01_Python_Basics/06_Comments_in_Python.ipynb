## 🤔 What is a Comment?

> A **comment** is a note that you write in your code for humans to read. Python **ignores** comments — they are not executed.

### Real-Life Analogy: Sticky Notes on a Recipe 📝

When your mother writes a recipe, she might add small notes:
- "Add salt according to taste" ← This is a comment!
- "Stir for 5 minutes" ← This is an instruction

The notes help YOU understand, but they don't change the recipe itself.

---

## 1️⃣ Single-Line Comments

Use the `#` symbol to write a single-line comment:

```python
# This is a comment - Python will ignore this line
print("Hello!")  # This comment is after the code

# Calculating student marks
hindi = 78     # Hindi marks out of 100
english = 85   # English marks out of 100
total = hindi + english  # Adding both marks

print("Total marks:", total)
```

## 2️⃣ Multi-Line Comments

For longer comments, use triple quotes `"""` or `'''`:

```python
"""
This is a multi-line comment.
It can span multiple lines.
Python will ignore all of this.

Program: Simple Calculator
Author: Rahul Kumar
Date: 2024
"""

# You can also use multiple single-line comments
# This is line 1 of the comment
# This is line 2 of the comment
# This is line 3 of the comment

print("Calculator is ready!")
```

## 3️⃣ Using Comments to Disable Code

Comments are useful to temporarily **turn off** a line of code:

```python
print("Line 1: This will run")
# print("Line 2: This is disabled — it won't run")
print("Line 3: This will run")

# If you want to test, remove the # from line 2 and run again!
```

---

## 📖 When to Use Comments

| Use Comments For | Example |
|---|---|
| Explaining what code does | `# Calculate the average` |
| Explaining WHY you did something | `# Using floor division because we need whole pieces` |
| Adding your name and date | `# Author: Rahul, Date: 2024` |
| Temporarily disabling code | `# print(debug_info)` |
| Writing TODOs | `# TODO: Add error handling later` |

---

## ⚠️ Comment Do's and Don'ts

| ✅ Do | ❌ Don't |
|---|---|
| Explain WHY the code exists | State what is obvious |
| `# Check if student passed (min 33%)` | `# Add a and b` (for `a + b`) |
| Keep comments updated | Leave outdated comments |
| Write clearly and simply | Write paragraphs in comments |
