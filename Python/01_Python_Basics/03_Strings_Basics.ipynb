## 🤔 What is a String?

> A **string** is a sequence of characters (letters, numbers, symbols) enclosed in quotes.

Think of a string as a **garland of flowers** 💐 — each flower is one character, and they are all connected together in a line.

```
String: "Hello"

┌───┬───┬───┬───┬───┐
│ H │ e │ l │ l │ o │  ← Each box is one character
└───┴───┴───┴───┴───┘
  0   1   2   3   4    ← Position numbers (index) start from 0
```

---

## 💻 Creating Strings

There are 3 ways to create strings:

````python
# Method 1: Using double quotes (most common)
name = "Rahul Kumar"
print(name)

# Method 2: Using single quotes (same thing!)
city = 'Pune'
print(city)

# Method 3: Using triple quotes (for multi-line text)
address = """Village Sundargram,
District Pune,
State Maharashtra,
India - 411001"""
print(address)
````

````python
# When to use single vs double quotes?
# Use double quotes when your text has a single quote (apostrophe)
message1 = "It's a beautiful day!"
print(message1)

# Use single quotes when your text has double quotes
message2 = 'He said "Hello" to me.'
print(message2)
````

---

## ➕ String Concatenation (Joining Strings)

You can **join** two strings together using the `+` sign. This is called **concatenation**.

Think of it like joining two garlands together! 💐 + 💐 = 💐💐

````python
# Joining strings with +
first_name = "Rahul"
last_name = "Kumar"

full_name = first_name + " " + last_name  # Adding a space in between
print("Full Name:", full_name)

# More examples
greeting = "Hello" + " " + "World" + "!"
print(greeting)

# Building a sentence
city = "Pune"
sentence = "I live in " + city + "."
print(sentence)
````

### ⚠️ Important: You Cannot Add Strings and Numbers Directly!

````python
# ❌ This will give an error:
# print("My age is " + 20)   # Error! Can't add string + integer

# ✅ Convert number to string first using str()
age = 20
print("My age is " + str(age))   # Works!

# ✅ Or use comma in print (easier!)
print("My age is", age)   # This also works!
````

## 🔄 String Repetition (Repeating)

You can **repeat** a string using the `*` sign:

````python
# Repeating strings with *
print("Ha" * 5)         # HaHaHaHaHa
print("=" * 30)          # A line of 30 equal signs
print("-" * 40)          # A line of 40 dashes
print("🌟" * 10)        # 10 stars

# Useful for decoration
print("*" * 30)
print("     Welcome!")
print("*" * 30)
````

---

## 📏 String Length — len()

Use `len()` to find how many characters are in a string:

````python
# Finding length of strings
name = "Rahul"
print("Length of 'Rahul':", len(name))  # 5

message = "Hello World"
print("Length of 'Hello World':", len(message))  # 11 (space counts!)

empty = ""
print("Length of empty string:", len(empty))  # 0

# Spaces count as characters!
text = "A B C"
print("Length of 'A B C':", len(text))  # 5 (3 letters + 2 spaces)
````

## 🔤 Accessing Characters by Position (Indexing)

Each character in a string has a **position number** called an **index**.

**Important: Indexing starts from 0, not 1!**

```
String:  P  y  t  h  o  n
Index:   0  1  2  3  4  5
```

````python
name = "Python"

# Accessing individual characters
print("Full string:", name)
print("First character (index 0):", name[0])    # P
print("Second character (index 1):", name[1])   # y
print("Third character (index 2):", name[2])    # t
print("Last character (index 5):", name[5])     # n
print("Last character (index -1):", name[-1])   # n (negative index!)
````

````python
# Negative indexing (counting from the end)
word = "Python"

# Positive:  0  1  2  3  4  5
# Negative: -6 -5 -4 -3 -2 -1

print("Last character:", word[-1])        # n
print("Second last:", word[-2])            # o
print("Third last:", word[-3])             # h
````

## 🔍 Checking if Text Exists in a String

Use `in` to check if some text is inside a string:

````python
sentence = "I love learning Python programming"

print("Python" in sentence)      # True
print("Java" in sentence)        # False
print("love" in sentence)        # True
print("LOVE" in sentence)        # False (case sensitive!)
````

## 🔠 Changing Case

````python
name = "rahul Kumar"

print("UPPERCASE:", name.upper())       # RAHUL KUMAR
print("lowercase:", name.lower())       # rahul kumar
print("Title Case:", name.title())      # Rahul Kumar
print("Capitalized:", name.capitalize()) # Rahul kumar
print("Swapped:", name.swapcase())      # RAHUL kUMAR
````

---

## 📊 Summary

| Operation | Syntax | Example | Result |
|---|---|---|---|
| Create string | `"text"` or `'text'` | `name = "Rahul"` | `Rahul` |
| Join strings | `+` | `"Hello" + " World"` | `Hello World` |
| Repeat string | `*` | `"Ha" * 3` | `HaHaHa` |
| Length | `len()` | `len("Hello")` | `5` |
| Access character | `[index]` | `"Hello"[0]` | `H` |
| Check if exists | `in` | `"lo" in "Hello"` | `True` |
| Uppercase | `.upper()` | `"hello".upper()` | `HELLO` |
| Lowercase | `.lower()` | `"HELLO".lower()` | `hello` |
