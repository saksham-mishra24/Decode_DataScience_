## 🧮 Python as a Calculator

Python can work like a powerful calculator! You can do all kinds of math:

| Operation | Symbol | Example | Result |
|---|---|---|---|
| Addition | `+` | `5 + 3` | `8` |
| Subtraction | `-` | `10 - 4` | `6` |
| Multiplication | `*` | `6 * 7` | `42` |
| Division | `/` | `20 / 4` | `5.0` |
| Floor Division | `//` | `17 // 3` | `5` |
| Modulus (Remainder) | `%` | `17 % 3` | `2` |
| Power (Exponent) | `**` | `2 ** 3` | `8` |

Let's try each one! 👇

```python
# ➕ Addition
print("5 + 3 =", 5 + 3)
print("100 + 250 =", 100 + 250)
print("10.5 + 2.3 =", 10.5 + 2.3)
```

```python
# ➖ Subtraction
print("10 - 4 =", 10 - 4)
print("100 - 75 =", 100 - 75)
print("5.5 - 2.2 =", 5.5 - 2.2)
```

```python
# ✖️ Multiplication
print("6 × 7 =", 6 * 7)
print("12 × 12 =", 12 * 12)
print("3.5 × 2 =", 3.5 * 2)
```

```python
# ➗ Division (always gives a float/decimal result)
print("20 ÷ 4 =", 20 / 4)     # Result: 5.0 (not 5!)
print("17 ÷ 3 =", 17 / 3)     # Result: 5.666...
print("10 ÷ 3 =", 10 / 3)     # Result: 3.333...
```

```python
# ⬇️ Floor Division (removes the decimal part - gives only whole number)
print("17 // 3 =", 17 // 3)   # Result: 5 (removes .666)
print("10 // 3 =", 10 // 3)   # Result: 3 (removes .333)
print("25 // 7 =", 25 // 7)   # Result: 3

# Real-life example:
# You have 17 sweets and 3 children. How many does each child get?
sweets = 17
children = 3
sweets_per_child = sweets // children
print("\nEach child gets", sweets_per_child, "sweets")
```

```python
# 🔄 Modulus / Remainder (what's left over after division)
print("17 % 3 =", 17 % 3)     # Result: 2 (because 17 = 5×3 + 2)
print("10 % 3 =", 10 % 3)     # Result: 1 (because 10 = 3×3 + 1)
print("20 % 5 =", 20 % 5)     # Result: 0 (divides perfectly!)

# Real-life example:
# You have 17 sweets and 3 children. How many sweets are left?
leftover = 17 % 3
print("\nSweets left over:", leftover)
```

```python
# 💪 Power / Exponent (multiply a number by itself)
print("2 ** 3 =", 2 ** 3)     # 2 × 2 × 2 = 8
print("5 ** 2 =", 5 ** 2)     # 5 × 5 = 25
print("10 ** 3 =", 10 ** 3)   # 10 × 10 × 10 = 1000

# Square root using power
print("Square root of 16 =", 16 ** 0.5)  # √16 = 4.0
print("Square root of 25 =", 25 ** 0.5)  # √25 = 5.0
```

## 📏 BODMAS Rule in Python

Remember **BODMAS** from school? Python follows the same rule!

| Letter | Stands For | Symbol | Priority |
|---|---|---|---|
| B | Brackets | `()` | 1st (Highest) |
| O | Orders (Power) | `**` | 2nd |
| D | Division | `/`, `//` | 3rd |
| M | Multiplication | `*` | 3rd (same as D) |
| A | Addition | `+` | 4th |
| S | Subtraction | `-` | 4th (same as A) |

```python
# BODMAS examples

# Without brackets
result1 = 2 + 3 * 4      # Multiplication first: 3*4=12, then 2+12=14
print("2 + 3 * 4 =", result1)  # 14 (NOT 20!)

# With brackets
result2 = (2 + 3) * 4    # Brackets first: 2+3=5, then 5*4=20
print("(2 + 3) * 4 =", result2)  # 20

# More examples
print("10 - 2 * 3 =", 10 - 2 * 3)       # 10 - 6 = 4
print("(10 - 2) * 3 =", (10 - 2) * 3)   # 8 * 3 = 24
print("2 ** 3 + 1 =", 2 ** 3 + 1)       # 8 + 1 = 9
print("2 ** (3 + 1) =", 2 ** (3 + 1))   # 2^4 = 16
```

## 🌾 Real-Life Math Problems

```python
# Problem 1: Farmer's Profit
# A farmer sells 50 kg of wheat at ₹25 per kg
# He spent ₹800 on seeds and ₹200 on fertilizer

selling_price_per_kg = 25
total_kg = 50
seed_cost = 800
fertilizer_cost = 200

total_income = selling_price_per_kg * total_kg
total_expense = seed_cost + fertilizer_cost
profit = total_income - total_expense

print("--- Farmer's Profit Calculator ---")
print("Total Income: ₹", total_income)
print("Total Expense: ₹", total_expense)
print("Profit: ₹", profit)
```

```python
# Problem 2: Student's Percentage
# Calculate percentage from marks

hindi = 78
english = 65
math = 82
science = 90
social = 75

total_marks = hindi + english + math + science + social
max_marks = 500
percentage = (total_marks / max_marks) * 100

print("--- Student Report Card ---")
print("Hindi:", hindi)
print("English:", english)
print("Math:", math)
print("Science:", science)
print("Social:", social)
print("Total:", total_marks, "/", max_marks)
print("Percentage:", percentage, "%")
```

```python
# Problem 3: Simple Interest
# Formula: SI = (P × R × T) / 100

principal = 10000    # ₹10,000
rate = 8             # 8% per year
time = 3             # 3 years

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print("--- Simple Interest Calculator ---")
print("Principal: ₹", principal)
print("Rate:", rate, "%")
print("Time:", time, "years")
print("Simple Interest: ₹", simple_interest)
print("Total Amount: ₹", total_amount)
```

## 🛠️ Useful Math Functions

Python has some built-in functions for math:

```python
# abs() - Absolute value (removes negative sign)
print("abs(-5) =", abs(-5))       # 5
print("abs(5) =", abs(5))         # 5
print("abs(-3.7) =", abs(-3.7))   # 3.7

# round() - Round a number
print("\nround(3.7) =", round(3.7))     # 4
print("round(3.2) =", round(3.2))       # 3
print("round(3.14159, 2) =", round(3.14159, 2))  # 3.14

# min() and max() - Find smallest and largest
print("\nmin(5, 3, 8, 1) =", min(5, 3, 8, 1))   # 1
print("max(5, 3, 8, 1) =", max(5, 3, 8, 1))     # 8
```

```python
# Using the math module for advanced math
import math

print("Square root of 144 =", math.sqrt(144))  # 12.0
print("Value of Pi =", math.pi)                 # 3.14159...
print("Ceiling of 4.2 =", math.ceil(4.2))      # 5 (rounds UP)
print("Floor of 4.8 =", math.floor(4.8))        # 4 (rounds DOWN)
print("Factorial of 5 =", math.factorial(5))    # 120 (5×4×3×2×1)
```
