## 🤔 What is Type Conversion?

Type conversion means **changing data from one type to another**.

### Real-Life Analogy: Currency Exchange 💱

When you travel from India to the USA, you convert **Rupees to Dollars**.
The money is the same value, but the **type** (currency) changes.

Similarly, in Python:
- `"25"` (string) → `25` (integer) — same value, different type!
- `25` (integer) → `25.0` (float) — same value, different type!

---

## 1️⃣ Implicit Conversion (Automatic)

Python sometimes converts types **automatically** when needed:

```python
# Python automatically converts int to float when mixed
a = 5       # int
b = 2.0     # float

result = a + b   # int + float = float (automatic!)
print("5 + 2.0 =", result)
print("Type of result:", type(result))  # float

# Another example
x = 10      # int
y = 3       # int
z = x / y   # Division always gives float
print("\n10 / 3 =", z)
print("Type:", type(z))  # float
```

## 2️⃣ Explicit Conversion (Manual)

When you manually convert data types using functions:

| Function | Converts To | Example |
|---|---|---|
| `int()` | Integer | `int("25")` → `25` |
| `float()` | Float | `float("3.14")` → `3.14` |
| `str()` | String | `str(25)` → `"25"` |
| `bool()` | Boolean | `bool(1)` → `True` |

```python
# Converting to INTEGER
print("--- Converting to int ---")
print(int(3.14))     # 3    (removes decimal)
print(int(9.99))     # 9    (does NOT round, just removes decimal)
print(int("25"))     # 25   (string to int)
print(int(True))     # 1    (True = 1)
print(int(False))    # 0    (False = 0)
```

```python
# Converting to FLOAT
print("--- Converting to float ---")
print(float(25))       # 25.0
print(float("3.14"))   # 3.14
print(float("100"))    # 100.0
print(float(True))     # 1.0
print(float(False))    # 0.0
```

```python
# Converting to STRING
print("--- Converting to str ---")
print(str(25))       # "25"
print(str(3.14))     # "3.14"
print(str(True))     # "True"
print(str(False))    # "False"

# Useful for joining text with numbers
age = 20
message = "I am " + str(age) + " years old."
print(message)
```

```python
# Converting to BOOLEAN
print("--- Converting to bool ---")

# Numbers: 0 is False, everything else is True
print(bool(0))        # False
print(bool(1))        # True
print(bool(-5))       # True
print(bool(100))      # True

# Strings: Empty string is False, everything else is True
print(bool(""))       # False (empty string)
print(bool("Hello"))  # True
print(bool(" "))      # True (space is not empty!)
```

## ⚠️ Common Conversion Errors

```python
# ❌ Cannot convert text to number
# int("Hello")    # Error! "Hello" is not a number

# ❌ Cannot convert decimal string directly to int
# int("3.14")     # Error! Use float() first, then int()

# ✅ Correct way to convert decimal string to int:
result = int(float("3.14"))   # First float, then int
print("int(float('3.14')) =", result)  # 3
```

## 🌾 Real-Life Example: Input Processing

```python
# Getting input and converting types
name = input("Enter your name: ")                    # String (default)
age = int(input("Enter your age: "))                  # Convert to int
height = float(input("Enter height in feet: "))       # Convert to float

# Now we can do math with age and height!
birth_year = 2024 - age
height_cm = height * 30.48

print("\n--- Your Profile ---")
print("Name:", name)
print("Age:", age)
print("Birth Year (approx):", birth_year)
print("Height:", height, "feet =", round(height_cm, 1), "cm")
```

## 📊 Complete Conversion Table

| From → To | int() | float() | str() | bool() |
|---|---|---|---|---|
| `25` (int) | `25` | `25.0` | `"25"` | `True` |
| `0` (int) | `0` | `0.0` | `"0"` | `False` |
| `3.14` (float) | `3` | `3.14` | `"3.14"` | `True` |
| `"25"` (str) | `25` | `25.0` | `"25"` | `True` |
| `""` (empty str) | ❌ Error | ❌ Error | `""` | `False` |
| `True` (bool) | `1` | `1.0` | `"True"` | `True` |
| `False` (bool) | `0` | `0.0` | `"False"` | `False` |
