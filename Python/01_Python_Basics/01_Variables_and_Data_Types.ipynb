## 🤔 What is a Variable?

### Real-Life Analogy: The Labelled Box (डिब्बा) 📦

Imagine you have boxes (डिब्बे) in your house:

```
┌─────────┐   ┌─────────┐   ┌─────────┐
│  Rice   │   │  Sugar  │   │  Flour  │
│  🍚    │   │  🍬    │   │  🌾    │
│  5 kg   │   │  2 kg   │   │  3 kg   │
└─────────┘   └─────────┘   └─────────┘
```

Each box has:
1. A **label** (name) → "Rice", "Sugar", "Flour"
2. Something **stored inside** → 5 kg, 2 kg, 3 kg

**A variable in Python works exactly like a labelled box:**
- The **label** is the variable's **name**
- The thing **inside** is the variable's **value**

> **Variable = A named container that stores a value**

---

## 💻 Creating Variables in Python

To create a variable, you use the **equals sign** `=`

```
variable_name = value
```

The `=` sign means **"put this value into this box"** (not "equal to" like in math!)

### Let's Create Our First Variables:

````python
# Creating variables - just like labelling boxes!

name = "Rahul"           # A box labelled 'name' containing "Rahul"
age = 20                  # A box labelled 'age' containing 20
village = "Sundargram"    # A box labelled 'village' containing "Sundargram"
height = 5.7              # A box labelled 'height' containing 5.7

# Now let's see what's inside each box
print("Name:", name)
print("Age:", age)
print("Village:", village)
print("Height:", height)
````

### 🔍 What Just Happened?

```
name = "Rahul"
 |        |
 |        └── Value (what's inside the box)
 └── Variable name (the label on the box)
```

When we write `print(name)`, Python opens the box labelled "name" and shows us what's inside → `"Rahul"`

---

## 📐 Rules for Variable Names

Just like you can't put any random label on a box, there are rules for variable names:

### ✅ Rules You MUST Follow:

| Rule | Good Example ✅ | Bad Example ❌ |
|---|---|---|
| Must start with a letter or underscore | `name`, `_age` | `2name` (starts with number) |
| Can contain letters, numbers, underscores | `student_1`, `my_name` | `my-name` (hyphen not allowed) |
| No spaces allowed | `first_name` | `first name` (space not allowed) |
| Case sensitive (capital ≠ small) | `Name` and `name` are DIFFERENT | — |
| Cannot use Python keywords | `my_class` | `class` (reserved word) |

### 💡 Tips for Good Variable Names:
- Use **meaningful names**: `student_age` is better than `x`
- Use **snake_case**: words separated by underscores → `first_name`, `total_marks`
- Keep names **short but clear**: `num` instead of `number_of_students_in_class`

````python
# ✅ GOOD variable names
student_name = "Priya"
total_marks = 450
is_passed = True
_private_data = "secret"
age2 = 25

print("Student:", student_name)
print("Marks:", total_marks)
print("Passed:", is_passed)
````

````python
# ❌ BAD variable names - These will give ERRORS
# Uncomment any line below to see the error:

# 2name = "Rahul"        # Error: starts with a number
# my name = "Rahul"      # Error: has a space
# my-name = "Rahul"      # Error: has a hyphen
# class = "10th"         # Error: 'class' is a Python keyword
````

## 🎨 What are Data Types?

Just like in real life, different boxes can contain different TYPES of things:
- A box can contain **rice** (a grain)
- A box can contain **water** (a liquid)
- A box can contain **clothes** (fabric)

Similarly, variables can store different TYPES of data:

### The 4 Basic Data Types in Python:

| Data Type | Python Name | What It Stores | Example |
|---|---|---|---|
| Integer | `int` | Whole numbers (no decimal) | `10`, `-5`, `0`, `100` |
| Float | `float` | Decimal numbers | `3.14`, `-2.5`, `0.0` |
| String | `str` | Text (words, sentences) | `"Hello"`, `"Rahul"`, `"123"` |
| Boolean | `bool` | True or False (Yes or No) | `True`, `False` |

---

## 1️⃣ Integer (int) — Whole Numbers

Integers are **whole numbers** without any decimal point.

**Real-Life Examples:**
- Number of students in a class: `45`
- Your age: `20`
- Number of chapatis you eat: `3`
- Temperature in Celsius: `-2`

````python
# Integer examples
students = 45
age = 20
chapatis = 3
temperature = -2
year = 2024

print("Students in class:", students)
print("My age:", age)
print("Chapatis I eat:", chapatis)
print("Temperature:", temperature, "°C")
print("Year:", year)

# Check the data type
print("\nType of 'age' variable:", type(age))
````

## 2️⃣ Float — Decimal Numbers

Float numbers are numbers **with a decimal point**.

**Real-Life Examples:**
- Price of rice per kg: `45.50`
- Your height in feet: `5.7`
- Your weight in kg: `65.5`
- Percentage in exam: `78.6`

````python
# Float examples
rice_price = 45.50
height = 5.7
weight = 65.5
percentage = 78.6
pi = 3.14159

print("Rice price per kg: ₹", rice_price)
print("Height:", height, "feet")
print("Weight:", weight, "kg")
print("Percentage:", percentage, "%")
print("Value of Pi:", pi)

# Check the data type
print("\nType of 'height' variable:", type(height))
````

## 3️⃣ String (str) — Text

Strings are **text** — any letters, words, sentences. They are always written inside **quotes** (`""` or `''`).

**Real-Life Examples:**
- Your name: `"Rahul Kumar"`
- Your village: `"Sundargram"`
- A message: `"Good morning!"`
- Even numbers as text: `"12345"` (this is NOT a number, it's text!)

````python
# String examples
name = "Rahul Kumar"
village = "Sundargram"
greeting = "Good morning!"
phone = "9876543210"    # This is a STRING, not a number!
address = "Village Sundargram, District Pune, Maharashtra"

print("Name:", name)
print("Village:", village)
print("Greeting:", greeting)
print("Phone:", phone)
print("Address:", address)

# Check the data type
print("\nType of 'name' variable:", type(name))
print("Type of 'phone' variable:", type(phone))  # It's str, not int!
````

### ⚠️ Important Note: Numbers in Quotes

```python
age = 20        # This is an INTEGER (number) ← Can do math with it
age = "20"      # This is a STRING (text) ← Cannot do math with it
```

If a number is inside quotes, Python treats it as TEXT, not as a number!

````python
# See the difference:
number_age = 20       # Integer
text_age = "20"       # String

print("Number age + 5 =", number_age + 5)   # Math works! → 25
print("Text age + '5' =", text_age + "5")    # Joins text! → "205"

print("\nType of number_age:", type(number_age))  # int
print("Type of text_age:", type(text_age))        # str
````

## 4️⃣ Boolean (bool) — True or False

Boolean is the simplest data type — it can only be **True** or **False**.

**Real-Life Examples:**
- Is it raining? → `True` or `False`
- Did you pass the exam? → `True` or `False`
- Is the shop open? → `True` or `False`
- Are you a student? → `True` or `False`

**Note:** `True` and `False` must start with a **capital letter**!

````python
# Boolean examples
is_raining = False
is_student = True
has_passed = True
is_shop_open = False

print("Is it raining?", is_raining)
print("Am I a student?", is_student)
print("Did I pass?", has_passed)
print("Is shop open?", is_shop_open)

# Check the data type
print("\nType of 'is_student':", type(is_student))
````

````python
# Booleans are also created by comparisons
print("Is 5 greater than 3?", 5 > 3)      # True
print("Is 2 equal to 7?", 2 == 7)          # False
print("Is 10 less than 20?", 10 < 20)      # True
````

---

## 🔍 Checking Data Type with type()

You can use the `type()` function to check what type of data a variable holds:

````python
# Let's check all data types
a = 10
b = 3.14
c = "Hello"
d = True

print("Value:", a, "→ Type:", type(a))   # int
print("Value:", b, "→ Type:", type(b))   # float
print("Value:", c, "→ Type:", type(c))   # str
print("Value:", d, "→ Type:", type(d))   # bool
````

---

## 🔄 Changing Variable Values

You can change what's inside a box anytime — just put a new value in it!

````python
# Variable values can be changed
age = 20
print("Age before birthday:", age)

age = 21   # Happy Birthday! Age changes!
print("Age after birthday:", age)

# You can even change the type!
x = 10
print("\nx is:", x, "| Type:", type(x))    # int

x = "Hello"
print("x is now:", x, "| Type:", type(x))   # str
````

---

## 📦 Multiple Variables at Once

Python lets you create multiple variables in one line:

````python
# Method 1: Multiple variables in one line
name, age, city = "Rahul", 20, "Pune"
print(name, age, city)

# Method 2: Same value to multiple variables
a = b = c = 100
print(a, b, c)

# Method 3: Swapping two variables
x = 10
y = 20
print("Before swap: x =", x, ", y =", y)

x, y = y, x  # Magic! They swap!
print("After swap: x =", x, ", y =", y)
````

---

## ⚠️ Common Mistakes

| Mistake | Why It's Wrong | Correct Way |
|---|---|---|
| `Name = "Rahul"` then `print(name)` | Capital N ≠ small n | Use same case everywhere |
| `true` (small t) | Must be `True` with capital T | `True` or `False` |
| `age = "20"` then `age + 5` | Can't add string + number | Use `age = 20` (without quotes) |
| `my name = "Rahul"` | Spaces not allowed | `my_name = "Rahul"` |
