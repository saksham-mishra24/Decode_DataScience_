## 📋 Membership Operators: `in` and `not in`

These check if a value **exists inside** a collection (like a list, string, or tuple).

### Real-Life Analogy: Attendance Register 📋

"Is Rahul **in** the class?" → True/False

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `in` | Is present? | `"a" in "apple"` | `True` |
| `not in` | Is absent? | `"z" in "apple"` | `False` |

```python
# Membership in STRINGS
sentence = "I love learning Python"

print("'Python' in sentence:", "Python" in sentence)       # True
print("'Java' in sentence:", "Java" in sentence)           # False
print("'love' in sentence:", "love" in sentence)           # True
print("'Java' not in sentence:", "Java" not in sentence)   # True
```

```python
# Membership in LISTS
fruits = ["apple", "banana", "mango", "grapes"]

print("'mango' in fruits:", "mango" in fruits)       # True
print("'orange' in fruits:", "orange" in fruits)     # False
print("'orange' not in fruits:", "orange" not in fruits)  # True

# Numbers in a list
marks = [78, 85, 92, 67, 88]
print("\n85 in marks:", 85 in marks)    # True
print("100 in marks:", 100 in marks)    # False
```

```python
# Real-life: Check attendance
students_present = ["Rahul", "Priya", "Amit", "Sunita", "Ravi"]

name = input("Enter student name to check: ")

if name in students_present:
    print(name, "is PRESENT ✅")
else:
    print(name, "is ABSENT ❌")
```

## 🆔 Identity Operators: `is` and `is not`

These check if two variables point to the **exact same object** in memory.

| Operator | Meaning | Example |
|---|---|---|
| `is` | Same object? | `x is y` |
| `is not` | Different object? | `x is not y` |

### `==` vs `is`:
- `==` checks if values are **equal**
- `is` checks if they are the **exact same object** in memory

```python
# Identity examples
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)     # True (same values)
print("a is b:", a is b)     # False (different objects in memory)
print("a is c:", a is c)     # True (c points to same object as a)

# Checking with None (most common use)
x = None
print("\nx is None:", x is None)       # True
print("x is not None:", x is not None) # False
```

---

## 📊 Summary

| Operator Type | Operator | Use Case |
|---|---|---|
| Membership | `in` | Check if item exists in a collection |
| Membership | `not in` | Check if item doesn't exist |
| Identity | `is` | Check if same object (mostly used with `None`) |
| Identity | `is not` | Check if different objects |
