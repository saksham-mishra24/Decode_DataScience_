## 🤔 What are Logical Operators?

Logical operators combine **two or more conditions** and give a single True/False result.

### Real-Life Analogy: College Admission 🎓

To get admission, you need:
- Marks >= 60 **AND** Age <= 25

Both conditions must be true! If one is false, no admission.

### The 3 Logical Operators:

| Operator | Meaning | Example |
|---|---|---|
| `and` | Both must be True | `True and True` → `True` |
| `or` | At least one must be True | `True or False` → `True` |
| `not` | Reverses True↔False | `not True` → `False` |

---

## 1️⃣ AND Operator

`and` returns **True only when BOTH conditions are True**.

Think of it like: "I will go to the picnic **IF** it's a holiday **AND** the weather is good."

| Condition A | Condition B | A and B |
|---|---|---|
| True | True | **True** ✅ |
| True | False | **False** ❌ |
| False | True | **False** ❌ |
| False | False | **False** ❌ |

```python
# AND examples
print("=== AND Operator ===")
print("True and True:", True and True)     # True
print("True and False:", True and False)   # False
print("False and True:", False and True)   # False
print("False and False:", False and False) # False

# Real example: College admission
marks = 75
age = 22
eligible = (marks >= 60) and (age <= 25)
print("\nMarks:", marks, ", Age:", age)
print("Eligible for admission?", eligible)  # True
```

```python
# More AND examples

# Can a person drive?
has_license = True
age = 20
can_drive = has_license and (age >= 18)
print("Can drive?", can_drive)  # True

# Is it a working day?
is_weekday = True
is_not_holiday = True
is_working_day = is_weekday and is_not_holiday
print("Working day?", is_working_day)  # True
```

## 2️⃣ OR Operator

`or` returns **True when AT LEAST ONE condition is True**.

Think: "I will buy it **IF** it's on sale **OR** I have enough money."

| Condition A | Condition B | A or B |
|---|---|---|
| True | True | **True** ✅ |
| True | False | **True** ✅ |
| False | True | **True** ✅ |
| False | False | **False** ❌ |

```python
# OR examples
print("=== OR Operator ===")
print("True or True:", True or True)       # True
print("True or False:", True or False)     # True
print("False or True:", False or True)     # True
print("False or False:", False or False)   # False

# Real example: Payment method
has_cash = False
has_upi = True
can_pay = has_cash or has_upi
print("\nHas cash:", has_cash, ", Has UPI:", has_upi)
print("Can pay?", can_pay)  # True (UPI works!)
```

## 3️⃣ NOT Operator

`not` simply **reverses** True to False, and False to True.

| Condition | not Condition |
|---|---|
| True | **False** |
| False | **True** |

```python
# NOT examples
print("=== NOT Operator ===")
print("not True:", not True)    # False
print("not False:", not False)  # True

# Real example
is_raining = True
go_outside = not is_raining
print("\nIs it raining?", is_raining)
print("Should I go outside?", go_outside)  # False
```

## 🔗 Combining Logical Operators

```python
# Complex example: Loan eligibility
print("=== Bank Loan Eligibility ===")

age = 30
salary = 25000
has_good_credit = True

# Must be 21-58 AND (salary >= 20000 OR has_good_credit)
age_ok = (age >= 21) and (age <= 58)
income_ok = (salary >= 20000) or has_good_credit
eligible = age_ok and income_ok

print("Age:", age, "→ Age OK?", age_ok)
print("Salary: ₹", salary, "→ Income OK?", income_ok)
print("Loan Eligible?", eligible)
```

---

## 📊 Summary

| Operator | Meaning | When True |
|---|---|---|
| `and` | Both conditions | Both must be True |
| `or` | Either condition | At least one must be True |
| `not` | Opposite | Reverses the condition |
