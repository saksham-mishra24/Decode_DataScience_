## 🤔 What are Comparison Operators?

Comparison operators **compare two values** and give you a result: **True** or **False**.

### Real-Life Analogy: The Judge ⚖️

Imagine a judge who answers questions with only "Yes" (True) or "No" (False):
- "Is 5 greater than 3?" → **True** (Yes!)
- "Is 2 equal to 7?" → **False** (No!)

### All 6 Comparison Operators:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to (same?) | `5 == 5` | `True` |
| `!=` | Not equal to (different?) | `5 != 3` | `True` |
| `>` | Greater than (bigger?) | `7 > 3` | `True` |
| `<` | Less than (smaller?) | `3 < 7` | `True` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `3 <= 7` | `True` |

---

## 1️⃣ Equal to ( == )

**⚠️ Note:** `==` (double equal) checks equality. `=` (single equal) assigns a value. They are DIFFERENT!

```python
print("=== Equal to (==) ===")
print("5 == 5:", 5 == 5)       # True
print("5 == 3:", 5 == 3)       # False
print("10 == 10.0:", 10 == 10.0)  # True (same value)

# With strings
print("'hello' == 'hello':", "hello" == "hello")  # True
print("'Hello' == 'hello':", "Hello" == "hello")  # False (capital H!)
```

## 2️⃣ Not Equal to ( != )

```python
print("=== Not Equal to (!=) ===")
print("5 != 3:", 5 != 3)       # True (they ARE different)
print("5 != 5:", 5 != 5)       # False (they are NOT different)
print("'cat' != 'dog':", "cat" != "dog")  # True
```

## 3️⃣ Greater Than ( > ) and Less Than ( < )

```python
print("=== Greater Than (>) ===")
print("7 > 3:", 7 > 3)         # True
print("3 > 7:", 3 > 7)         # False
print("5 > 5:", 5 > 5)         # False (not greater, they're equal)

print("\n=== Less Than (<) ===")
print("3 < 7:", 3 < 7)         # True
print("7 < 3:", 7 < 3)         # False
print("5 < 5:", 5 < 5)         # False (not less, they're equal)
```

## 4️⃣ Greater/Less Than or Equal to ( >=, <= )

```python
print("=== Greater Than or Equal To (>=) ===")
print("7 >= 3:", 7 >= 3)       # True (7 is greater)
print("5 >= 5:", 5 >= 5)       # True (5 is equal)
print("3 >= 7:", 3 >= 7)       # False

print("\n=== Less Than or Equal To (<=) ===")
print("3 <= 7:", 3 <= 7)       # True
print("5 <= 5:", 5 <= 5)       # True
print("7 <= 3:", 7 <= 3)       # False
```

## 🌾 Real-Life Examples

```python
# Example 1: Can a person vote?
age = 18
can_vote = age >= 18
print("Age:", age, "→ Can vote?", can_vote)  # True

# Example 2: Did student pass?
marks = 35
passing_marks = 33
passed = marks >= passing_marks
print("Marks:", marks, "→ Passed?", passed)  # True

# Example 3: Is temperature hot?
temperature = 42
is_hot = temperature > 40
print("Temperature:", temperature, "°C → Is hot?", is_hot)  # True

# Example 4: Is password correct?
saved_password = "python123"
entered_password = "python123"
is_correct = saved_password == entered_password
print("Password correct?", is_correct)  # True
```

```python
# Using comparisons with variables
print("=== Comparing Variables ===")

ram_marks = 85
shyam_marks = 72

print("Ram's marks:", ram_marks)
print("Shyam's marks:", shyam_marks)
print("Ram scored more than Shyam?", ram_marks > shyam_marks)   # True
print("Both scored same?", ram_marks == shyam_marks)             # False
print("Difference exists?", ram_marks != shyam_marks)            # True
```

## 📊 Summary

| Operator | Meaning | True Example | False Example |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `5 == 3` |
| `!=` | Not equal | `5 != 3` | `5 != 5` |
| `>` | Greater than | `7 > 3` | `3 > 7` |
| `<` | Less than | `3 < 7` | `7 < 3` |
| `>=` | Greater or equal | `5 >= 5` | `3 >= 7` |
| `<=` | Less or equal | `5 <= 5` | `7 <= 3` |

### Key Points:
- Comparison operators always return **True** or **False**
- `==` checks equality (not `=` which assigns values!)
- String comparisons are **case-sensitive**
