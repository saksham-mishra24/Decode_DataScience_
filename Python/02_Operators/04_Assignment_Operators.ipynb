## 🤔 What are Assignment Operators?

Assignment operators **put values into variables** (assign values to boxes).

We already know `=` — but there are shortcuts too!

### All Assignment Operators:

| Operator | Meaning | Example | Same As |
|---|---|---|---|
| `=` | Assign | `x = 5` | — |
| `+=` | Add and assign | `x += 3` | `x = x + 3` |
| `-=` | Subtract and assign | `x -= 2` | `x = x - 2` |
| `*=` | Multiply and assign | `x *= 4` | `x = x * 4` |
| `/=` | Divide and assign | `x /= 2` | `x = x / 2` |
| `//=` | Floor divide and assign | `x //= 3` | `x = x // 3` |
| `%=` | Modulus and assign | `x %= 3` | `x = x % 3` |
| `**=` | Power and assign | `x **= 2` | `x = x ** 2` |

---

```python
# Basic assignment
x = 10
print("Start: x =", x)

# Add and assign
x += 5    # Same as: x = x + 5
print("After x += 5: x =", x)   # 15

# Subtract and assign
x -= 3    # Same as: x = x - 3
print("After x -= 3: x =", x)   # 12

# Multiply and assign
x *= 2    # Same as: x = x * 2
print("After x *= 2: x =", x)   # 24

# Divide and assign
x /= 4    # Same as: x = x / 4
print("After x /= 4: x =", x)   # 6.0
```

## 🌾 Real-Life Example: Bank Account

```python
# Bank account simulation
balance = 5000
print("Opening Balance: ₹", balance)

# Deposit ₹2000
balance += 2000
print("After deposit of ₹2000: ₹", balance)   # 7000

# Withdraw ₹1500
balance -= 1500
print("After withdrawal of ₹1500: ₹", balance)  # 5500

# Interest (multiply by 1.05 for 5% interest)
balance *= 1.05
print("After 5% interest: ₹", round(balance, 2))  # 5775.0
```

```python
# Counter example
count = 0

count += 1   # Increment by 1
print("Count:", count)  # 1

count += 1   # Increment again
print("Count:", count)  # 2

count += 1   # And again
print("Count:", count)  # 3

# This is very useful in loops (we'll learn loops soon!)
```

---

## 📊 Summary

| Shortcut | Full Form | Meaning |
|---|---|---|
| `x += 5` | `x = x + 5` | Add 5 to x |
| `x -= 3` | `x = x - 3` | Subtract 3 from x |
| `x *= 2` | `x = x * 2` | Multiply x by 2 |
| `x /= 4` | `x = x / 4` | Divide x by 4 |

**Key Point:** These are just shortcuts! `x += 5` does the exact same thing as `x = x + 5`.
