## 🧮 What are Arithmetic Operators?

Arithmetic operators are symbols used to perform **mathematical calculations** — just like the symbols you used in school math!

### All 7 Arithmetic Operators:

| Operator | Name | Symbol | Example | Result |
|---|---|---|---|---|
| Addition | Jod (जोड़) | `+` | `5 + 3` | `8` |
| Subtraction | Ghataav (घटाव) | `-` | `10 - 4` | `6` |
| Multiplication | Guna (गुणा) | `*` | `6 * 7` | `42` |
| Division | Bhaag (भाग) | `/` | `15 / 4` | `3.75` |
| Floor Division | Poora Bhaag | `//` | `15 // 4` | `3` |
| Modulus | Shesh (शेष) | `%` | `15 % 4` | `3` |
| Exponent | Ghaatank (घातांक) | `**` | `2 ** 4` | `16` |

## 1️⃣ Addition ( + )

```python
# Addition examples
print("=== Addition (+) ===")
print("5 + 3 =", 5 + 3)           # 8
print("100 + 250 =", 100 + 250)   # 350
print("-5 + 3 =", -5 + 3)         # -2
print("2.5 + 1.5 =", 2.5 + 1.5)   # 4.0

# Real-life: Total price of groceries
rice = 250
dal = 180
oil = 150
total = rice + dal + oil
print("\nGrocery Total: ₹", total)
```

## 2️⃣ Subtraction ( - )

```python
# Subtraction examples
print("=== Subtraction (-) ===")
print("10 - 4 =", 10 - 4)         # 6
print("3 - 8 =", 3 - 8)           # -5 (negative result)
print("100.5 - 50.2 =", 100.5 - 50.2)  # 50.3

# Real-life: Money left after shopping
wallet = 1000
spent = 650
remaining = wallet - spent
print("\nMoney left in wallet: ₹", remaining)
```

## 3️⃣ Multiplication ( * )

```python
# Multiplication examples
print("=== Multiplication (*) ===")
print("6 * 7 =", 6 * 7)           # 42
print("12 * 12 =", 12 * 12)       # 144
print("2.5 * 4 =", 2.5 * 4)       # 10.0

# Real-life: Total cost of items
price_per_kg = 45
quantity = 5
total_cost = price_per_kg * quantity
print("\n5 kg rice at ₹45/kg = ₹", total_cost)
```

## 4️⃣ Division ( / )

**Important:** Division ALWAYS gives a float (decimal) result!

```python
# Division examples
print("=== Division (/) ===")
print("20 / 4 =", 20 / 4)         # 5.0 (float, not int!)
print("15 / 4 =", 15 / 4)         # 3.75
print("10 / 3 =", 10 / 3)         # 3.3333...

# Real-life: Dividing equally
total_amount = 1000
num_people = 3
each_person_gets = total_amount / num_people
print("\n₹1000 divided among 3 people: ₹", round(each_person_gets, 2), "each")
```

## 5️⃣ Floor Division ( // )

Floor division gives only the **whole number** part (removes the decimal).

```python
# Floor Division examples
print("=== Floor Division (//) ===")
print("15 / 4 =", 15 / 4)         # 3.75 (normal division)
print("15 // 4 =", 15 // 4)       # 3    (floor division - only whole part)
print("17 // 5 =", 17 // 5)       # 3
print("10 // 3 =", 10 // 3)       # 3

# Real-life: How many full bags?
total_apples = 23
apples_per_bag = 5
full_bags = total_apples // apples_per_bag
print("\n23 apples in bags of 5:", full_bags, "full bags")
```

## 6️⃣ Modulus / Remainder ( % )

Modulus gives the **remainder** after division.

```python
# Modulus examples
print("=== Modulus (%) ===")
print("15 % 4 =", 15 % 4)         # 3 (remainder)
print("17 % 5 =", 17 % 5)         # 2
print("20 % 5 =", 20 % 5)         # 0 (divides perfectly!)

# Real-life: Leftover apples
total_apples = 23
apples_per_bag = 5
leftover = total_apples % apples_per_bag
print("\n23 apples in bags of 5:", leftover, "apples left over")

# Useful: Check if a number is even or odd
num = 15
print("\n15 % 2 =", 15 % 2)  # 1 → odd
print("20 % 2 =", 20 % 2)    # 0 → even
```

## 7️⃣ Exponent / Power ( ** )

```python
# Exponent examples
print("=== Exponent (**) ===")
print("2 ** 3 =", 2 ** 3)         # 8   (2×2×2)
print("5 ** 2 =", 5 ** 2)         # 25  (5×5)
print("10 ** 4 =", 10 ** 4)       # 10000
print("9 ** 0.5 =", 9 ** 0.5)     # 3.0 (square root!)

# Real-life: Area of a square
side = 7
area = side ** 2
print("\nArea of square with side 7 =", area, "sq units")
```

## 🏆 Operator Precedence (Which Runs First?)

| Priority | Operator | Name |
|---|---|---|
| 1st (Highest) | `()` | Parentheses |
| 2nd | `**` | Exponent |
| 3rd | `*`, `/`, `//`, `%` | Multiply, Divide, Floor Div, Modulus |
| 4th (Lowest) | `+`, `-` | Add, Subtract |

```python
# Precedence examples
print("2 + 3 * 4 =", 2 + 3 * 4)       # 14 (multiply first)
print("(2 + 3) * 4 =", (2 + 3) * 4)   # 20 (brackets first)
print("10 - 2 ** 2 =", 10 - 2 ** 2)   # 6  (power first: 4, then 10-4)
print("20 / 4 + 3 =", 20 / 4 + 3)     # 8.0 (divide first)
```

## 🌾 Complete Real-Life Program

```python
# Kisan (Farmer) Profit Calculator
print("🌾 === KISAN PROFIT CALCULATOR === 🌾")
print()

crop = input("Enter crop name: ")
harvest_kg = float(input("Total harvest (kg): "))
price_per_kg = float(input("Selling price per kg (₹): "))
seed_cost = float(input("Seed cost (₹): "))
fertilizer_cost = float(input("Fertilizer cost (₹): "))
labour_cost = float(input("Labour cost (₹): "))

# Calculations using arithmetic operators
total_income = harvest_kg * price_per_kg              # Multiplication
total_expense = seed_cost + fertilizer_cost + labour_cost  # Addition
profit = total_income - total_expense                  # Subtraction
profit_per_kg = profit / harvest_kg                    # Division
profit_percentage = (profit / total_expense) * 100     # Percentage

print("\n" + "=" * 40)
print("    PROFIT REPORT")
print("=" * 40)
print("Crop:            ", crop)
print("Total Income:     ₹", total_income)
print("Total Expense:    ₹", total_expense)
print("Net Profit:       ₹", profit)
print("Profit per kg:    ₹", round(profit_per_kg, 2))
print("Profit %:         ", round(profit_percentage, 2), "%")
print("=" * 40)
```
