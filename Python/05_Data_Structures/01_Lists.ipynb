## 🤔 What is a List?

> A **list** is an ordered collection of items stored in a single variable.

### Real-Life Analogy: Shopping List 🛒

When you go to the market, you carry a shopping list:

```
Shopping List:
1. Rice
2. Sugar
3. Oil
4. Tea
5. Soap
```

A Python list works the same way — it holds multiple items in one place!

### Key Properties:
- **Ordered** — items have a fixed position (index)
- **Changeable** — you can add, remove, modify items
- **Allows duplicates** — same item can appear multiple times
- **Can hold different types** — numbers, strings, even other lists!

---

## 💻 Creating Lists

Lists are created using **square brackets** `[]`:

````python
# Different types of lists

# List of strings
fruits = ["Apple", "Banana", "Mango", "Grapes"]
print("Fruits:", fruits)

# List of numbers
marks = [78, 85, 92, 67, 88]
print("Marks:", marks)

# List of mixed types
student = ["Rahul", 20, 5.7, True, "Pune"]
print("Student:", student)

# Empty list
empty = []
print("Empty list:", empty)

# Length of list
print("\nNumber of fruits:", len(fruits))
````

---

## 🔤 Accessing List Elements (Indexing)

Each item has a position number (index), starting from **0**.

```
fruits = ["Apple", "Banana", "Mango", "Grapes"]
Index:      0        1        2        3
Negative:  -4       -3       -2       -1
```

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Orange"]

# Positive indexing
print("First fruit:", fruits[0])    # Apple
print("Second fruit:", fruits[1])   # Banana
print("Third fruit:", fruits[2])    # Mango

# Negative indexing (from the end)
print("\nLast fruit:", fruits[-1])         # Orange
print("Second last:", fruits[-2])           # Grapes
````

---

## ✂️ List Slicing

Get a **portion** of the list using `list[start:end]`:

````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

print("Original:", numbers)
print("First 3:", numbers[0:3])     # [10, 20, 30]
print("Index 2-5:", numbers[2:5])   # [30, 40, 50]
print("Last 3:", numbers[-3:])      # [60, 70, 80]
print("Every 2nd:", numbers[::2])   # [10, 30, 50, 70]
print("Reversed:", numbers[::-1])   # [80, 70, 60, 50, 40, 30, 20, 10]
````

## ➕ Adding Elements

````python
# Starting with a shopping list
shopping = ["Rice", "Sugar", "Oil"]
print("Original:", shopping)

# append() — Add to the END
shopping.append("Tea")
print("After append('Tea'):", shopping)

# insert() — Add at a specific position
shopping.insert(1, "Dal")   # Insert at index 1
print("After insert(1, 'Dal'):", shopping)

# extend() — Add multiple items
shopping.extend(["Soap", "Shampoo"])
print("After extend:", shopping)
````

---

## ➖ Removing Elements

````python
fruits = ["Apple", "Banana", "Mango", "Grapes", "Banana"]
print("Original:", fruits)

# remove() — Remove by value (first occurrence)
fruits.remove("Banana")
print("After remove('Banana'):", fruits)

# pop() — Remove by index (returns the removed item)
removed = fruits.pop(1)   # Remove index 1
print("Popped:", removed)
print("After pop(1):", fruits)

# pop() — Remove last item (no index)
last = fruits.pop()
print("Popped last:", last)
print("After pop():", fruits)

# clear() — Remove ALL items
fruits.clear()
print("After clear():", fruits)  # []
````

## 🔧 Useful List Methods

````python
# sort() — Sort the list
marks = [78, 45, 92, 67, 88, 34]
print("Original:", marks)

marks.sort()    # Ascending order
print("Sorted:", marks)

marks.sort(reverse=True)  # Descending order
print("Reverse sorted:", marks)

# reverse() — Reverse the order
names = ["Amit", "Priya", "Rahul"]
names.reverse()
print("\nReversed names:", names)

# count() — Count occurrences
numbers = [1, 3, 5, 3, 2, 3, 8]
print("\nCount of 3:", numbers.count(3))  # 3

# index() — Find position
print("Index of 5:", numbers.index(5))    # 2
````

## 🔁 Looping Through Lists

````python
# Simple loop
fruits = ["Apple", "Banana", "Mango", "Grapes"]

for fruit in fruits:
    print("🍎", fruit)

# Loop with index using enumerate()
print("\n=== With Index ===")
for index, fruit in enumerate(fruits):
    print(f"{index + 1}. {fruit}")
````

````python
# Real-life: Student marks analysis
marks = [78, 85, 92, 67, 88, 45, 73, 95]

print("Total students:", len(marks))
print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))
print("Total marks:", sum(marks))
print("Average:", round(sum(marks) / len(marks), 2))

# Count passed students
passed = 0
for mark in marks:
    if mark >= 33:
        passed += 1
print("Passed:", passed, "out of", len(marks))
````

## 📊 Summary of List Methods

| Method | What It Does | Example |
|---|---|---|
| `append(x)` | Add x to end | `list.append("Tea")` |
| `insert(i, x)` | Add x at position i | `list.insert(0, "First")` |
| `extend(list2)` | Add all items from list2 | `list.extend([1,2,3])` |
| `remove(x)` | Remove first x | `list.remove("Tea")` |
| `pop(i)` | Remove at index i | `list.pop(0)` |
| `sort()` | Sort ascending | `list.sort()` |
| `reverse()` | Reverse order | `list.reverse()` |
| `count(x)` | Count x | `list.count(3)` |
| `index(x)` | Find position of x | `list.index("Mango")` |
| `clear()` | Remove all items | `list.clear()` |
| `copy()` | Copy the list | `list2 = list.copy()` |
