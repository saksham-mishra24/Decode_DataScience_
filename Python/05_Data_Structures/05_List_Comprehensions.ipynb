## 🤔 What is List Comprehension?

List comprehension is a **shortcut** to create lists using just one line of code!

### Normal Way (3+ lines):
```python
squares = []
for i in range(1, 6):
    squares.append(i ** 2)
```

### List Comprehension Way (1 line!):
```python
squares = [i ** 2 for i in range(1, 6)]
```

Both give: `[1, 4, 9, 16, 25]`

### Syntax:
```python
new_list = [expression for item in sequence]
new_list = [expression for item in sequence if condition]
```

---

````python
# Basic list comprehension
squares = [i ** 2 for i in range(1, 11)]
print("Squares 1-10:", squares)

# Cubes
cubes = [i ** 3 for i in range(1, 6)]
print("Cubes 1-5:", cubes)

# Double each number
numbers = [1, 2, 3, 4, 5]
doubled = [n * 2 for n in numbers]
print("Doubled:", doubled)
````

````python
# With condition (filter)

# Only even numbers
evens = [i for i in range(1, 21) if i % 2 == 0]
print("Even 1-20:", evens)

# Only passed students
all_marks = [78, 25, 92, 15, 88, 45, 30, 95]
passed = [m for m in all_marks if m >= 33]
print("Passed marks:", passed)

# Failed marks
failed = [m for m in all_marks if m < 33]
print("Failed marks:", failed)
````

````python
# With strings
names = ["rahul", "priya", "amit", "sunita"]
capitalized = [name.capitalize() for name in names]
print("Capitalized:", capitalized)

# Length of each word
words = ["apple", "banana", "cat", "elephant"]
lengths = [len(word) for word in words]
print("Lengths:", lengths)

# With if-else
marks = [78, 25, 92, 15, 88]
results = ["Pass" if m >= 33 else "Fail" for m in marks]
print("Results:", results)
````

---

## 📊 Summary

| Type | Syntax | Example |
|---|---|---|
| Basic | `[expr for item in seq]` | `[i**2 for i in range(5)]` |
| With filter | `[expr for item in seq if cond]` | `[i for i in range(10) if i%2==0]` |
| With if-else | `[a if cond else b for item in seq]` | `["Pass" if m>=33 else "Fail" for m in marks]` |
