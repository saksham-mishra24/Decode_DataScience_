## 🤔 What is a Dictionary?

> A **dictionary** stores data as **key-value pairs** — like a real dictionary where each word (key) has a meaning (value).

### Real-Life Analogy: Student Mark Sheet 📝

```
Subject  → Marks
Hindi    → 78
English  → 85
Math     → 92
Science  → 67
```

Each subject (key) has marks (value) associated with it.

### Key Properties:
- Stores **key: value** pairs
- Keys must be **unique** (no duplicate keys)
- Keys are used to access values (not index numbers)
- Uses **curly braces** `{key: value}`

---

````python
# Creating dictionaries
marks = {
    "Hindi": 78,
    "English": 85,
    "Math": 92,
    "Science": 67,
    "Social": 75
}
print("Marks:", marks)

# Student profile
student = {
    "name": "Rahul Kumar",
    "age": 20,
    "village": "Sundargram",
    "is_student": True
}
print("Student:", student)
````

````python
# Accessing values using keys
marks = {"Hindi": 78, "English": 85, "Math": 92}

print("Hindi marks:", marks["Hindi"])       # 78
print("Math marks:", marks["Math"])          # 92

# Using get() — safer (no error if key missing)
print("Science marks:", marks.get("Science", "Not found"))  # Not found
````

````python
# Adding and modifying
student = {"name": "Rahul", "age": 20}
print("Original:", student)

# Add new key-value pair
student["city"] = "Pune"
print("After adding city:", student)

# Modify existing value
student["age"] = 21
print("After updating age:", student)

# Remove
del student["city"]
print("After deleting city:", student)

# pop() — remove and get the value
age = student.pop("age")
print("Popped age:", age)
print("After pop:", student)
````

````python
# Looping through dictionaries
marks = {"Hindi": 78, "English": 85, "Math": 92, "Science": 67, "Social": 75}

# Loop through keys
print("=== Subjects ===")
for subject in marks:
    print("📚", subject)

# Loop through key-value pairs
print("\n=== Report Card ===")
for subject, mark in marks.items():
    status = "✅" if mark >= 33 else "❌"
    print(f"{status} {subject}: {mark}")

# Calculate total and average
total = sum(marks.values())
average = total / len(marks)
print(f"\nTotal: {total} / {len(marks) * 100}")
print(f"Average: {round(average, 2)}")
````

````python
# Useful dictionary methods
data = {"a": 1, "b": 2, "c": 3}

print("Keys:", list(data.keys()))       # ['a', 'b', 'c']
print("Values:", list(data.values()))   # [1, 2, 3]
print("Items:", list(data.items()))     # [('a',1), ('b',2), ('c',3)]
print("Length:", len(data))              # 3
print("'a' exists?", "a" in data)        # True
````

````python
# Nested dictionaries — dictionary inside a dictionary
classroom = {
    "Rahul": {"age": 20, "marks": 85, "city": "Pune"},
    "Priya": {"age": 19, "marks": 92, "city": "Mumbai"},
    "Amit":  {"age": 21, "marks": 78, "city": "Delhi"}
}

# Access nested data
print("Rahul's marks:", classroom["Rahul"]["marks"])
print("Priya's city:", classroom["Priya"]["city"])

# Loop through nested dict
print("\n=== Class Details ===")
for name, info in classroom.items():
    print(f"{name}: Age={info['age']}, Marks={info['marks']}, City={info['city']}")
````

## 📊 Comparison of All Data Structures

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Brackets | `[]` | `()` | `{}` | `{key: val}` |
| Ordered | ✅ Yes | ✅ Yes | ❌ No | ✅ Yes (3.7+) |
| Changeable | ✅ Yes | ❌ No | ✅ Yes | ✅ Yes |
| Duplicates | ✅ Allowed | ✅ Allowed | ❌ No | ❌ Keys unique |
| Indexing | ✅ Yes | ✅ Yes | ❌ No | By key |
