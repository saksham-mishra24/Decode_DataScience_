## 🤔 What is a Tuple?

> A **tuple** is like a list, but it **cannot be changed** after creation.

### Real-Life Analogy: Aadhaar Card 🪪

Your Aadhaar number is fixed — you can't change it!
- Aadhaar Number: `(1234, 5678, 9012)` → This is a tuple! Fixed forever.
- Shopping List: `[Rice, Sugar, Oil]` → This is a list. You can add/remove items.

### List vs Tuple:

| Feature | List | Tuple |
|---|---|---|
| Brackets | `[square]` | `(round)` |
| Changeable? | ✅ Yes | ❌ No (immutable) |
| Speed | Slower | Faster |
| Use when | Data will change | Data should NOT change |

---

```python
# Creating tuples
coordinates = (28.6139, 77.2090)   # Delhi coordinates
colors = ("Red", "Green", "Blue")
student = ("Rahul", 20, "Pune")
single = (42,)    # Single element tuple needs a comma!

print("Coordinates:", coordinates)
print("Colors:", colors)
print("Student:", student)
print("Type:", type(colors))
```

```python
# Accessing elements (same as lists)
colors = ("Red", "Green", "Blue", "Yellow", "Orange")

print("First:", colors[0])      # Red
print("Last:", colors[-1])       # Orange
print("Slice:", colors[1:4])     # ('Green', 'Blue', 'Yellow')
print("Length:", len(colors))     # 5
```

```python
# Tuples are IMMUTABLE (cannot change)
my_tuple = (1, 2, 3)

# ❌ This will give an error:
# my_tuple[0] = 10   # TypeError: 'tuple' does not support item assignment

# ❌ Cannot append or remove
# my_tuple.append(4)   # Error!
# my_tuple.remove(1)   # Error!

print("Tuples cannot be changed after creation! 🔒")
```

```python
# Tuple unpacking — very useful!
student = ("Rahul", 20, "Pune")

# Unpack into separate variables
name, age, city = student
print("Name:", name)
print("Age:", age)
print("City:", city)

# Swap using tuples
a, b = 10, 20
print("\nBefore swap:", a, b)
a, b = b, a   # This is actually tuple unpacking!
print("After swap:", a, b)
```

```python
# Tuple methods
numbers = (1, 3, 5, 3, 2, 3, 8)

print("Count of 3:", numbers.count(3))   # 3
print("Index of 5:", numbers.index(5))   # 2

# Looping through tuples
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
for day in days:
    print("📅", day)
```

## 📊 When to Use Tuples

| Use Tuple For | Example |
|---|---|
| Coordinates | `(28.61, 77.20)` |
| RGB Colors | `(255, 0, 0)` |
| Database records | `("Rahul", 20, "Pune")` |
| Days/Months | `("Mon", "Tue", ...)` |
| Function return values | `return (min, max)` |

**Rule of thumb:** Use tuples when data should NOT change. Use lists when it should.
