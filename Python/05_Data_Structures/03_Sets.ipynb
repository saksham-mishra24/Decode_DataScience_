## 🤔 What is a Set?

> A **set** is an unordered collection of **unique** items (no duplicates!).

### Real-Life Analogy: Unique Students in Class

If you take attendance and someone's name appears twice, you still count them once. That's a set!

### Key Properties:
- ❌ No duplicates
- ❌ No order (no indexing)
- ✅ Very fast for checking membership
- Uses **curly braces** `{}`

---

```python
# Creating sets
fruits = {"Apple", "Banana", "Mango", "Apple", "Banana"}
print("Fruits:", fruits)    # Duplicates automatically removed!
print("Count:", len(fruits))  # Only 3, not 5

# Numbers set
numbers = {1, 5, 3, 5, 2, 1, 4}
print("Numbers:", numbers)  # Each appears only once
```

```python
# Removing duplicates from a list using set
names_with_duplicates = ["Rahul", "Priya", "Rahul", "Amit", "Priya", "Sunita"]
unique_names = set(names_with_duplicates)
print("Original:", names_with_duplicates)
print("Unique:", unique_names)
print("Unique count:", len(unique_names))
```

```python
# Adding and removing
fruits = {"Apple", "Banana"}

fruits.add("Mango")    # Add one item
print("After add:", fruits)

fruits.update(["Grapes", "Orange"])  # Add multiple
print("After update:", fruits)

fruits.remove("Banana")  # Remove (error if not found)
print("After remove:", fruits)

fruits.discard("Kiwi")   # Remove (no error if not found)
print("After discard:", fruits)
```

## 🔗 Set Operations (Like Math!)

Think of Venn diagrams from school:

```python
# Students who play cricket and football
cricket = {"Rahul", "Amit", "Priya", "Ravi", "Sunita"}
football = {"Amit", "Ravi", "Suresh", "Meena", "Priya"}

# UNION — All students (both sports combined)
all_players = cricket | football  # or cricket.union(football)
print("All players:", all_players)

# INTERSECTION — Students who play BOTH
both = cricket & football  # or cricket.intersection(football)
print("Play both:", both)

# DIFFERENCE — Only cricket (not football)
only_cricket = cricket - football  # or cricket.difference(football)
print("Only cricket:", only_cricket)

# SYMMETRIC DIFFERENCE — Play only one sport
one_sport = cricket ^ football
print("Only one sport:", one_sport)
```

```python
# Membership check (very fast!)
students = {"Rahul", "Priya", "Amit", "Sunita"}
print("Rahul present?", "Rahul" in students)   # True
print("Ravi present?", "Ravi" in students)      # False
```

---

## 📊 Summary

| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `\|` | `.union()` | All items from both |
| Intersection | `&` | `.intersection()` | Common items |
| Difference | `-` | `.difference()` | Items in first but not second |
| Symmetric Diff | `^` | `.symmetric_difference()` | Items in one but not both |
