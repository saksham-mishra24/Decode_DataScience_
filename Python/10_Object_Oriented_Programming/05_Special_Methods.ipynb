# ✨ Special (Magic/Dunder) Methods

---

## 📌 Methods with double underscores like `__init__`, `__str__`, `__len__`

These let your objects work with built-in Python operations.

---

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):     # When you print() the object
        return f"'{self.title}' by {self.author}"
    
    def __len__(self):     # When you use len()
        return self.pages
    
    def __eq__(self, other):   # When you use ==
        return self.title == other.title
    
    def __lt__(self, other):   # When you use <
        return self.pages < other.pages

# Use it
b1 = Book("Python Basics", "Rahul", 200)
b2 = Book("Data Science", "Priya", 350)
b3 = Book("Python Basics", "Amit", 180)

print(b1)                    # __str__: 'Python Basics' by Rahul
print(f"Pages: {len(b1)}")  # __len__: 200
print(f"b1 == b3? {b1 == b3}")  # __eq__: True (same title)
print(f"b1 < b2? {b1 < b2}")   # __lt__: True (200 < 350)

# Can even sort!
books = [b1, b2, b3]
for book in sorted(books):
    print(f"  📚 {book} ({len(book)} pages)")
```

---

## 📊 Common Magic Methods

| Method | Called When | Example |
|---|---|---|
| `__init__` | Creating object | `obj = MyClass()` |
| `__str__` | `print(obj)` | Human-readable string |
| `__repr__` | `repr(obj)` | Developer string |
| `__len__` | `len(obj)` | Length |
| `__eq__` | `obj1 == obj2` | Equality |
| `__lt__` | `obj1 < obj2` | Less than |
| `__add__` | `obj1 + obj2` | Addition |
