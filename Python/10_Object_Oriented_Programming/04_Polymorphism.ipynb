---

## 📌 Same method name, different behavior

### Analogy: The word "Open" 🚪
- "Open" the **door** → push/pull
- "Open" a **book** → lift the cover
- "Open" a **bottle** → twist the cap

Same word, different actions depending on the object!

---

```python
# Polymorphism with classes
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def describe(self):
        return f"Circle with radius {self.radius}"

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):   # Same method name!
        return self.length * self.width
    
    def describe(self):
        return f"Rectangle {self.length}×{self.width}"

class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):   # Same method name!
        return 0.5 * self.base * self.height
    
    def describe(self):
        return f"Triangle base={self.base}, height={self.height}"

# Polymorphism in action!
shapes = [Circle(5), Rectangle(4, 6), Triangle(3, 8)]

for shape in shapes:
    print(f"{shape.describe()} → Area = {shape.area():.2f}")
```

```python
# Built-in polymorphism in Python
# len() works on different types!
print("len('Hello'):", len("Hello"))       # String → 5
print("len([1,2,3]):", len([1, 2, 3]))     # List → 3
print("len({'a':1}):", len({"a": 1}))      # Dict → 1

# + operator works differently for different types
print("\n2 + 3 =", 2 + 3)               # Numbers → addition
print("'Hi' + '!' =", "Hi" + "!")       # Strings → concatenation
print("[1,2] + [3] =", [1,2] + [3])     # Lists → merge
```
