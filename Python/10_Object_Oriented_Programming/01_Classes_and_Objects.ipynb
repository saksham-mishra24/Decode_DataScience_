## 🤔 What is OOP?

### Real-Life Analogy: Blueprint and House 🏠

- A **class** is like a **blueprint** (नक्शा) of a house
- An **object** is the **actual house** built from that blueprint
- From ONE blueprint, you can build MANY houses!

```
Class (Blueprint)  →  Object 1 (Rahul's house)
                   →  Object 2 (Priya's house)
                   →  Object 3 (Amit's house)
```

---

````python
# Creating a class
class Student:
    """A class to represent a student"""
    
    def __init__(self, name, age, marks):
        """Initialize student with name, age, and marks"""
        self.name = name      # Attribute
        self.age = age        # Attribute
        self.marks = marks    # Attribute
    
    def display(self):
        """Display student info"""
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Marks: {self.marks}")
    
    def get_grade(self):
        """Return grade based on marks"""
        if self.marks >= 90: return "A+"
        elif self.marks >= 80: return "A"
        elif self.marks >= 70: return "B"
        elif self.marks >= 60: return "C"
        elif self.marks >= 33: return "D"
        else: return "F"

# Creating objects (instances)
student1 = Student("Rahul", 20, 85)
student2 = Student("Priya", 19, 92)
student3 = Student("Amit", 21, 67)

# Using objects
print("=== Student 1 ===")
student1.display()
print(f"Grade: {student1.get_grade()}")

print("\n=== Student 2 ===")
student2.display()
print(f"Grade: {student2.get_grade()}")
````

````python
# Accessing and modifying attributes
print("Name:", student1.name)       # Access
student1.marks = 90                  # Modify
print("New marks:", student1.marks)
print("New grade:", student1.get_grade())
````

````python
# Another example: Bank Account
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f"✅ Deposited ₹{amount}. Balance: ₹{self.balance}")
    
    def withdraw(self, amount):
        if amount > self.balance:
            print(f"❌ Insufficient balance! Available: ₹{self.balance}")
        else:
            self.balance -= amount
            print(f"✅ Withdrawn ₹{amount}. Balance: ₹{self.balance}")
    
    def check_balance(self):
        print(f"💰 {self.owner}'s Balance: ₹{self.balance}")

# Use it
account = BankAccount("Rahul", 5000)
account.check_balance()
account.deposit(3000)
account.withdraw(2000)
account.withdraw(10000)
account.check_balance()
````

## 📊 Key Terms

| Term | Meaning | Example |
|---|---|---|
| **Class** | Blueprint/template | `class Student:` |
| **Object** | Instance of class | `s1 = Student(...)` |
| **Attribute** | Data/property | `self.name` |
| **Method** | Function inside class | `def display(self):` |
| **`__init__`** | Constructor (auto-runs) | Sets up initial values |
| **`self`** | Refers to current object | `self.name = name` |
