# 🔐 Encapsulation

---

## 📌 Hiding data and controlling access

### Analogy: Medicine Capsule 💊
A capsule hides the medicine inside. You can't directly touch the medicine — you take the whole capsule. Similarly, encapsulation hides data inside a class.

---

```python
# Public, Protected, Private attributes
class Account:
    def __init__(self, name, balance):
        self.name = name            # Public — accessible everywhere
        self._account_type = "Savings"  # Protected — convention: don't use outside
        self.__balance = balance     # Private — can't access directly outside
    
    # Getter — controlled access to private data
    def get_balance(self):
        return self.__balance
    
    # Setter — controlled modification
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"✅ Deposited ₹{amount}")
        else:
            print("❌ Amount must be positive!")
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"✅ Withdrawn ₹{amount}")
        else:
            print("❌ Invalid amount or insufficient balance!")

acc = Account("Rahul", 5000)
print("Name:", acc.name)              # ✅ Public — works
print("Type:", acc._account_type)      # ⚠️ Protected — works but shouldn't
# print(acc.__balance)                 # ❌ Private — Error!
print("Balance:", acc.get_balance())   # ✅ Controlled access

acc.deposit(2000)
acc.withdraw(1000)
print("Final Balance:", acc.get_balance())
```

---

## 📊 Access Levels

| Type | Syntax | Access |
|---|---|---|
| Public | `self.name` | Everywhere |
| Protected | `self._name` | Class + subclasses (convention) |
| Private | `self.__name` | Only inside the class |
