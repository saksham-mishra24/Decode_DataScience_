# 🎨 String Formatting

---

## 📌 What You Will Learn

- 3 ways to format strings in Python
- f-strings (the modern way)
- .format() method
- % formatting (old way)

---

## 🤔 Why String Formatting?

Instead of awkward concatenation:
```python
print("My name is " + name + " and I am " + str(age) + " years old.")
```

We can write clean, readable code:
```python
print(f"My name is {name} and I am {age} years old.")
```

---

## 1️⃣ f-strings (Recommended! Python 3.6+)

Put `f` before the string and use `{variable}` inside:

````python
name = "Rahul"
age = 20
city = "Pune"

# f-string — clean and easy!
print(f"My name is {name}")
print(f"I am {age} years old")
print(f"I live in {city}")
print(f"{name} is {age} years old and lives in {city}.")

# Calculations inside f-strings!
print(f"Next year I will be {age + 1}")
print(f"5 × 7 = {5 * 7}")

# Formatting numbers
price = 1234.5678
print(f"Price: ₹{price:.2f}")        # 2 decimal places
print(f"Price: ₹{price:,.2f}")       # With comma separator
print(f"Percentage: {78.5:.1f}%")     # 1 decimal place
````

## 2️⃣ .format() Method

````python
# Using .format()
print("My name is {} and I am {} years old.".format("Rahul", 20))
print("I live in {city} and study {subject}.".format(city="Pune", subject="Python"))

# Numbered placeholders
print("{0} is friends with {1}. {1} likes {0}.".format("Rahul", "Priya"))
````

## 3️⃣ % Formatting (Old Way)

````python
# % formatting (you may see this in old code)
name = "Rahul"
age = 20

print("My name is %s" % name)              # %s = string
print("I am %d years old" % age)            # %d = integer
print("Height: %.2f feet" % 5.7)            # %f = float
print("%s is %d years old" % (name, age))   # Multiple values
````

## 🌾 Real-Life Example: Receipt Generator

````python
# Beautiful receipt using f-strings
shop = "Kisan General Store"
items = [("Rice 5kg", 250), ("Sugar 2kg", 80), ("Oil 1L", 150), ("Tea 250g", 100)]

print(f"{'=' * 40}")
print(f"{'🏪 ' + shop:^40}")
print(f"{'=' * 40}")
print(f"{'Item':<20} {'Price':>15}")
print(f"{'-' * 40}")

total = 0
for item, price in items:
    print(f"{item:<20} {'₹' + str(price):>15}")
    total += price

print(f"{'-' * 40}")
print(f"{'TOTAL':<20} {'₹' + str(total):>15}")
print(f"{'=' * 40}")
print(f"{'Thank you! Visit again!':^40}")
````

---

## 📊 Summary

| Method | Syntax | Best For |
|---|---|---|
| **f-string** | `f"Hello {name}"` | ✅ Modern, recommended |
| **.format()** | `"Hello {}".format(name)` | Older Python versions |
| **% format** | `"Hello %s" % name` | Very old code |

**Always use f-strings** when possible — they are the easiest and fastest!
