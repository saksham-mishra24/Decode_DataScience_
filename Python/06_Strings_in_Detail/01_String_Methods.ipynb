## 🔠 Case Changing Methods

```python
text = "hello WORLD python"

print("Original:   ", text)
print("upper():    ", text.upper())        # HELLO WORLD PYTHON
print("lower():    ", text.lower())        # hello world python
print("title():    ", text.title())        # Hello World Python
print("capitalize():", text.capitalize())  # Hello world python
print("swapcase(): ", text.swapcase())     # HELLO world PYTHON
```

## 🔍 Search Methods

```python
sentence = "I love Python programming. Python is awesome!"

# find() — Returns index of first occurrence (-1 if not found)
print("find('Python'):", sentence.find("Python"))      # 7
print("find('Java'):", sentence.find("Java"))           # -1

# count() — Count occurrences
print("count('Python'):", sentence.count("Python"))    # 2

# startswith() and endswith()
print("starts with 'I':", sentence.startswith("I"))    # True
print("ends with '!':", sentence.endswith("!"))        # True
```

## ✂️ Modify Methods

```python
# replace() — Replace text
msg = "I love Java"
new_msg = msg.replace("Java", "Python")
print("Original:", msg)
print("Replaced:", new_msg)  # I love Python

# strip() — Remove whitespace from edges
messy = "   Hello World!   "
print("Before strip: '", messy, "'")
print("After strip:  '", messy.strip(), "'")
print("lstrip():     '", messy.lstrip(), "'")   # Left only
print("rstrip():     '", messy.rstrip(), "'")   # Right only
```

## 🔗 Split and Join

```python
# split() — Break string into a list
sentence = "Python is easy to learn"
words = sentence.split()   # Split by space (default)
print("Words:", words)  # ['Python', 'is', 'easy', 'to', 'learn']

# Split by comma
csv_data = "Rahul,20,Pune,India"
parts = csv_data.split(",")
print("CSV parts:", parts)

# join() — Combine list into a string
words = ["Python", "is", "fun"]
sentence = " ".join(words)     # Join with space
print("Joined:", sentence)

# Join with different separators
print("With dash:", "-".join(words))
print("With arrow:", " → ".join(words))
```

## ✅ Check Methods

```python
# isalpha() — Only letters?
print("'Hello'.isalpha():", "Hello".isalpha())     # True
print("'Hello123'.isalpha():", "Hello123".isalpha()) # False

# isdigit() — Only digits?
print("'12345'.isdigit():", "12345".isdigit())      # True
print("'123.5'.isdigit():", "123.5".isdigit())      # False

# isalnum() — Letters and/or digits?
print("'Hello123'.isalnum():", "Hello123".isalnum()) # True

# isspace() — Only whitespace?
print("'   '.isspace():", "   ".isspace())           # True

# isupper() / islower()
print("'HELLO'.isupper():", "HELLO".isupper())       # True
print("'hello'.islower():", "hello".islower())       # True
```

## 📊 Complete String Methods Table

| Method | What It Does | Example |
|---|---|---|
| `upper()` | All uppercase | `"hello".upper()` → `"HELLO"` |
| `lower()` | All lowercase | `"HELLO".lower()` → `"hello"` |
| `title()` | Title case | `"hello world".title()` → `"Hello World"` |
| `strip()` | Remove spaces | `" hi ".strip()` → `"hi"` |
| `replace(a,b)` | Replace a with b | `"cat".replace("c","b")` → `"bat"` |
| `split(sep)` | Break into list | `"a,b,c".split(",")` → `["a","b","c"]` |
| `join(list)` | List to string | `",".join(["a","b"])` → `"a,b"` |
| `find(x)` | Find position | `"hello".find("l")` → `2` |
| `count(x)` | Count x | `"hello".count("l")` → `2` |
| `startswith(x)` | Starts with x? | `"hello".startswith("he")` → `True` |
| `endswith(x)` | Ends with x? | `"hello".endswith("lo")` → `True` |
