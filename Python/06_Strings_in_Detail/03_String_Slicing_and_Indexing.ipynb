## 🔢 Indexing Refresher

```
String:    P   y   t   h   o   n
Positive:  0   1   2   3   4   5
Negative: -6  -5  -4  -3  -2  -1
```

````python
text = "PROGRAMMING"

print("String:", text)
print("First char:", text[0])      # P
print("Last char:", text[-1])       # G
print("5th char:", text[4])         # R
print("Length:", len(text))          # 11
````

## ✂️ Slicing: `string[start:end:step]`

- `start` — where to begin (inclusive)
- `end` — where to stop (exclusive)
- `step` — how many to skip

````python
text = "Hello World"

print("Full:", text)               # Hello World
print("[0:5]:", text[0:5])         # Hello
print("[6:11]:", text[6:11])       # World
print("[:5]:", text[:5])           # Hello (start from 0)
print("[6:]:", text[6:])           # World (go to end)
print("[-5:]:", text[-5:])         # World
print("[::2]:", text[::2])         # HloWrd (every 2nd char)
print("[::-1]:", text[::-1])       # dlroW olleH (reversed!)
````

````python
# Practical examples
phone = "91-9876543210"
country_code = phone[:2]
number = phone[3:]
print(f"Country code: +{country_code}")
print(f"Phone number: {number}")

# Palindrome check
word = "madam"
reversed_word = word[::-1]
print(f"\n'{word}' reversed is '{reversed_word}'")
print(f"Is palindrome? {word == reversed_word}")

# Extract first and last name
full_name = "Rahul Kumar"
space_pos = full_name.find(" ")
first = full_name[:space_pos]
last = full_name[space_pos + 1:]
print(f"\nFirst name: {first}")
print(f"Last name: {last}")
````

---

## 📊 Slicing Quick Reference

| Syntax | Result | Meaning |
|---|---|---|
| `s[0]` | First char | Index 0 |
| `s[-1]` | Last char | Last index |
| `s[2:5]` | Index 2,3,4 | Start to end-1 |
| `s[:5]` | First 5 chars | From beginning |
| `s[5:]` | From index 5 to end | To end |
| `s[::2]` | Every 2nd char | Step of 2 |
| `s[::-1]` | Reversed | Step of -1 |
