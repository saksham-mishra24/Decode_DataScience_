## 🤔 What are Regular Expressions?

> **Regular expressions** (regex) are special patterns used to search, match, and manipulate text.

Think of it like a **detective's magnifying glass** 🔍 — it helps you find specific patterns in text.

Examples:
- Find all phone numbers in a document
- Check if an email address is valid
- Find all words that start with capital letter

---

```python
import re  # Import the regex module

# re.findall() — Find all matches
text = "My phone is 9876543210 and office is 8765432109"
phones = re.findall(r"\d{10}", text)   # Find 10-digit numbers
print("Phone numbers found:", phones)

# re.search() — Find first match
result = re.search(r"\d+", "Price is 450 rupees")
if result:
    print("Number found:", result.group())  # 450
```

```python
# Common patterns
text = "Contact rahul@gmail.com or priya123@yahoo.com for details"

# Find all email addresses
emails = re.findall(r"[\w.]+@[\w.]+", text)
print("Emails:", emails)

# Find all words starting with capital letter
text2 = "Hello World Python Is Amazing"
caps = re.findall(r"[A-Z][a-z]+", text2)
print("Capital words:", caps)
```

```python
# re.sub() — Replace patterns
text = "My number is 9876543210"
hidden = re.sub(r"\d", "X", text)   # Replace all digits with X
print("Hidden:", hidden)  # My number is XXXXXXXXXX

# re.split() — Split by pattern
text = "apple;banana,mango orange:grapes"
fruits = re.split(r"[;,: ]", text)  # Split by ; , : or space
print("Fruits:", fruits)
```

---

## 📊 Common Regex Patterns

| Pattern | Meaning | Example |
|---|---|---|
| `\d` | Any digit (0-9) | `\d+` matches `123` |
| `\w` | Any letter/digit/underscore | `\w+` matches `hello_123` |
| `\s` | Any whitespace | `\s+` matches spaces |
| `.` | Any character | `a.b` matches `acb` |
| `+` | One or more | `\d+` matches one or more digits |
| `*` | Zero or more | `\d*` matches zero or more digits |
| `{n}` | Exactly n times | `\d{10}` matches 10 digits |
| `[abc]` | Any one of a, b, c | `[aeiou]` matches vowels |
| `^` | Start of string | `^Hello` |
| `$` | End of string | `world$` |
