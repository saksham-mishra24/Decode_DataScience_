## 🤔 What is Recursion?

> **Recursion** is when a function **calls itself**.

### Analogy: Mirrors Facing Each Other 🪞🪞
When you put two mirrors facing each other, you see infinite reflections. Each reflection contains another reflection.

A recursive function works similarly — it calls itself with a smaller problem until it reaches a **stopping point** (base case).

### Two Essential Parts:
1. **Base Case** — When to STOP (prevents infinite loop)
2. **Recursive Case** — Function calls itself with a smaller problem

---

```python
# Countdown using recursion
def countdown(n):
    if n <= 0:            # Base case: STOP!
        print("🚀 Launch!")
    else:
        print(n)           # Print current number
        countdown(n - 1)   # Call itself with n-1

countdown(5)
```

```python
# Factorial: n! = n × (n-1) × (n-2) × ... × 1
# 5! = 5 × 4 × 3 × 2 × 1 = 120

def factorial(n):
    if n == 0 or n == 1:    # Base case
        return 1
    else:
        return n * factorial(n - 1)   # Recursive case

print("5! =", factorial(5))    # 120
print("7! =", factorial(7))    # 5040
print("10! =", factorial(10))  # 3628800
```

```python
# Fibonacci: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...
# Each number = sum of previous two

def fibonacci(n):
    if n <= 0: return 0       # Base case 1
    elif n == 1: return 1     # Base case 2
    else:
        return fibonacci(n-1) + fibonacci(n-2)  # Recursive

print("First 10 Fibonacci numbers:")
for i in range(10):
    print(fibonacci(i), end=" ")
```

```python
# Sum of digits: 1234 → 1+2+3+4 = 10
def sum_digits(n):
    if n < 10:              # Base case: single digit
        return n
    else:
        return (n % 10) + sum_digits(n // 10)  # Last digit + rest

print("Sum of digits of 1234:", sum_digits(1234))   # 10
print("Sum of digits of 9876:", sum_digits(9876))   # 30
```

---

## ⚠️ Important Notes
- Always have a **base case** or you get infinite recursion!
- Python has a limit of ~1000 recursive calls
- Loops are often faster than recursion
- Use recursion when the problem naturally breaks into smaller sub-problems
