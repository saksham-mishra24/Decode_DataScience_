## 🤔 What is Return?

`return` sends a value **back** to the code that called the function.

### Analogy: ATM Machine 🏧
- You **give** your card and PIN (input)
- The machine **returns** money to you (output/return)

---

```python
# Function that RETURNS a value
def add(a, b):
    result = a + b
    return result   # Send the answer back

# Store the returned value
answer = add(5, 3)
print("5 + 3 =", answer)  # 8

# Use directly in expressions
print("10 + 20 =", add(10, 20))
total = add(100, add(50, 25))  # Nested calls!
print("100 + (50 + 25) =", total)
```

```python
# print() vs return
def add_print(a, b):
    print(a + b)     # Just shows on screen, can't reuse

def add_return(a, b):
    return a + b     # Sends value back, can reuse!

# With print — can't reuse the result
x = add_print(5, 3)    # Prints 8
print("x =", x)         # x = None (nothing returned!)

# With return — can reuse!
y = add_return(5, 3)   # Returns 8
print("y =", y)         # y = 8
print("y * 2 =", y * 2) # Can do math with it!
```

```python
# Returning multiple values
def get_stats(numbers):
    """Return min, max, average of a list"""
    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)
    return minimum, maximum, average   # Return 3 values!

marks = [78, 85, 92, 67, 88]
low, high, avg = get_stats(marks)
print(f"Lowest: {low}")
print(f"Highest: {high}")
print(f"Average: {avg}")
```

```python
# Practical: Grade calculator function
def get_grade(marks):
    """Return grade based on marks"""
    if marks >= 90: return "A+"
    elif marks >= 80: return "A"
    elif marks >= 70: return "B"
    elif marks >= 60: return "C"
    elif marks >= 33: return "D"
    else: return "F"

# Use it!
students = {"Rahul": 85, "Priya": 92, "Amit": 67, "Sunita": 28}
for name, marks in students.items():
    grade = get_grade(marks)
    print(f"{name}: {marks} marks → Grade {grade}")
```
