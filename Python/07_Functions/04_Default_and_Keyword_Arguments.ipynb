# 🎯 Default and Keyword Arguments

---

## 📌 What You Will Learn
- Default parameter values
- Keyword arguments for clarity
- Combining default and required parameters

---

````python
# Default parameters — pre-set values
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}! 🙏")

greet("Rahul")               # Uses default: "Hello, Rahul!"
greet("Priya", "Good morning")  # Custom: "Good morning, Priya!"
````

````python
# Real-life: Electricity bill with default rates
def calculate_bill(units, rate=5, tax_percent=10):
    subtotal = units * rate
    tax = subtotal * tax_percent / 100
    total = subtotal + tax
    return total

print("Bill for 100 units (default rate):", calculate_bill(100))
print("Bill for 100 units (₹8 rate):", calculate_bill(100, rate=8))
print("Bill for 200 units (₹7, 15% tax):", calculate_bill(200, rate=7, tax_percent=15))
````

````python
# Power function with default exponent
def power(base, exponent=2):
    return base ** exponent

print("5 squared:", power(5))       # 25
print("5 cubed:", power(5, 3))      # 125
print("2 to the 10:", power(2, 10)) # 1024
````

---

**Rule:** Default parameters must come AFTER required parameters.

```python
def func(required, optional=default):   # ✅ Correct
def func(optional=default, required):   # ❌ Error!
```
