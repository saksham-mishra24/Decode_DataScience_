## 📌 Tour of Python's Most Useful Built-in Functions

Python comes with many ready-to-use functions. Here are the most important ones:

---

```python
# Input/Output
# print() — display output
# input() — get user input
print("Hello World!")

# Type functions
print("\n--- Type Functions ---")
print("type(42):", type(42))          # <class 'int'>
print("int('25'):", int("25"))        # 25
print("float(10):", float(10))       # 10.0
print("str(100):", str(100))          # '100'
print("bool(0):", bool(0))            # False
```

```python
# Math functions
print("--- Math Functions ---")
print("abs(-7):", abs(-7))              # 7
print("round(3.7):", round(3.7))       # 4
print("round(3.14159, 2):", round(3.14159, 2))  # 3.14
print("pow(2, 10):", pow(2, 10))       # 1024
print("min(5,3,8):", min(5, 3, 8))     # 3
print("max(5,3,8):", max(5, 3, 8))     # 8
print("sum([1,2,3,4,5]):", sum([1,2,3,4,5]))  # 15
```

```python
# Sequence functions
print("--- Sequence Functions ---")
print("len('Hello'):", len("Hello"))       # 5
print("len([1,2,3]):", len([1,2,3]))       # 3
print("sorted([3,1,4,2]):", sorted([3,1,4,2]))   # [1,2,3,4]
print("reversed:", list(reversed([1,2,3])))       # [3,2,1]
print("list(range(5)):", list(range(5)))            # [0,1,2,3,4]

# enumerate() — get index + value
fruits = ["Apple", "Banana", "Mango"]
for i, fruit in enumerate(fruits, 1):
    print(f"  {i}. {fruit}")
```

```python
# zip() — combine two lists
names = ["Rahul", "Priya", "Amit"]
marks = [85, 92, 67]

print("--- zip() ---")
for name, mark in zip(names, marks):
    print(f"{name}: {mark} marks")

# any() and all()
print("\nany([False, False, True]):", any([False, False, True]))  # True
print("all([True, True, True]):", all([True, True, True]))       # True
print("all([True, True, False]):", all([True, True, False]))     # False
```

```python
# String/Character functions
print("--- Character Functions ---")
print("chr(65):", chr(65))      # 'A' (number → character)
print("chr(97):", chr(97))      # 'a'
print("ord('A'):", ord("A"))    # 65 (character → number)
print("ord('a'):", ord("a"))    # 97

# help() — get documentation
# help(print)   # Shows documentation for print function
```

## 📊 Quick Reference Table

| Function | Purpose | Example |
|---|---|---|
| `print()` | Display output | `print("Hi")` |
| `input()` | Get user input | `input("Name: ")` |
| `len()` | Length | `len("Hello")` → 5 |
| `type()` | Check type | `type(42)` → int |
| `int/float/str()` | Convert type | `int("25")` → 25 |
| `abs()` | Absolute value | `abs(-5)` → 5 |
| `round()` | Round number | `round(3.7)` → 4 |
| `min/max()` | Smallest/Largest | `max(5,3,8)` → 8 |
| `sum()` | Sum of list | `sum([1,2,3])` → 6 |
| `sorted()` | Sort | `sorted([3,1,2])` → [1,2,3] |
| `range()` | Number sequence | `range(5)` → 0-4 |
| `enumerate()` | Index + value | Loop with numbers |
| `zip()` | Combine lists | Pair items together |
