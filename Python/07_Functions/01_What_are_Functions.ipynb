## 🤔 What is a Function?

> A **function** is a reusable block of code that performs a specific task.

### Real-Life Analogy: Juice Machine 🧃

A juice machine:
1. Takes **input** (fruits)
2. **Processes** them (blends)
3. Gives **output** (juice)

```
🍎 Fruits (Input)  →  [MACHINE (Function)]  →  🧃 Juice (Output)
```

You don't rebuild the machine every time — you just USE it! That's a function.

### Why Functions?
- **Reusability** — Write once, use many times
- **Organization** — Break big code into small pieces
- **Readability** — Code is easier to understand

## 💻 Creating a Function

```python
def function_name():
    """Description of what this function does"""
    code here
    code here
```

- `def` — keyword that means "define a function"
- `function_name` — name you choose (like variable names)
- `()` — parentheses (for parameters, we'll learn soon)
- `:` — colon at the end
- Indented code — the function's body

````python
# Defining a simple function
def greet():
    """This function prints a greeting message"""
    print("Hello! 👋")
    print("Welcome to Python!")
    print("Have a great day! 🌟")

# Calling (using) the function
greet()          # First call
print("---")
greet()          # Second call — same code runs again!
print("---")
greet()          # Third call — reusable!
````

````python
# Function with a parameter (input)
def greet_person(name):
    """Greet a specific person"""
    print(f"Hello, {name}! Welcome! 🙏")

# Call with different names
greet_person("Rahul")
greet_person("Priya")
greet_person("Amit")
````

````python
# Function with multiple parameters
def student_info(name, age, city):
    """Display student information"""
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
    print("-" * 20)

student_info("Rahul", 20, "Pune")
student_info("Priya", 19, "Mumbai")
````

````python
# Practical function: Multiplication table
def print_table(number):
    """Print multiplication table for given number"""
    print(f"\n=== Table of {number} ===")
    for i in range(1, 11):
        print(f"{number} × {i} = {number * i}")

# Now I can print ANY table easily!
print_table(5)
print_table(7)
````

---

## 📊 Summary

| Concept | Syntax |
|---|---|
| Define function | `def function_name():` |
| Call function | `function_name()` |
| With parameter | `def greet(name):` |
| Call with argument | `greet("Rahul")` |

**Key Rule:** A function must be **defined** before it can be **called**.
