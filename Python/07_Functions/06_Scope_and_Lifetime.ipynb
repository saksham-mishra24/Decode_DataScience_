## 🤔 What is Scope?

**Scope** determines where a variable can be accessed.

### Analogy: Your Room vs Your House 🏠
- Things in **your room** (local) — only you can access them
- Things in the **living room** (global) — everyone in the house can access them

---

```python
# Local variable — exists only INSIDE the function
def my_function():
    x = 10   # Local variable
    print("Inside function: x =", x)

my_function()
# print(x)   # ❌ Error! x doesn't exist outside the function
```

```python
# Global variable — exists EVERYWHERE
y = 100   # Global variable

def my_function():
    print("Inside function: y =", y)   # Can READ global

my_function()
print("Outside function: y =", y)     # Can access here too
```

```python
# The global keyword — modify global variable inside function
count = 0   # Global

def increment():
    global count    # Tell Python we want the GLOBAL count
    count += 1

print("Before:", count)   # 0
increment()
increment()
increment()
print("After 3 calls:", count)  # 3
```

```python
# Same name, different scope
x = "Global"

def test():
    x = "Local"   # This is a DIFFERENT x (local)
    print("Inside:", x)   # Local

test()
print("Outside:", x)    # Global (unchanged!)
```

---

## 📊 Summary

| Scope | Where Defined | Where Accessible |
|---|---|---|
| Local | Inside function | Only inside that function |
| Global | Outside all functions | Everywhere |
| `global` keyword | Inside function | Makes local → global |

**Tip:** Avoid using `global` too much. It's better to use `return` to pass data.
