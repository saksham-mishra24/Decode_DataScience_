```python
import cv2

# Load file
image = cv2.imread(r"C:\Users\Edunet Foundation\Desktop\github _upload\color_image.jpg")

# shape[height: width : channels] -> [h:w:c] => image. shape =
# rows - heights of the image
# column - widht of the image

rows, cols = image.shape[:2]


# rotation matrix
M = cv2.getRotationMatrix2D((cols / 2, rows / 2), 45, 1)

# getRotationMatrix2D(center, angle, scale) is a function in OpenCV that generates a 2D rotation matrix. 
# The parameters are:
# cols / 2: The x-coordinate of the center of rotation (in this case, the center of the image).
# rows / 2: The y-coordinate of the center of rotation (in this case, the center of the image).
# 45: The angle of rotation in degrees (positive values rotate counterclockwise).
# 1: The scale factor (1 means no scaling, values greater than 1 will enlarge the image, and 
# values less than 1 will shrink the image).


# apply the rotation to the image
rotated_image = cv2.warpAffine(image, M, (cols, rows))

# warpAffine(src, M, dsize) is a function in OpenCV that applies an affine transformation to an image.
# The parameters are:
# src: The source image to be transformed.
# M: The transformation matrix (in this case, the rotation matrix generated by getRotationMatrix2D).
# dsize: The size of the output image (in this case, the same size as the original image).
# Display the original and rotated images
cv2.imshow('Original Image', image)
```

```python
# Shi–Tomasi Corner Detection
 
import cv2
import matplotlib.pyplot as plt 
import numpy as np
 
# Read image 
img = cv2.imread(r"C:\Users\Edunet Foundation\Desktop\github _upload\color_image.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # RGB  
 
# convert to GrayScal-image 
gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)  # RGB - GRAY  
 
# Corner detection 
corners = cv2.goodFeaturesToTrack(gray,
                                  maxCorners=100,
                                  qualityLevel=0.01,
                                  minDistance=10) 
# convert to integer 
corners = np.int32(corners)
 
# drawing corners 
for corner in corners:
    x,y = corner.ravel() 
    cv2.circle(img_rgb, (x,y), 5, (255,0,0, -1))
 
plt.imshow(img_rgb)
plt.title("Shi-Tomasi Corners")
plt.axis('off')
plt.show() 
```

```python
# Image Alignment and Feature Matching  
 
import cv2
import numpy as np
import matplotlib.pyplot as plt
 
img1 = cv2.imread(r'C:\Users\Edunet Foundation\Desktop\github _upload\download (1).jpeg')
img2 = cv2.imread(r'C:\Users\Edunet Foundation\Desktop\github _upload\download.jpeg')
 
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
 
# Detect ORB Features  - ORB (Oriented FAST and Rotated BRIEF) 
orb = cv2.ORB_create(500) 
kp1, des1 = orb.detectAndCompute(gray1, None) 
kp2, des2 = orb.detectAndCompute(gray2, None)
 
# Match Features 
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True) 
matches = bf.match(des1, des2)
 
matches = sorted(matches, key = lambda x : x.distance)
 
# draw matches 
matched_img = cv2.drawMatches(img1, kp1, img2, kp2, matches[:50], None) 
plt.imshow(cv2.cvtColor(matched_img, cv2.COLOR_BGR2RGB)) 
plt.title("Feature Matching")
plt.axis('off')
plt.show()
```

```python
# Image Alignment (Homography)
 
# Extract matched points 
pts1 = np.float32([kp1[m.queryIdx].pt for m in matches]).reshape(-1,1,2)
pts2 = np.float32([kp2[m.queryIdx].pt for m in matches]).reshape(-1,1,2)
 
# compute homography 
H, mask = cv2.findHomography(pts2, pts1, cv2.RANSAC)
 
# warp image 
aligned = cv2.warpPerspective(img2, H, (img1.shape[1],img1.shape[0]))
 
plt.imshow(cv2.cvtColor(aligned, cv2.COLOR_BGR2RGB))
plt.title("Aligned Image")
plt.axis('off')
plt.show()
```

```python
# Image Blending
 
import cv2
import numpy as np
import matplotlib.pyplot as plt 
blend = cv2.addWeighted(img1, 0.5, aligned, 0.5, 0)
plt.imshow(cv2.cvtColor(blend, cv2.COLOR_BGR2RGB))
plt.title("Simple Blending")
plt.axis('off')
plt.show() 
```

```python
import cv2
import numpy as np
 
cap = cv2.VideoCapture('video.mp4')
ret, frame = cap.read()
 
r, h, c, w = 200, 100, 300, 200
track_window = (c, r, w, h)
 
roi = frame[r:r+h, c:c+w]
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
 
mask = cv2.inRange(hsv_roi, (0, 60, 32), (180, 255, 255))
roi_hist = cv2.calcHist([hsv_roi], [0], mask, [180], [0, 180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)
 
term_crit = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 10, 1)
 
while True:
    ret, frame = cap.read()
    if not ret:
        break
 
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    dst = cv2.calcBackProject([hsv], [0], roi_hist, [0, 180], 1)
 
    ret, track_window = cv2.CamShift(dst, track_window, term_crit)
 
    pts = cv2.boxPoints(ret)
    pts = pts.astype(int)
 
    cv2.polylines(frame, [pts], True, (0,255,0), 2)
 
    cv2.imshow("CamShift", frame)
 
    if cv2.waitKey(30) & 0xFF == ord('q'):
        break
 
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
 
# Load Haar Cascade for face detection
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
 
# Start video capture
cap = cv2.VideoCapture(0)
 
# Read first frame
ret, frame = cap.read()
x, y, w, h = 200, 150, 100, 100  # Initial tracking window
track_window = (x, y, w, h)
 
# Setup ROI for tracking
roi = frame[y:y+h, x:x+w]
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
roi_hist = cv2.calcHist([hsv_roi], [0], None, [180], [0, 180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)
 
# Termination criteria
term_crit = (cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 10, 1)
 
while True:
    ret, frame = cap.read()
    if not ret:
        break
 
    # Face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
 
    for (fx, fy, fw, fh) in faces:
        cv2.rectangle(frame, (fx, fy), (fx+fw, fy+fh), (255, 0, 0), 2)
 
    # Object tracking using CamShift
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    dst = cv2.calcBackProject([hsv], [0], roi_hist, [0, 180], 1)
 
    ret, track_window = cv2.CamShift(dst, track_window, term_crit)
    pts = cv2.boxPoints(ret)
    pts = np.int(pts)
    cv2.polylines(frame, [pts], True, (0, 255, 0), 2)
 
    cv2.imshow("Face Detection & Object Tracking", frame)
 
    if cv2.waitKey(1) & 0xFF == 27:
        break
 
cap.release()
cv2.destroyAllWindows()
```
