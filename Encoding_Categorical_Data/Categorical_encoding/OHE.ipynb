# 3. OneHotEncoding using Sklearn

#### when we do one hot encoding using pandas it will not be good because it does not remeber the order of the columns and
####  it will create new columns if we have new categories in the test data which were not present in the training data. 
#### So we will use sklearn for one hot encoding.

```python
x = df.iloc[:, 0:4]
y = df.iloc[:, -1]
```
